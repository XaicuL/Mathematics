# 복소수

```python
import sympy as sp


sp.init_printing()

z1 = 3 + 4*sp.I
z2 = 1 - 2*sp.I

print('(3 + 4i) + (1 - 2i) = ')
display(z1 + z2)


print('(3 + 4i) - (1 - 2i) = ')
display(z1 - z2)
```

복소수를 나타내는 방법으로는 <br><br>
**n * sympy.I 를 사용해야한다( 단 n = 1,2,3,,,,)**

<br>

다음으로 **복소수가 서로 같을 조건을 보자.**

```python
import sympy as sp

sp.init_printing()

a = sp.symbols('a' ,real = True)
b = sp.symbols('b' ,real = True)

z1 = a + 4*sp.I
z2 = -1 + b*sp.I

eq = sp.Eq(z1, z2)
sol_eq = sp.solve(eq)

print(f'{z1} = {z2}')
display(sol_eq)
```

**sp.symbol('name', real = True)** 에서 **real = True**의 의미 <br>
-> 해당 **name** 이라는 변수는 **실수** 이다.

<br>

**sp.Eq** <br>
방정식에서의 **=(등호)** 역할이라고 보면 된다.
<br> 그러나 **Eq** 함수까지만 쓰면 단지 식을 세팅해둔 것이니, 방정식을 계산해내기 위해 **solve** 함수를 사용해야한다.

```python
#복소수의 덧셈과 뺄셈

import sympy as sp

sp.init_printing()

a = sp.symbols('a' ,real = True)
b = sp.symbols('b' ,real = True)
c = sp.symbols('c' ,real = True)
d = sp.symbols('d', real = True)

z1 = (a + b*sp.I) + (c + d*sp.I)
z2 = (a + b*sp.I) - (c + d*sp.I)

z1_a = (a+c) + (b+d)*sp.I
z2_a = (a-c) + (b-d)*sp.I

display(z1)
print('=')
display(z1_a)

print('################################################')

display(z2)
print('=')
display(z2_a)
```

```python
# 복소수의 곱셈

import sympy as sp

sp.init_printing()

a = sp.symbols('a' ,real = True)
b = sp.symbols('b' ,real = True)
c = sp.symbols('c' ,real = True)
d = sp.symbols('d', real = True)

z = (a + b*sp.I) * (c + d*sp.I)
zCal = (a*c - b*d) + (a*d + b*c)*sp.I

display(z)
print('=')
display(zCal)
```

```python
# 복소수의 나눗셈

import sympy as sp

sp.init_printing()

a = sp.symbols('a' ,real = True)
b = sp.symbols('b' ,real = True)
c = sp.symbols('c' ,real = True)
d = sp.symbols('d', real = True)

exP = c + d*sp.I
exC = a + b*sp.I

ex_pa = c**2 + d**2
ex_ca = a*c + b*d

ex_pb = c**2 + d**2
ex_cb = b*c - a*d

ex1 = exC/exP
ex2 = ex_ca/ex_pa
ex3 = ex_cb/ex_pb

display(ex1)
print('=')

display(ex2 + ex3)

```
