# 1. 평균과 표준편차

자료 $x_1, \ldots, x_n$에 대해:

평균: $\bar{x} = \dfrac{1}{n}\sum x_i$

분산: $s^2 = \dfrac{1}{n}\sum (x_i - \bar{x})^2$ (모집단) / $n-1$ (표본)

표준편차: $s = \sqrt{s^2}$

## 1. 계산

```python
import numpy as np
import sympy as sp

data = np.array([72, 85, 90, 68, 95, 88, 76])
mean = np.mean(data)
std_pop = np.std(data, ddof=0)
std_sample = np.std(data, ddof=1)

print(f"자료: {data}")
print(f"평균: {mean:.2f}")
print(f"표준편차 (모): {std_pop:.2f}")
print(f"표준편차 (표본, ddof=1): {std_sample:.2f}")
```

**np.mean(x)**: <br>
=> 산술평균

**np.std(x, ddof=0/1)**: <br>
=> 표준편차 (0=모집단, 1=표본)

## 2. 시각화

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(range(len(data)), data, color='steelblue', alpha=0.7)
ax.axhline(mean, color='red', ls='--', label=f'mean={mean:.1f}')
ax.axhline(mean + std_sample, color='orange', ls=':', label=f'±1σ')
ax.axhline(mean - std_sample, color='orange', ls=':')
ax.set_xlabel('student')
ax.set_ylabel('score')
ax.legend()
plt.tight_layout()
plt.show()
```
