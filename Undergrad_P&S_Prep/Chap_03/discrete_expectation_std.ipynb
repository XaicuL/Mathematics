# 3. 이산확률변수의 기댓값(평균)과 표준편차

기댓값: $E(X) = \sum x \cdot P(X=x)$

분산: $V(X) = E(X^2) - [E(X)]^2$

표준편차: $\sigma(X) = \sqrt{V(X)}$

## 1. 기댓값과 분산

```python
import numpy as np
import sympy as sp
from scipy.stats import rv_discrete

sp.init_printing()

# X: 0,1,2 with P = 1/4, 1/2, 1/4
xk = np.array([0, 1, 2])
pk = np.array([0.25, 0.5, 0.25])
dist = rv_discrete(values=(xk, pk))

EX = dist.mean()
VX = dist.var()
SX = dist.std()
print(f"E(X) = {EX}")
print(f"V(X) = {VX}")
print(f"σ(X) = {SX:.4f}")

# sympy 검증
E_sym = sum(x * p for x, p in zip(xk, pk))
print(f"sympy E(X) = {E_sym}")
```

**dist.mean() / .var() / .std()**: <br>
=> scipy 분포 객체의 모멘트

**수식**: <br>
=> $E(X)=\sum x_i p_i$
