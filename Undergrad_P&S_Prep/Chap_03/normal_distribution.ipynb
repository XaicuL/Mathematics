# 2. 정규분포

$$X \sim N(\mu, \sigma^2): \quad f(x) = \dfrac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\dfrac{(x-\mu)^2}{2\sigma^2}\right)$$

## 1. 정규분포

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
import sympy as sp

mu, sigma = 0, 1
x = sp.symbols('x')
f = sp.exp(-x**2 / 2) / sp.sqrt(2 * sp.pi)
display(sp.Eq(sp.Symbol('f(x)'), f))

P = norm.cdf(1, loc=mu, scale=sigma) - norm.cdf(-1, loc=mu, scale=sigma)
print(f"P(-1≤Z≤1) ≈ {P:.4f}")
```

**scipy.stats.norm**: <br>
=> `norm.pdf(x, loc=μ, scale=σ)`, `norm.cdf(x)`

**표준정규**: <br>
=> $N(0,1)$, `norm(0,1)`

## 2. 종 모양 곡선

```python
xs = np.linspace(-4, 4, 300)
fig, ax = plt.subplots(figsize=(8, 4))
for mu, sigma, label in [(0, 1, 'N(0,1)'), (0, 2, 'N(0,4)'), (2, 1, 'N(2,1)')]:
    ax.plot(xs, norm.pdf(xs, mu, sigma), lw=2, label=label)
ax.set_xlabel('x')
ax.set_ylabel('f(x)')
ax.legend()
ax.set_title('normal distribution')
plt.tight_layout()
plt.show()
```
