# 1. 모집단과 표본

**모집단**: 관심 대상 전체

**표본**: 모집단에서 추출한 일부

**표본평균** $\bar{X}$: 표본의 평균 (확률변수)

**모평균** $\mu$: 모집단의 평균 (상수)

## 1. 모집단 vs 표본

```python
import numpy as np

np.random.seed(42)
population = np.random.normal(100, 15, size=10_000)  # 모집단
mu = population.mean()
sigma = population.std(ddof=0)

sample = np.random.choice(population, size=30, replace=False)
x_bar = sample.mean()
s = sample.std(ddof=1)

print(f"모평균 μ ≈ {mu:.2f}, 모표준편차 σ ≈ {sigma:.2f}")
print(f"표본평균 x̄ = {x_bar:.2f}, 표본표준편차 s = {s:.2f}")
```

**np.random.choice**: <br>
=> 모집단에서 표본 추출

**ddof=1**: <br>
=> 표본 표준편차 (불편추정)
