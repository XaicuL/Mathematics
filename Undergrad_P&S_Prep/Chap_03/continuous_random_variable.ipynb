# 1. 연속확률변수

확률밀도함수 $f(x)$:
1. $f(x) \ge 0$
2. $\int_{-\infty}^{\infty} f(x)\,dx = 1$

$$P(a \le X \le b) = \int_a^b f(x)\,dx$$

## 1. 균등분포 U(a,b)

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import uniform
import sympy as sp

sp.init_printing()
a, b = 0, 4
x = sp.symbols('x')
f = sp.Piecewise((0, x < a), (sp.Rational(1, b - a), (x >= a) & (x <= b)), (0, True))
display(sp.Eq(sp.Symbol('f(x)'), f))

P_1_3 = sp.integrate(f, (x, 1, 3))
print(f"P(1≤X≤3) = {P_1_3}")
```

**scipy.stats.uniform**: <br>
=> `uniform(loc=a, scale=b-a)`

**sp.integrate**: <br>
=> PDF 적분으로 확률 계산

## 2. PDF 그래프

```python
xs = np.linspace(-1, 5, 200)
pdf = uniform.pdf(xs, loc=a, scale=b - a)

plt.figure(figsize=(7, 4))
plt.plot(xs, pdf, 'b-', lw=2)
plt.fill_between(xs, pdf, where=(xs >= 1) & (xs <= 3), alpha=0.3, color='orange', label='P(1≤X≤3)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title(f'uniform distribution U({a},{b})')
plt.legend()
plt.tight_layout()
plt.show()
```
