# 2. 확률변수와 확률분포

**확률변수 $X$**: 시행 결과를 실수로 대응

**확률질량함수(PMF)**: $P(X=x)$ (이산)

**확률밀도함수(PDF)**: $f(x)$, $P(a \le X \le b) = \int_a^b f(x)\,dx$ (연속)

## 1. 이산 확률분포

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import rv_discrete

# 주사위 X ~ {1,...,6}, 각 1/6
xk = np.arange(1, 7)
pk = np.ones(6) / 6
dist = rv_discrete(name='dice', values=(xk, pk))

print("PMF:")
for x in xk:
    print(f"  P(X={x}) = {dist.pmf(x):.4f}")
print(f"합: {dist.pmf(xk).sum():.4f}")
```

**scipy.stats.rv_discrete**: <br>
=> 사용자 정의 이산분포

**`.pmf(x)`**: <br>
=> $P(X=x)$

## 2. PMF 막대그래프

```python
plt.figure(figsize=(6, 4))
plt.bar(xk, pk, width=0.6, color='steelblue', edgecolor='black')
plt.xlabel('x')
plt.ylabel('P(X=x)')
plt.title('PMF')
plt.xticks(xk)
plt.tight_layout()
plt.show()
```
