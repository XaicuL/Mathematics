# 3. 이항분포와 정규분포

De Moivre–Laplace 근사: $n$이 크면
$$B(n,p) \approx N(np, \, np(1-p))$$

## 1. 정규근사

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, norm
import sympy as sp

n, p = 50, 0.4
mu = n * p
sigma = np.sqrt(n * p * (1 - p))
k = 20
P_binom = binom.pmf(k, n, p)
P_norm = norm.pdf(k, mu, sigma)  # 연속 근사
print(f"B({n},{p}): P(X={k}) = {P_binom:.4f}")
print(f"N({mu:.1f}, {sigma**2:.1f}) PDF at {k}: {P_norm:.4f}")
```

**정규근사**: <br>
=> $n$ 크고 $np, n(1-p)$ 모두 ≥ 5 정도일 때 사용

**연속 보정**: <br>
=> $P(X=k) \approx P(k-0.5 < Y < k+0.5)$

## 2. 겹쳐 그리기

```python
ks = np.arange(0, n + 1)
pmf = binom.pmf(ks, n, p)
xs = np.linspace(0, n, 300)
pdf = norm.pdf(xs, mu, sigma)

plt.figure(figsize=(9, 4))
plt.bar(ks, pmf, alpha=0.6, label=f'B({n},{p})', color='steelblue')
plt.plot(xs, pdf, 'r-', lw=2, label=f'N({mu:.0f}, {sigma**2:.1f})')
plt.xlabel('k')
plt.ylabel('probability / density')
plt.legend()
plt.title('binomial and normal approximation')
plt.tight_layout()
plt.show()
```
