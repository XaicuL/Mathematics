# 2. 모집단과 표본평균의 분포

모집단 $N(\mu, \sigma^2)$에서 크기 $n$ 표본의 $\bar{X}$:

$$\bar{X} \sim N\left(\mu, \dfrac{\sigma^2}{n}\right) \quad (n \text{ 충분히 크면, CLT})$$

## 1. 표본평균의 분포 (CLT)

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

np.random.seed(0)
mu, sigma = 50, 10
n = 30
n_samples = 2000

sample_means = [np.random.normal(mu, sigma, n).mean() for _ in range(n_samples)]
se = sigma / np.sqrt(n)

print(f"이론: E(X̄)={mu}, SE={se:.3f}")
print(f"시뮬: mean={np.mean(sample_means):.3f}, std={np.std(sample_means):.3f}")
```

**표준오차 SE**: <br>
=> $\sigma / \sqrt{n}$

**CLT (중심극한정리)**: <br>
=> $n$↑ 이면 $\bar{X}$가 정규에 가까워짐

## 2. 표본평균 히스토그램

```python
plt.figure(figsize=(8, 4))
plt.hist(sample_means, bins=40, density=True, alpha=0.7, color='steelblue', label='sample mean')
xs = np.linspace(min(sample_means), max(sample_means), 200)
plt.plot(xs, norm.pdf(xs, mu, se), 'r-', lw=2, label=f'N({mu}, {se**2:.2f})')
plt.xlabel('sample mean')
plt.ylabel('density')
plt.legend()
plt.title('CLT')
plt.tight_layout()
plt.show()
```
