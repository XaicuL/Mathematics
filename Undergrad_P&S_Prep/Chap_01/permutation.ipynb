# 1. 순열

서로 다른 $n$개에서 $r$개를 **순서를 구분**하여 나열하는 방법의 수.

$${}_nP_r = \dfrac{n!}{(n-r)!} = n(n-1)\cdots(n-r+1)$$

## 1. 순열 공식

```python
import sympy as sp
sp.init_printing()

n, r = sp.symbols('n r', integer=True, nonnegative=True)
nPr = sp.factorial(n) / sp.factorial(n - r)
display(sp.Eq(sp.Symbol('_nP_r'), nPr))

# 예: 5명 중 3명 줄 세우기
result = sp.factorial(5) // sp.factorial(5 - 3)
print(f"5P3 = {result}")
```

**sp.factorial(n)**: <br>
=> $n!$ 계산

**sp.binomial(n, r)**: <br>
=> 조합 ${}_nC_r$ (순열과 구분)

## 2. Example

```python
from itertools import permutations

items = ['A', 'B', 'C']
r = 2
perms = list(permutations(items, r))
print(f"3P2 = {len(perms)}")
print("나열:", perms)
```
