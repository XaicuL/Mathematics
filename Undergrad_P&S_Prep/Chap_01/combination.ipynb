# 1. 조합

서로 다른 $n$개에서 $r$개를 **순서 없이** 선택하는 방법.

$${}_nC_r = \binom{n}{r} = \dfrac{n!}{r!(n-r)!}$$

**대칭**: ${}_nC_r = {}_nC_{n-r}$

## 1. 조합 공식

```python
import sympy as sp
sp.init_printing()

n, r = 10, 3
ncr = sp.binomial(n, r)
print(f"{n}C{r} = {ncr}")

# 대칭성
print(f"{n}C{r} = {n}C{n-r}:", sp.binomial(n, r) == sp.binomial(n, n - r))
```

**sp.binomial(n, r)**: <br>
=> ${}_nC_r = \binom{n}{r}$ 계산

**itertools.combinations**: <br>
=> 조합 나열

## 2. Example

```python
from itertools import combinations

items = list('ABCDE')
r = 2
combs = list(combinations(items, r))
print(f"5C2 = {len(combs)}")
print(combs)
```
