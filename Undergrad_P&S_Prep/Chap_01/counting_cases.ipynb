# 1. 경우의 수

어떤 일이 일어날 수 있는 **모든 경우**를 세는 방법이다.

**합의 법칙**: 두 일이 동시에 일어날 수 없을 때, 전체 경우의 수는 각 경우의 수를 **더**한다.

$$|A \cup B| = |A| + |B| \quad (A \cap B = \varnothing)$$

**곱의 법칙**: 연속된 선택에서 전체 경우의 수는 각 단계 경우의 수를 **곱**한다.

$$|A \times B| = |A| \cdot |B|$$

## 1. 합의 법칙

```python
import sympy as sp
sp.init_printing()

# 예: 1~5 중 짝수 또는 3의 배수
evens = {2, 4}
mult3 = {3}
union = evens | mult3
print(f"짝수: {len(evens)}, 3의 배수: {len(mult3)}, 합집합: {len(union)}")
print("경우:", sorted(union))
```

**집합 연산**: <br>
=> `|` (합집합), `&` (교집합)으로 겹치는 경우를 확인한다.

## 2. 곱의 법칙

```python
# 예: 3자리 PIN (0~9, 중복 허용)
digits = 10
length = 3
total = digits ** length
print(f"{length}자리 PIN: {digits}^{length} = {total}")

# 예: 메뉴 A 4종 × 음료 B 3종
menu, drink = 4, 3
print(f"주문 조합: {menu} × {drink} = {menu * drink}")
```

**곱의 벡칙**: <br>
=> 독립적인 단계가 있을 때 `**` 또는 `math.prod`로 계산한다.

## 3. 합·곱 혼합 Example

```python
# 2명의 학생에게 서로 다른 책 5권 중 2권 배부 (순서 있음)
n, r = 5, 2
permutation = sp.factorial(n) // sp.factorial(n - r)
print(f"5P2 = {permutation}")

# 경로: A→B 3가지, B→C 2가지
paths = 3 * 2
print(f"A→B→C 경로 수: {paths}")
```
