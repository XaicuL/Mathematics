# 2. 이항계수의 성질

1. $\binom{n}{0} = \binom{n}{n} = 1$
2. $\binom{n}{k} = \binom{n}{n-k}$ (대칭)
3. $\binom{n}{k} = \binom{n-1}{k-1} + \binom{n-1}{k}$ (파스칼)
4. $\sum_{k=0}^{n} \binom{n}{k} = 2^n$

## 1. 파스칼의 삼각형

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

sp.init_printing()
n_rows = 8

for n in range(n_rows):
    row = [sp.binomial(n, k) for k in range(n + 1)]
    print(" " * (n_rows - n) + " ".join(f"{int(c):3d}" for c in row))

# 파스칼 관계 검증
n, k = 6, 3
lhs = sp.binomial(n, k)
rhs = sp.binomial(n - 1, k - 1) + sp.binomial(n - 1, k)
print(f"\nC({n},{k}) = C({n-1},{k-1}) + C({n-1},{k}): {lhs == rhs}")
```

**sp.binomial(n, k)**: <br>
=> 이항계수; 파스칼 관계로 재귀 정의 가능

**합 성질**: <br>
=> `sum(sp.binomial(n,k) for k in range(n+1)) == 2**n`

## 2. 파스칼 삼각형 시각화

```python
n_rows = 10
fig, ax = plt.subplots(figsize=(8, 5))
for n in range(n_rows):
    for k in range(n + 1):
        val = int(sp.binomial(n, k))
        ax.text(k - n / 2, -n, str(val), ha='center', va='center', fontsize=9)
ax.set_xlim(-n_rows / 2 - 1, n_rows / 2 + 1)
ax.set_ylim(-n_rows - 0.5, 0.5)
ax.axis('off')
ax.set_title('triangle of Pascal')
plt.tight_layout()
plt.show()
```
