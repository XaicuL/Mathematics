# 4. 원순열

원형 테이블에 $n$명을 **회전만 구분**하여 배치하는 방법.

$$(n-1)! \quad (\text{한 자리 고정 후 나머지 순열})$$

## 1. 원순열 공식

```python
import sympy as sp
sp.init_printing()

n = 5
linear = sp.factorial(n)
circular = sp.factorial(n - 1)
print(f"{n}명 일렬: {linear}")
print(f"{n}명 원형: ({n}-1)! = {circular}")
print(f"비율: {linear / circular} (= n)")
```

**원순열**: <br>
=> 회전 대칭: $n$가지 중복 제거 → $(n-1)!$

**거울 대칭까지 같으면**: <br>
=> $(n-1)!/2$ (반사 대칭 포함)

## 2. Example

```python
people = ['A', 'B', 'C', 'D']
n = len(people)
print(f"{n}명 원순열: {sp.factorial(n - 1)}")

# 4명 중 2명만 원형 배치
r = 2
partial = sp.factorial(n - 1) // sp.factorial(n - r - 1)
print(f"4명 중 2명 원순열: {partial}")
```
