# 3. 같은 것이 있는 순열

일부 원소가 **구별 불가**할 때의 순열.

$$\dfrac{n!}{a!\,b!\,\cdots} \quad (\text{같은 것 } a\text{개, } b\text{개, } \ldots)$$

## 1. MISSISSIPPI Example

```python
import sympy as sp
from collections import Counter

word = 'MISSISSIPPI'
n = len(word)
counts = Counter(word)
denom = sp.prod(sp.factorial(v) for v in counts.values())
result = sp.factorial(n) // denom

print(f"'{word}' 순열 수:")
print(f"{n}! / (", " × ".join(f"{v}!" for v in counts.values()), f") = {result}")
```

**Counter**: <br>
=> 문자열·리스트 원소 개수 집계

**sp.prod(...)**: <br>
=> 팩토리얼 곱 계산

## 2. 일반 공식

```python
n, a, b = sp.symbols('n a b', integer=True, positive=True)
formula = sp.factorial(n) / (sp.factorial(a) * sp.factorial(b))
display(sp.Eq(sp.Symbol('P'), formula))

# AABB (4글자, A 2개, B 2개)
print("AABB:", sp.factorial(4) // (sp.factorial(2) * sp.factorial(2)))
```
