# 2. 중복순열

같은 것을 **여러 번 사용**할 수 있는 순열.

$$n\text{개에서 } r\text{개를 중복 허용하여 나열: } n^r$$

## 1. 공식과 Example

```python
import sympy as sp
sp.init_printing()

n, r = 4, 3  # 4가지 숫자로 3자리 수
count = n ** r
print(f"{n}^r = {n}^{r} = {count}")

# 비밀번호: 0~9 중 4자리 (중복 허용)
digits, length = 10, 4
print(f"PIN 경우의 수: {digits}^{length} = {digits ** length}")
```

**중복순열**: <br>
=> 각 자리에 $n$가지 선택 → $n^r$

**itertools.product**: <br>
=> 중복순열 나열에 유용 (`repeat=r`)

## 2. 나열 확인

```python
from itertools import product

items = ['R', 'B']
r = 3
arrangements = list(product(items, repeat=r))
print(f"2^3 = {len(arrangements)}")
print(arrangements[:8], "...")
```
