# 2. 중복조합

같은 것을 **여러 번 선택**할 수 있는 조합.

$$H(n, r) = {}_{n+r-1}C_r = \binom{n+r-1}{r}$$

## 1. 중복조합 공식

```python
import sympy as sp
sp.init_printing()

n, r = 4, 3  # 4종류 중 3개 선택 (중복 허용)
H = sp.binomial(n + r - 1, r)
print(f"H({n}, {r}) = C({n+r-1}, {r}) = {H}")

# 예: 과일 3종 중 5개 담기
n, r = 3, 5
print(f"과일 상자: H(3, 5) = {sp.binomial(n + r - 1, r)}")
```

**중복조합 H(n,r)**: <br>
=> $n$종류에서 $r$개 선택 (복원)

**별칭**: ${}_{n}H_r = {}_{n+r-1}C_r$

## 2. Example

```python
from itertools import combinations_with_replacement

items = ['A', 'B', 'C']
r = 2
combs = list(combinations_with_replacement(items, r))
print(f"H(3, 2) = {len(combs)}")
print(combs)
```
