# 1. 이항정리

$$(a+b)^n = \sum_{k=0}^{n} \binom{n}{k} a^{n-k} b^k$$

## 1. 전개

```python
import sympy as sp
sp.init_printing()

a, b, n = sp.symbols('a b n', integer=True, nonnegative=True)
expr = (a + b) ** n
expanded = sp.expand(expr)
display(sp.Eq((a + b) ** n, expanded))

# n=4 구체적 전개
display(sp.expand((a + b) ** 4))
```

**sp.expand((a+b)**n)**: <br>
=> 이항정리로 전개

**sp.binomial(n, k)**: <br>
=> 각 항의 계수 $\binom{n}{k}$

## 2. 계수 확인

```python
n = 5
for k in range(n + 1):
    coeff = sp.binomial(n, k)
    print(f"a^{n-k} b^{k} 계수: C({n},{k}) = {coeff}")
```
