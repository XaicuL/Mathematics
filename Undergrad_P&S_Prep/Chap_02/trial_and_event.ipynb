# 1. 시행과 사건

**시행(trial)**: 결과가 불확실한 실험 (동전 던지기, 주사위 등)

**표본공간 $S$**: 모든 가능한 결과의 집합

**사건(event)**: $S$의 부분집합

| 용어 | 예 (주사위) |
|------|-------------|
| 표본공간 | $S = \{1,2,3,4,5,6\}$ |
| 사건 A (짝수) | $A = \{2,4,6\}$ |

## 1. 표본공간과 사건

```python
import sympy as sp
from itertools import product

sp.init_printing()

# 주사위 2개
S = list(product(range(1, 7), repeat=2))
print(f"표본공간 크기: |S| = {len(S)}")

A = {(a, b) for a, b in S if a + b == 7}  # 합이 7
B = {(a, b) for a, b in S if a == b}      # 더블
print(f"A (합=7): {sorted(A)}")
print(f"B (더블): {sorted(B)}")
print(f"A ∩ B: {A & B}")
```

**itertools.product**: <br>
=> 복수 시행의 표본공간 생성

**집합 연산**: <br>
=> `|`, `&`, `-`로 합·교·차 사건 표현
