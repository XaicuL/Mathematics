# 1. 확률의 덧셈정리

$$P(A \cup B) = P(A) + P(B) - P(A \cap B)$$

$A, B$가 **서로소** ($A \cap B = \varnothing$)이면:
$$P(A \cup B) = P(A) + P(B)$$

## 1. 덧셈정리

```python
import sympy as sp
sp.init_printing()

PA, PB, PAB = sp.Rational(1, 3), sp.Rational(1, 4), sp.Rational(1, 12)
P_union = PA + PB - PAB
print(f"P(A) = {PA}, P(B) = {PB}, P(A∩B) = {PAB}")
print(f"P(A∪B) = {PA} + {PB} - {PAB} = {P_union}")
```

**덧셈정리**: <br>
=> 교집합을 한 번 빼서 이중 계산 방지

**서로소**: <br>
=> $P(A \cap B) = 0$이면 $P(A \cup B) = P(A) + P(B)$

## 2. 카드 Example

```python
# 52장: 하트 또는 킹
hearts = 13
kings = 4
heart_king = 1
total = 52
P = sp.Rational(hearts + kings - heart_king, total)
print(f"P(하트 또는 킹) = (13+4-1)/52 = {P}")
```
