# 2. 확률의 정의

**고전적 확률** (모든 결과가 동일할 때):
$$P(A) = \dfrac{|A|}{|S|}$$

**통계적(빈도) 확률**: 반복 시행에서 상대도수의 극한

## 1. 고전적 확률

```python
import sympy as sp
from itertools import product

sp.init_printing()

# 주사위: 3의 배수
S = {1, 2, 3, 4, 5, 6}
A = {3, 6}
P_A = sp.Rational(len(A), len(S))
print(f"P(3의 배수) = {len(A)}/{len(S)} = {P_A}")

# 동전 3번: 앞면 2번
S_coin = list(product(['H', 'T'], repeat=3))
E = [s for s in S_coin if s.count('H') == 2]
P_E = sp.Rational(len(E), len(S_coin))
print(f"P(앞 2번) = {len(E)}/{len(S_coin)} = {P_E}")
```

**sp.Rational(p, q)**: <br>
=> 정확한 유리수 확률 (부동소수 오차 방지)

**확률 공리**: <br>
=> $0 \le P(A) \le 1$, $P(S)=1$, $P(\varnothing)=0$

## 2. 빈도적 확률 (시뮬레이션)

```python
import numpy as np

np.random.seed(42)
trials = 10_000
rolls = np.random.randint(1, 7, size=trials)
freq = np.mean(np.isin(rolls, [3, 6]))
print(f"{trials}회 시뮬: P(3의 배수) ≈ {freq:.4f}")
print(f"이론값: {float(P_A):.4f}")
```
