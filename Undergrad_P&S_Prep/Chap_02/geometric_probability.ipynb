# 3. 기하적 확률

표본공간이 **연속적(길이·넓이)**일 때:
$$P(A) = \dfrac{\text{유리한 영역의 크기}}{\text{전체 영역의 크기}}$$

## 1. 다트판 Example

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Wedge

fig, ax = plt.subplots(figsize=(6, 6))
ax.set_aspect('equal')

# 원판 반지름 1
board = Circle((0, 0), 1, fill=False, color='black', lw=2)
ax.add_patch(board)

# 중심 원 (r=0.5) — 3배 점수
inner = Circle((0, 0), 0.5, fill=True, alpha=0.3, color='red', label='Score 3 Area')
ax.add_patch(inner)

ax.set_xlim(-1.2, 1.2)
ax.set_ylim(-1.2, 1.2)
ax.legend(handles=[inner])
ax.set_title('Geometric Probability')
plt.show()

# P(중심 원 적중) = π(0.5)² / π(1)² = 1/4
P_inner = 0.5 ** 2
print(f"P(중심 적중) = (0.5)² = {P_inner}")
```

**기하적 확률**: <br>
=> 넓이·길이·부피 비율로 계산

**matplotlib.patches**: <br>
=> `Circle`, `Wedge` 등 도형 그리기

## 2. 무작위 점 시뮬레이션

```python
np.random.seed(0)
n = 5000
x = np.random.uniform(-1, 1, n)
y = np.random.uniform(-1, 1, n)
inside_board = x**2 + y**2 <= 1
inside_inner = (x**2 + y**2 <= 0.25) & inside_board

P_sim = inside_inner.sum() / inside_board.sum()
print(f"시뮬 P(중심) ≈ {P_sim:.4f}, 이론 = 0.25")
```
