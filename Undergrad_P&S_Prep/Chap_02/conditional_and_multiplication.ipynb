# 1. 조건부확률과 확률의 곱셈정리

**조건부확률**:
$$P(A|B) = \dfrac{P(A \cap B)}{P(B)} \quad (P(B) > 0)$$

**곱셈정리**:
$$P(A \cap B) = P(B) \cdot P(A|B) = P(A) \cdot P(B|A)$$

## 1. 조건부확률

```python
import sympy as sp
sp.init_printing()

# 52장 카드: B=하트, A=킹
P_B = sp.Rational(13, 52)
P_AB = sp.Rational(1, 52)
P_A_given_B = P_AB / P_B
print(f"P(킹|하트) = P(킹∩하트)/P(하트) = {P_AB}/{P_B} = {P_A_given_B}")
```

**조건부확률 P(A|B)**: <br>
=> B가 주어졌을 때 A의 확률

**곱셈정리**: <br>
=> 연속 사건의 교집합 확률

## 2. 곱셈정리 Example

```python
# 주사위 2개: A=첫째 6, B=합 11
from itertools import product
S = list(product(range(1, 7), repeat=2))
A = {s for s in S if s[0] == 6}
B = {s for s in S if sum(s) == 11}
AB = A & B
P_A = sp.Rational(len(A), len(S))
P_B = sp.Rational(len(B), len(S))
P_AB = sp.Rational(len(AB), len(S))
print(f"P(A∩B) = {P_AB}")
print(f"P(A)·P(B|A) = {P_A} × {P_AB/P_A} = {P_A * P_AB/P_A}")
```
