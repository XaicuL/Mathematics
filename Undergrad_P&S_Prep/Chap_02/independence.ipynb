# 2. 사건의 독립과 종속

$A, B$가 **독립** $\Leftrightarrow$ $P(A \cap B) = P(A) \cdot P(B)$

동치: $P(A|B) = P(A)$, $P(B|A) = P(B)$

## 1. 독립 vs 종속

```python
import sympy as sp
sp.init_printing()

# 독립: 동전 2번
P_A = sp.Rational(1, 2)
P_B = sp.Rational(1, 2)
P_AB_indep = P_A * P_B
print(f"독립: P(A∩B) = {P_A}×{P_B} = {P_AB_indep}")

# 종속: 주사위 (A=짝수, B=3 이상)
S = 6
A = {2, 4, 6}
B = {3, 4, 5, 6}
AB = A & B
P_A2 = sp.Rational(len(A), S)
P_B2 = sp.Rational(len(B), S)
P_AB2 = sp.Rational(len(AB), S)
print(f"\n종속: P(A)P(B) = {P_A2*P_B2}, P(A∩B) = {P_AB2}")
print(f"독립? {P_A2 * P_B2 == P_AB2}")
```

**독립**: <br>
=> 한 사건이 다른 사건에 영향 없음

**종속**: <br>
=> $P(A \cap B) \ne P(A)P(B)$
