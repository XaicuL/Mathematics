# 2. 여사건의 확률

사건 $A$의 **여사건** $A^c$: $A$가 일어나지 않는 경우.

$$P(A^c) = 1 - P(A)$$

## 1. 여사건 공식

```python
import sympy as sp
sp.init_printing()

PA = sp.Rational(2, 5)
PAc = 1 - PA
print(f"P(A) = {PA}")
print(f"P(A^c) = 1 - {PA} = {PAc}")
```

**여사건**: <br>
=> '적어도 하나' 문제에서 `1 - P(모두 아님)` 활용

**sp.Rational**: <br>
=> 정확한 확률 계산

## 2. Example: 주사위 3번, 모두 6 아님

```python
P_not6 = sp.Rational(5, 6)
P_all_not6 = P_not6 ** 3
P_at_least_one6 = 1 - P_all_not6
print(f"P(적어도 하나 6) = 1 - (5/6)³ = {P_at_least_one6}")
print(f"≈ {float(P_at_least_one6):.4f}")
```
