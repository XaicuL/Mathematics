# 3. 독립시행의 확률

성공확률 $p$인 시행을 $n$번 **독립** 반복.

정확히 $k$번 성공:
$$P(X=k) = \binom{n}{k} p^k (1-p)^{n-k}$$

## 1. 이항확률

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

sp.init_printing()

n, p = 10, sp.Rational(1, 2)
k = 4
P_k = sp.binomial(n, k) * p**k * (1 - p)**(n - k)
print(f"P(X={k}) = C({n},{k}) × ({p})^{k} × ({1-p})^{n-k} = {P_k}")
```

**scipy.stats.binom.pmf(k, n, p)**: <br>
=> $P(X=k)$ 계산

**sp.binomial(n, k)**: <br>
=> 이항계수

## 2. PMF 그래프

```python
n, p = 10, 0.3
ks = np.arange(0, n + 1)
pmf = binom.pmf(ks, n, p)

plt.figure(figsize=(7, 4))
plt.bar(ks, pmf, color='steelblue', edgecolor='black')
plt.xlabel('k (success count)')
plt.ylabel('P(X=k)')
plt.title(f'independent trials n={n}, p={p}')
plt.xticks(ks)
plt.tight_layout()
plt.show()
```
