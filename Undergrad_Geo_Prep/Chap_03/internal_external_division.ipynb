# 2. 선분의 내분점과 외분점

$A$, $B$를 $m:n$으로 **내분**하는 점 $P$:

$$
P = \frac{nA + mB}{m+n}
$$

**외분**: $P = \dfrac{-nA + mB}{m-n}$ ($m\neq n$).


## 1. 내분·외분

```python
import numpy as np
import matplotlib.pyplot as plt

A = np.array([1, 1])
B = np.array([7, 4])
m, n = 2, 1

P_int = (n*A + m*B) / (m + n)
P_ext = (-n*A + m*B) / (m - n)

print(f"{m}:{n} 내분점 P =", P_int)
print(f"{m}:{n} 외분점 P' =", P_ext)
```

**(n*A + m*B)/(m+n)**: <br>
=> $m:n$ 내분점 좌표


## 2. 시각화

```python
fig, ax = plt.subplots(figsize=(8, 4))
for pt, c, lbl in [(A, 'b', 'A'), (B, 'r', 'B'), (P_int, 'g', 'internal'), (P_ext, 'm', 'external')]:
    ax.plot(pt[0], pt[1], 'o', color=c, ms=10, label=lbl)
ax.plot([A[0], B[0]], [A[1], B[1]], 'k--', alpha=0.4)
ax.set_aspect('equal'); ax.grid(True, alpha=0.3); ax.legend()
plt.show()
```
