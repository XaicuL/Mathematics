# 3. 삼수선의 정리

평면 $\alpha$ 위의 직선 $l$에 대해, $l$에 수직인 평면 $\beta$가 $\alpha$와 만나는 직선 $m$에 대해:

**$\beta$ 위의 직선 $PA$가 $m$에 수직 $\Leftrightarrow$ $PA$의 $\alpha$ 위 투영 $PA'$가 $l$에 수직.**


## 1. 좌표 예시

```python
import numpy as np

# plane alpha: z=0, line l along x-axis
# P = (0, 2, 3), A on l: (a, 0, 0)
P = np.array([0, 2, 3])
A = np.array([2, 0, 0])
A_proj = np.array([A[0], A[1], 0])  # projection onto z=0... actually A already on plane

# PA vector and its projection PA' (drop z component from P to A direction)
PA = A - P
PA_proj = np.array([PA[0], PA[1], 0])

l_dir = np.array([1, 0, 0])
print("PA =", PA)
print("PA' (xy 투영) =", PA_proj)
print("PA' ⊥ l:", abs(np.dot(PA_proj, l_dir)) < 1e-9)
print("PA ⊥ l (3D):", abs(np.dot(PA, l_dir)) < 1e-9)
```

**투영**: <br>
=> $xy$평면에 투영할 때 $z$성분 제거


## 2. 시각화

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
# plane z=0
xx, yy = np.meshgrid(np.linspace(-1, 4, 5), np.linspace(-1, 4, 5))
ax.plot_surface(xx, yy, np.zeros_like(xx), alpha=0.2)
# line l on x-axis
ax.plot([0, 4], [0, 0], [0, 0], 'b', lw=2, label='l (on α)')
ax.plot([P[0], A[0]], [P[1], A[1]], [P[2], A[2]], 'r', lw=2, label='PA')
ax.plot([P[0], A[0]], [P[1], A[1]], [0, 0], 'g--', lw=2, label="PA'")
ax.scatter(*P, color='k', s=60)
ax.scatter(*A, color='r', s=60)
ax.legend()
plt.show()
```
