# 1. 두 점 사이의 거리

공간에서 $A(x_1,y_1,z_1)$, $B(x_2,y_2,z_2)$:

$$
AB = \sqrt{(x_2-x_1)^2+(y_2-y_1)^2+(z_2-z_1)^2}
$$


## 1. 3D 거리 공식

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

A = np.array([1, 2, 3])
B = np.array([4, 6, 7])

dist = np.linalg.norm(B - A)
print("A =", A)
print("B =", B)
print("AB =", dist)
```

**np.linalg.norm(B-A)**: <br>
=> 3D(또는 n차) 유클리드 거리


## 2. 3D 시각화

```python
fig = plt.figure(figsize=(7, 6))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(*A, color='b', s=80, label='A')
ax.scatter(*B, color='r', s=80, label='B')
ax.plot([A[0], B[0]], [A[1], B[1]], [A[2], B[2]], 'g--', lw=2, label=f'd={dist:.2f}')
ax.set_xlabel('x'); ax.set_ylabel('y'); ax.set_zlabel('z')
ax.legend()
plt.show()
```
