# 2. 직선과 평면이 이루는 각

직선의 방향벡터 $\vec{v}$, 평면의 법선 $\vec{n}$:

$$
\sin\theta = \frac{|\vec{v}\cdot\vec{n}|}{|\vec{v}||\vec{n}|}
$$

($\theta$: 직선과 평면이 이루는 예각)


## 1. 각도 계산

```python
import numpy as np

v = np.array([1, 1, 1])
n = np.array([0, 0, 1])

sin_theta = abs(np.dot(v, n)) / (np.linalg.norm(v) * np.linalg.norm(n))
theta = np.arcsin(np.clip(sin_theta, 0, 1))
print(f"sin θ = {sin_theta:.4f}")
print(f"θ = {np.degrees(theta):.1f}°")
```

**np.arcsin**: <br>
=> $\sin\theta$에서 $\theta$ (라디안)


## 2. 예시

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(111, projection='3d')
ax.quiver(0, 0, 0, v[0], v[1], v[2], color='r', label='v')
ax.quiver(0, 0, 0, n[0], n[1], n[2], color='b', label='n')
ax.set_xlim(-1, 2); ax.set_ylim(-1, 2); ax.set_zlim(-1, 2)
ax.legend()
plt.show()
```
