# 1. 직선과 평면의 위치 관계

평면 $\pi: ax+by+cz+d=0$, 직선 $\ell: \vec{r}=\vec{r_0}+t\vec{v}$.

$\vec{v}\cdot(a,b,c)=0$이면 평행, 아니면 한 점에서 만남.


## 1. 교점 판별

```python
import numpy as np

# plane: x + y + z - 6 = 0
n = np.array([1, 1, 1])
d = -6

# line: r0 + t*v
r0 = np.array([0, 0, 1])
v = np.array([1, 1, 0])  # parallel to plane

dot_nv = np.dot(n, v)
print("n·v =", dot_nv)
if abs(dot_nv) < 1e-9:
    # check if on plane
    if abs(np.dot(n, r0) + d) < 1e-9:
        print("직선이 평면 위에 있음")
    else:
        print("직선이 평면과 평행 (불일치)")
else:
    t = -(np.dot(n, r0) + d) / dot_nv
    P = r0 + t * v
    print(f"교점 t={t:.3f}, P={P}")
```

**n·v**: <br>
=> 0이면 직선 ∥ 평면

**t = -(n·r0+d)/(n·v)**: <br>
=> 교점 매개변수


## 2. 교점 있는 경우

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

v2 = np.array([1, 0, 1])
t_hit = -(np.dot(n, r0) + d) / np.dot(n, v2)
P_hit = r0 + t_hit * v2
print("교점 P =", P_hit)

fig = plt.figure(figsize=(7, 6))
ax = fig.add_subplot(111, projection='3d')
xx, yy = np.meshgrid(np.linspace(-1, 5, 10), np.linspace(-1, 5, 10))
zz = (6 - xx - yy)  # x+y+z=6
ax.plot_surface(xx, yy, zz, alpha=0.3)
ts = np.linspace(-1, 3, 50)
line = r0 + np.outer(ts, v2)
ax.plot(line[:,0], line[:,1], line[:,2], 'r', lw=2)
ax.scatter(*P_hit, color='k', s=80)
plt.show()
```
