# 3. 구의 방정식

중심 $C(a,b,c)$, 반지름 $r$:

$$
(x-a)^2+(y-b)^2+(z-c)^2=r^2
$$


## 1. 방정식

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

sp.init_printing()
x, y, z = sp.symbols('x y z')
a, b, c, r = 1, -2, 3, 4

eq = (x-a)**2 + (y-b)**2 + (z-c)**2 - r**2
print("구의 방정식:")
display(sp.expand(eq))

center = np.array([a, b, c])
P = np.array([5, -2, 3])
print("P에서 중심까지:", np.linalg.norm(P - center))
```

**3변수 sp.symbols**: <br>
=> $(x-a)^2+(y-b)^2+(z-c)^2=r^2$ 기호 처리


## 2. 3D 시각화

```python
from mpl_toolkits.mplot3d import Axes3D

u = np.linspace(0, 2*np.pi, 40)
v = np.linspace(0, np.pi, 20)
xs = a + r * np.outer(np.cos(u), np.sin(v))
ys = b + r * np.outer(np.sin(u), np.sin(v))
zs = c + r * np.outer(np.ones_like(u), np.cos(v))

fig = plt.figure(figsize=(7, 6))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(xs, ys, zs, alpha=0.6, color='steelblue')
ax.scatter(a, b, c, color='k', s=50)
ax.set_xlabel('x'); ax.set_ylabel('y'); ax.set_zlabel('z')
ax.set_title(f'center of sphere: ({a},{b},{c}), r={r}')
plt.show()
```
