# 1. 타원의 방정식

두 초점 $F(c,0)$, $F'(-c,0)$ ($c>0$), $c^2=a^2-b^2$ ($a>b>0$):

$$
\frac{x^2}{a^2} + \frac{y^2}{b^2} = 1
$$

장축 $2a$, 단축 $2b$, 이심률 $e=c/a$.


## 1. 초점과 방정식

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

sp.init_printing()
a, b = sp.symbols('a b', positive=True)
c = sp.sqrt(a**2 - b**2)

print("타원: x²/a² + y²/b² = 1")
print("초점 (±c, 0), c =")
display(c)

a_val, b_val = 5, 3
c_val = float(sp.sqrt(a_val**2 - b_val**2))
print(f"a={a_val}, b={b_val} → c={c_val:.3f}, e={c_val/a_val:.3f}")
```

**sp.sqrt(expr)**: <br>
=> 기호적 제곱근


## 2. 시각화

```python
theta = np.linspace(0, 2*np.pi, 400)
xs = a_val * np.cos(theta)
ys = b_val * np.sin(theta)

fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(xs, ys, 'b', lw=2)
ax.plot([c_val, -c_val], [0, 0], 'red', ms=8, label='focus')
ax.set_aspect('equal'); ax.grid(True, alpha=0.3); ax.legend()
ax.set_title(f'$x²/{a_val}² + y²/{b_val}² = 1$')
plt.show()
```
