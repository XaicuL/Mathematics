# 2. 쌍곡선과 직선의 위치 관계

쌍곡선 $\dfrac{x^2}{a^2}-\dfrac{y^2}{b^2}=1$과 $y=mx+n$의 교점 이차방정식의 판별식 $D$로 위치 관계를 판정.


## 1. 판별식

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

sp.init_printing()
x, m, n = sp.symbols('x m n', real=True)
a, b = sp.symbols('a b', positive=True)

A = 1/a**2 - m**2/b**2
B = -2*m*n/b**2
C = -n**2/b**2 - 1
D = sp.expand(B**2 - 4*A*C)
print("판별식 D =")
display(sp.factor(D))
```

**sp.factor**: <br>
=> 판별식을 인수분해하여 접·교·이등 조건 파악


## 2. 예시

```python
a_val, b_val = 3, 2
m_val, n_val = 0.5, 1

A = 1/a_val**2 - m_val**2/b_val**2
B = -2*m_val*n_val/b_val**2
C = -n_val**2/b_val**2 - 1
D = B**2 - 4*A*C
print(f"D = {D:.4f}")

fig, ax = plt.subplots(figsize=(7, 5))
for xs in [np.linspace(a_val, 8, 200), np.linspace(-8, -a_val, 200)]:
    ys = b_val/a_val * np.sqrt(xs**2 - a_val**2)
    ax.plot(xs, ys, 'b'); ax.plot(xs, -ys, 'b')
xline = np.linspace(-6, 6, 100)
ax.plot(xline, m_val*xline + n_val, 'r--', label=f'y={m_val}x+{n_val}')
ax.set_aspect('equal'); ax.grid(True, alpha=0.3); ax.legend()
plt.show()
```
