# 1. 쌍곡선의 방정식

초점 $(\pm c, 0)$, $c^2=a^2+b^2$ ($a,b>0$):

$$
\frac{x^2}{a^2} - \frac{y^2}{b^2} = 1
$$

점근선: $y = \pm\dfrac{b}{a}x$.


## 1. 초점과 점근선

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

sp.init_printing()
a, b, x = sp.symbols('a b x', positive=True)
c = sp.sqrt(a**2 + b**2)
asym = (b / a) * x

print("쌍곡선: x²/a² - y²/b² = 1")
print("c ="); display(c)
print("점근선 y ="); display(asym)

a_val, b_val = 3, 2
c_val = float(sp.sqrt(a_val**2 + b_val**2))
print(f"a={a_val}, b={b_val} → c={c_val:.3f}")
```

**sp.symbols('a b x', positive=True)**: <br>
=> 양수 기호 변수 정의

**(b / a) * x**: <br>
=> 기호식 점근선 $y = \dfrac{b}{a}x$ (정수 전용 `sp.Rational`과 달리 기호 $a,b$에 사용)


## 2. 시각화

```python
x_right = np.linspace(a_val, 8, 200)
x_left = np.linspace(-8, -a_val, 200)
for xs in [x_right, x_left]:
    ys = b_val/a_val * np.sqrt(xs**2 - a_val**2)
    plt.plot(xs, ys, 'b', lw=2)
    plt.plot(xs, -ys, 'b', lw=2)

x_asym = np.linspace(-8, 8, 100)
plt.plot(x_asym, (b_val/a_val)*x_asym, 'r--', alpha=0.6, label='asymptote')
plt.plot(x_asym, -(b_val/a_val)*x_asym, 'r--', alpha=0.6)
plt.plot([c_val, -c_val], [0, 0], 'ko', ms=8, label='focus')
plt.gca().set_aspect('equal'); plt.grid(True, alpha=0.3); plt.legend()
plt.title('$x²/9 - y²/4 = 1$'); plt.show()
```
