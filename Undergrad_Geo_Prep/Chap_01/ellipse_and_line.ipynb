# 2. 타원과 직선의 위치 관계

타원 $\dfrac{x^2}{a^2}+\dfrac{y^2}{b^2}=1$과 직선 $y=mx+n$의 교점은

$$
\left(\frac{1}{a^2}+\frac{m^2}{b^2}\right)x^2 + \frac{2mn}{b^2}x + \left(\frac{n^2}{b^2}-1\right)=0
$$


## 1. 판별식

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

sp.init_printing()
x, m, n = sp.symbols('x m n', real=True)
a, b = sp.symbols('a b', positive=True)

A = 1/a**2 + m**2/b**2
B = 2*m*n/b**2
C = n**2/b**2 - 1
D = sp.expand(B**2 - 4*A*C)
print("판별식 D =")
display(sp.factor(D))
```

## 2. 예시

```python
a_val, b_val = 5, 3
m_val, n_val = 0.3, 0

A = 1/a_val**2 + m_val**2/b_val**2
B = 2*m_val*n_val/b_val**2
C = n_val**2/b_val**2 - 1
D = B**2 - 4*A*C
print(f"D = {D:.4f}, 교점 {'2개' if D>0 else '1개' if D==0 else '0개'}")

theta = np.linspace(0, 2*np.pi, 400)
fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(a_val*np.cos(theta), b_val*np.sin(theta), 'b', lw=2)
xs = np.linspace(-6, 6, 100)
ax.plot(xs, m_val*xs + n_val, 'r--', label=f'y={m_val}x')
ax.set_aspect('equal'); ax.grid(True, alpha=0.3); ax.legend()
plt.show()
```
