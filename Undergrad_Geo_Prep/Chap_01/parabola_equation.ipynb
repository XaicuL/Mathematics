# 1. 포물선의 방정식

초점 $F(p,0)$, 준선 $x=-p$ ($p>0$)인 포물선:

$$
y^2 = 4px
$$

꼭짓점 $(0,0)$, 초점 $(p,0)$, 준선 $x=-p$.


## 1. 표준형과 초점·준선

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

sp.init_printing()
x, y, p = sp.symbols('x y p', positive=True)

eq = sp.Eq(y**2, 4*p*x)
print("포물선:", eq)
print("초점 F = (p, 0), 준선 x = -p")

p_val = 2
focus = (p_val, 0)
directrix = -p_val
print(f"p={p_val} → 초점 {focus}, 준선 x={directrix}")
```

**sp.Eq(lhs, rhs)**: <br>
=> 방정식 $lhs = rhs$ 객체

**sp.symbols(..., positive=True)**: <br>
=> 양수 가정 기호


## 2. 시각화

```python
p_val = 2
t = np.linspace(-4, 4, 400)
y_pos = np.sqrt(4 * p_val * np.maximum(t, 0))
y_neg = -y_pos

fig, ax = plt.subplots(figsize=(7, 5))
ax.plot(t, y_pos, 'b', lw=2)
ax.plot(t, y_neg, 'b', lw=2)
ax.axvline(-p_val, color='r', ls='--', label=f'directrix x={-p_val}')
ax.plot(p_val, 0, 'ko', ms=8, label=f'focus ({p_val},0)')
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.legend()
ax.set_xlabel('x'); ax.set_ylabel('y')
ax.set_title(f'$y^2 = 4·{p_val}x$')
plt.show()
```

**np.linspace(a, b, n)**: <br>
=> $a$부터 $b$까지 $n$개 등간격 점

**ax.axvline(x)**: <br>
=> 수직선 (준선 표시)
