# 1. 평면벡터의 성분

$\vec{a}=(a_1,a_2)$, $\vec{b}=(b_1,b_2)$일 때

$$
\vec{a}+\vec{b}=(a_1+b_1,\, a_2+b_2), \quad k\vec{a}=(ka_1,\, ka_2)
$$

$|\vec{a}|=\sqrt{a_1^2+a_2^2}$.


## 1. 성분 연산

```python
import numpy as np
import matplotlib.pyplot as plt

a = np.array([3, -4])
b = np.array([-1, 2])

print("a =", a, ", |a| =", np.linalg.norm(a))
print("b =", b)
print("a + b =", a + b)
print("2a - 3b =", 2*a - 3*b)
```

**np.linalg.norm**: <br>
=> $\sqrt{a_1^2+a_2^2}$


## 2. 단위벡터

```python
unit_a = a / np.linalg.norm(a)
print("a의 단위벡터:", unit_a)
print("검증 |unit| =", np.linalg.norm(unit_a))

fig, ax = plt.subplots(figsize=(5, 5))
ax.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color='b', label='a')
ax.quiver(0, 0, unit_a[0], unit_a[1], angles='xy', scale_units='xy', scale=1, color='r', label='unit a')
ax.set_aspect('equal'); ax.grid(True, alpha=0.3); ax.legend()
plt.show()
```
