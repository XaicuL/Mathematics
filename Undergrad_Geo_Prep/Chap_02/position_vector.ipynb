# 4. 위치벡터

원점 $O$에서 점 $P(x,y)$까지: $\overrightarrow{OP}=\vec{p}=(x,y)$.

$A(x_1,y_1)$, $B(x_2,y_2)$이면 $\overrightarrow{AB}=\vec{b}-\vec{a}$.


## 1. 위치벡터 계산

```python
import numpy as np
import matplotlib.pyplot as plt

A = np.array([1, 2])
B = np.array([4, 6])

p_A = A  # position vector of A
p_B = B
AB = B - A

print("A의 위치벡터:", p_A)
print("B의 위치벡터:", p_B)
print("AB = B - A =", AB)
```

**B - A**: <br>
=> 두 점을 잇는 벡터 $\overrightarrow{AB}$


## 2. 시각화

```python
fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(A[0], A[1], 'bo', ms=10, label='A')
ax.plot(B[0], B[1], 'ro', ms=10, label='B')
ax.quiver(0, 0, A[0], A[1], angles='xy', scale_units='xy', scale=1, color='b', alpha=0.5)
ax.quiver(A[0], A[1], AB[0], AB[1], angles='xy', scale_units='xy', scale=1, color='g', label='AB')
ax.set_aspect('equal'); ax.grid(True, alpha=0.3); ax.legend()
plt.show()
```
