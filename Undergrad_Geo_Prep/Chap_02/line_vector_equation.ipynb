# 1. 직선의 벡터방정식

점 $P_0(x_0,y_0)$, 방향벡터 $\vec{v}=(v_1,v_2)$:

$$
\vec{r} = \vec{r_0} + t\vec{v} \quad (t\in\mathbb{R})
$$

매개변수 형: $x=x_0+tv_1$, $y=y_0+tv_2$.


## 1. 벡터방정식

```python
import numpy as np
import matplotlib.pyplot as plt

r0 = np.array([1, 2])
v = np.array([2, 1])

t_vals = np.linspace(-2, 3, 100)
points = r0 + np.outer(t_vals, v)

print("r0 =", r0, ", v =", v)
print("t=0 →", r0)
print("t=1 →", r0 + v)
```

**np.outer(t_vals, v)**: <br>
=> 각 $t$에 대해 $t\vec{v}$ 배열


## 2. 시각화

```python
fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(points[:, 0], points[:, 1], 'b', lw=2, label='straight line')
ax.plot(r0[0], r0[1], 'ko', ms=8, label='P0')
ax.quiver(r0[0], r0[1], v[0], v[1], angles='xy', scale_units='xy', scale=1, color='r', label='v')
ax.set_aspect('equal'); ax.grid(True, alpha=0.3); ax.legend()
plt.show()
```
