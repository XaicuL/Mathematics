# 3. 평면벡터의 수직과 평행

$\vec{a}\parallel\vec{b} \Leftrightarrow a_1b_2-a_2b_1=0$ (외적 성분)

$\vec{a}\perp\vec{b} \Leftrightarrow a_1b_1+a_2b_2=0$.


## 1. 평행·수직 판별

```python
import numpy as np

def is_parallel(a, b, tol=1e-9):
    return abs(a[0]*b[1] - a[1]*b[0]) < tol

def is_perpendicular(a, b, tol=1e-9):
    return abs(np.dot(a, b)) < tol

a = np.array([2, 4])
b = np.array([1, 2])   # parallel
c = np.array([-2, 1])  # perpendicular

print("a∥b:", is_parallel(a, b))
print("a⊥c:", is_perpendicular(a, c))
```

**a[0]*b[1]-a[1]*b[0]**: <br>
=> 2D 외적 (평행 ⇔ 0)


## 2. 시각화

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 6))
for v, c, lbl in [(a, 'b', 'a'), (b, 'r', 'b ∥ a'), (c, 'g', 'c ⊥ a')]:
    ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color=c, label=lbl)
ax.set_aspect('equal'); ax.grid(True, alpha=0.3); ax.legend()
plt.show()
```
