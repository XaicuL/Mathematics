# 2. 평면벡터의 내적

$$
\vec{a}\cdot\vec{b} = a_1b_1 + a_2b_2 = |\vec{a}||\vec{b}|\cos\theta
$$

$\vec{a}\perp\vec{b} \Leftrightarrow \vec{a}\cdot\vec{b}=0$.


## 1. 내적 계산

```python
import numpy as np
import matplotlib.pyplot as plt

a = np.array([3, 1])
b = np.array([1, 2])

dot = np.dot(a, b)
cos_theta = dot / (np.linalg.norm(a) * np.linalg.norm(b))
theta = np.arccos(np.clip(cos_theta, -1, 1))

print("a·b =", dot)
print(f"각 θ = {np.degrees(theta):.1f}°")
```

**np.dot(a, b)**: <br>
=> 내적 $a_1b_1+a_2b_2$

**np.arccos(x)**: <br>
=> 역코사인 (라디안)


## 2. 수직·예각 판별

```python
vecs = {
    'acute': np.array([1, 1]),
    'right': np.array([-1, 1]),
    'obtuse': np.array([-1, -1]),
}
base = np.array([2, 0])
for name, v in vecs.items():
    d = np.dot(base, v)
    print(f"{name}: a·b = {d}")

fig, ax = plt.subplots(figsize=(5, 5))
ax.quiver(0, 0, base[0], base[1], angles='xy', scale_units='xy', scale=1, color='red', label='a')
for v in vecs.values():
    ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, alpha=0.7)
ax.set_aspect('equal'); ax.grid(True, alpha=0.3); ax.legend()
plt.show()
```
