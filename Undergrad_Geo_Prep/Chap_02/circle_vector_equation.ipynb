# 2. 원의 벡터방정식

중심 $C(\vec{c})$, 반지름 $r$:

$$
|\vec{r}-\vec{c}| = r
$$

전개: $(x-c_1)^2+(y-c_2)^2=r^2$.


## 1. 벡터형과 좌표형

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

sp.init_printing()
x, y = sp.symbols('x y')
c1, c2, r = 2, -1, 3

eq = (x - c1)**2 + (y - c2)**2 - r**2
print("원의 방정식 (전개):")
display(sp.expand(eq))

# point on circle?
P = (5, -1)
dist = np.sqrt((P[0]-c1)**2 + (P[1]-c2)**2)
print(f"점 {P}에서 중심까지 거리 = {dist} (r={r})")
```

**sp.expand**: <br>
=> 원방정식 전개·인수분해


## 2. 시각화

```python
theta = np.linspace(0, 2*np.pi, 200)
xs = c1 + r * np.cos(theta)
ys = c2 + r * np.sin(theta)

fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(xs, ys, 'b', lw=2)
ax.plot(c1, c2, 'ko', ms=8, label='center C')
ax.plot(P[0], P[1], 'r^', ms=10, label='point P')
ax.set_aspect('equal'); ax.grid(True, alpha=0.3); ax.legend()
plt.show()
```
