# 1. 벡터의 뜻과 표시법

크기와 방향을 가진 양. $\vec{a}$, $\overrightarrow{AB}$로 표기.

$$
|\vec{a}| = \text{벡터의 크기(길이)}
$$


## 1. 벡터 표현

```python
import numpy as np
import matplotlib.pyplot as plt

a = np.array([3, 4])
b = np.array([-2, 1])

print("벡터 a =", a)
print("|a| =", np.linalg.norm(a))
print("벡터 b =", b)
print("|b| =", np.linalg.norm(b))
```

**np.array([x, y])**: <br>
=> 2차원 벡터 생성

**np.linalg.norm(v)**: <br>
=> 벡터의 크기 $|v|$


## 2. 시각화

```python
fig, ax = plt.subplots(figsize=(6, 6))
ax.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color='b', label='a')
ax.quiver(0, 0, b[0], b[1], angles='xy', scale_units='xy', scale=1, color='r', label='b')
ax.set_xlim(-3, 5); ax.set_ylim(-1, 5)
ax.set_aspect('equal'); ax.grid(True, alpha=0.3); ax.legend()
ax.axhline(0, color='k', lw=0.5); ax.axvline(0, color='k', lw=0.5)
plt.show()
```
