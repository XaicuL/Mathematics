# 3. 벡터의 실수배

$k\vec{a}$: 방향은 $\vec{a}$와 같거나($k>0$) 반대($k<0$), 크기는 $|k||\vec{a}|$.


## 1. 실수배

```python
import numpy as np
import matplotlib.pyplot as plt

a = np.array([2, 1])
for k in [2, -1, 0.5]:
    ka = k * a
    print(f"{k}·a = {ka}, |{k}·a| = {np.linalg.norm(ka):.3f}")
```

**k * v**: <br>
=> 스칼라와 벡터의 곱 (실수배)


## 2. 시각화

```python
fig, ax = plt.subplots(figsize=(6, 6))
colors = ['b', 'g', 'r', 'm']
for k, c in zip([2, 1, -1, 0.5], colors):
    v = k * a
    ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color=c, label=f'{k}a')
ax.set_aspect('equal'); ax.grid(True, alpha=0.3); ax.legend()
plt.show()
```
