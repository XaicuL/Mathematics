# 2. 벡터의 덧셈과 뺄셈

$$
\vec{a}+\vec{b}, \quad \vec{a}-\vec{b}=\vec{a}+(-\vec{b})
$$

평행사변형 법칙, 삼각형 법칙.


## 1. 덧셈·뺄셈

```python
import numpy as np
import matplotlib.pyplot as plt

a = np.array([3, 1])
b = np.array([1, 3])

print("a + b =", a + b)
print("a - b =", a - b)
```

**벡터 +, -**: <br>
=> 성분별 덧셈·뺄셈


## 2. 평행사변형 법칙

```python
fig, ax = plt.subplots(figsize=(6, 6))
origin = np.array([0, 0])
for v, c, lbl in [(a, 'b', 'a'), (b, 'r', 'b'), (a+b, 'g', 'a+b')]:
    ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color=c, label=lbl)
# parallelogram
ax.plot([a[0], a[0]+b[0]], [a[1], a[1]+b[1]], 'r--', alpha=0.5)
ax.plot([b[0], a[0]+b[0]], [b[1], a[1]+b[1]], 'b--', alpha=0.5)
ax.set_aspect('equal'); ax.grid(True, alpha=0.3); ax.legend()
plt.show()
```
