## 계수비교법

- 양변의 동류항 계수를 비교. <br>

계수비교법은 두 가지의 전제를 필요로 한다. <br>

1. 항등식이여야 한다.
2. 두 개의 식이 있어야 한다.
<br>

항등식이어야 한다는 점은 우리가 쉽게 이해할 수 있다. <br>
그러나 두 번째는 의아할 수 있다. <br>

??: 식 하나로만 해도 복잡해 죽겠는데 두개를 하라고? <br>
라고 생각 할 수 있는데 두 눈을 크게 떠서 계수 **비교** 법을 읽어보면 좋겠다. <br>
그러하다. 비교를 해야한다. 비교를 하려면 최소 2개는 있어야 한다는 것은 너무나 자명한 사실일테니 계수 비교법 또한 식이 최소 2개 필요하다.
<br>
그럼 어떤 방법일까? 다음의 예를 보자  <br>
~~코드를 보자는 말이다.~~

```python
import sympy as sp
sp.init_printing()

x = sp.symbols('x')
a = sp.symbols('a')
b = sp.symbols('b')

lhs = a*(x - 1)**2 + b * (x - 1) - 2
rhs = x**2 -2*x - 1

display(lhs)
print('위 식을 좌변이라 가정하고')
display(rhs)
print('위 식을 우변이라 가정하자')
print('우선 좌변을 정리하자.')

lhs_exp = sp.expand(lhs)
display(lhs_exp)

print('이제 전개한 좌변과 우변의 식을 = 관계로 두고 보면 좌변엔 미정계수가 있는데,')
print('좌변의 이차식 계수인 a와 우변의 이차식 계수를 = 관계로 두면 a = 1이라는 결과와 b = 0 이라는 결과가 나온다.')

expr = lhs - rhs

ex = sp.Poly(expr, x)
coeffs = ex.coeffs()

ans = sp.solve(coeffs, (a, b))


display(ans)
```

sp.Poly(cal_variableName, x): <br>
=> 수식을 다항식으로 변환한다고 생각하면 된다. <br>
**,x** 는 인터프리터에게 **x에 대한 식** 이라는 점을 시사한다고 생각하면 된다. <br><br>

그럼 여기서 궁금해진다. <br>
왜 x에 대한 식이라고 알려줘야만 할까?<br>
그 이유는 현재 우리가 다루는 식 속엔 미정계수가 포함되었기에 무엇이 항을 구분하는 기준인지 모르기 때문이다. <br>
쉬운 예를 들자면 콩밥에서 콩을 골라내는 것이 우리의 목표인데, 이때 컴퓨터는 무엇이 기준인지 모르기에 우리가 하얀 밥을 하나 내어준다 생각하면 된다. <br>

## 수치대입법
- 문자에 편리한 숫자를 대입하여 계산

<br>

사실 말그대로 **직관** 을 발휘하여 적당한 수를 **대입** 하는 것인데,<br>
이것 또한 앞서 설명한 계수 비교법과 전제는 동일하다
<br>
아래 코드로 보자.

```python
import sympy as sp
sp.init_printing()

x, a, b = sp.symbols('x a b')


lhs = a*(x - 1)**2 + b*(x - 1) - 2
rhs = x**2 - 2*x - 1


eq = lhs - rhs

print("--- 수치대입법 시작 ---")

eq1 = eq.subs(x, 1)
print(f"x=1 대입 시 식: {eq1} = 0")
display(eq1)


eq2 = eq.subs(x, 0)
print(f"x=0 대입 시 식: {eq2} = 0")
display(eq2)

eq3 = eq.subs(x,2)
print(f"x = 2 대입 시 식: {eq3} = 0")
display(eq3)

ans_sub = sp.solve([eq1, eq2], (a, b))

print("\n--- 최종 결과 ---")
print(f"미정계수 해: {ans_sub}")
display(ans_sub)
```

subs(cal_variableName, n): (n = 1,2,3,4,5...) <br>
=> **subs()** 함수는 Substitution(대입/교체)의 약자인데, **n** 은 대입을 시킬 수를 지칭한다 <br>

solve([eqN , eqN], (sb1,sb2))<br>
=> **solve()** 함수는 해당 식이 **= 0** 이 되게 하는 값은 무엇인지를 구하게 하는 기능을 한다.

# 나머지 정리
<br>

나머지 정리란 **다항식의 나눗셈**을 진행하지 않고, 쉽게 **몫**과 **나머지** 를 구하기 위해 사용한다고 이해하면 좋다.
<br>

그런데 여기서 식이 좀 처음에 보면 복잡해보일 수 있는데, 다음과 같다 <br>
**P(x) = (x-α)Q(x)+R** 이때 **몫 ≈ Q(x)** **나머지 ≈ R** 이라고 칭하는데, 해당 식에 **x = α** 을 대입하면 <br>
**R = P(α)** 가 만족된다 <br>

정리해서 요약하여 쓴다면 다음과 같다. <br>
다항식 P(x)를 일차식 x-α 로 나누었을 때의 나머지를 R 이라고 하면 **R = P(α)** 가 성립한다. <br>

~~독자가 존재할지 모른다면, 독자에게 양해를 구한다. 본인이 봐도 가독성도, 글 솜씨도 없는 것 같다. 코드는 잘 쓰기 위해 노력해보겠다..~~

```python
import sympy as sp
sp.init_printing()

x = sp.symbols('x')

px = 2*x**3 + 1 * x ** 2 + 1 * x + 1
display(px)

R = px.subs(x,-1)
display(R)
```

새로 사용하는 함수가 없다.. 그러하다... 그렇기에 검산의 과정을 보여주고자 한다. **feat 다항식의 나눗셈**

```python
quotient, remainder = sp.div(px, x + 1)

print("몫:", quotient)       # 2*x**2 - x + 2
print("나머지:", remainder) # -1
```

정확하게 일치함이 확인된다. 이때 주의해야하는 것을 하나 상기하고 가도록 하자. <br>

**나머지 정리** 는 나눗셈을 하지 않고 **나머지** 를 쉽게 구하는 것이지 , 이것으로 **몫** 까지 구할 순 없다. <br>
검산의 과정에선 **다항식의 나눗셈** 을 이용하면 **몫** 과 **나머지** 를 구할 수 있으니 다항식의 나눗셈을 적극적으로 활용할 수 있어야 할 것 같다.

# 인수 정리 <br>

인수 정리는 처음에 배우면 도대체 **나머지 정리** 와 무슨 차이인지 감이 잡히지 않을 수 있다. <br>
딱 이거만 기억해주면 완벽한데, 다음과 같다. <br>

    - 나머지 정리의 나머지는 상수 OR 0
    - 인수 정리의 나머지는 Must 0

이 사실이 중요한 이유는 **인수 정리** 는 결국 **나누어 떨어지는 인수를 찾아내는** 과정이다. <br>
필자의 교과서엔 다음과 같은 서술로 정리되어 있기에 공유하고자 한다. <br>

---------------------------

다항식 P(x)에 대하여
1. P(α) = 0 이면 일차식 x - α 로 나누어 떨어진다.
2. P(x)가 일차식 x - α 로 나누어떨어지면 P(α) = 0이다.
