이떄 text로만 보면 더 난감해질 수 있다. 다음과 같은 공식은 보자 <br>

a*x**2 + b*x + c = 0 <br>

해당 식에서 **a,b,c** 는 항등식의 입장에서 계수임은 자명하다. <br>
그런데 **a,b,c** 는 결코 숫자가 아닌 것은 누구나 알 수 있다 ~~엥? 싶었다면 당신은..... 말을 아끼겠다. 화이팅 하길 바란다.~~
<br>

이때 우리는 항등식으로 그래프 혹은 계산을 하기 위해선 미지수가 남아선 안된다 <br>
결론적으로 미지수를 없애기 위해선 미정계수법을 이용하여 미정계수를 규정해야 한다. 이렇게 <br>
**a = 5**
<br>

그럼 여기서 계수비교법이란 무엇일까?
그건 이후 설명하도록 하겠다. 여기서 설명하면 이미 백스텝 할 것이 자명할 것 같기 때문이다.

```python
import sympy as sp

x = sp.symbols('x')
a = sp.symbols('a')
b = sp.symbols('b')

lhs = a*(x - 1)**2 + b * (x - 1) - 2
rhs = x**2 -2*x - 1

expr = lhs - rhs

ex = sp.Poly(expr, x)
coeffs = ex.coeffs()

ans = sp.solve(coeffs, (a, b))

display(ans)
```

sp.Poly(cal_variableName, x): <br>
=> 수식을 다항식으로 변환한다고 생각하면 된다. <br>
**,x** 는 인터프리터에게 **x에 대한 식** 이라는 점을 시사한다고 생각하면 된다. <br><br>

그럼 여기서 궁금해진다. <br>
왜 x에 대한 식이라고 알려줘야만 할까?<br>
그 이유는 현재 우리가 다루는 식 속엔 미정계수가 포함되었기에 무엇이 항을 구분하는 기준인지 모르기 때문이다. <br>
쉬운 예를 들자면 콩밥에서 콩을 골라내는 것이 우리의 목표인데, 이때 컴퓨터는 무엇이 기준인지 모르기에 우리가 하얀 밥을 하나 내어준다 생각하면 된다. <br>
그러면 아무래도 콩밥엔 보기만 해도 먹으면 맛이 없을 것 같은 콩이 보이니 그걸 기준 삼아 볼 수 있을 것 이기 때문이다.

coeffs(): <br>
=> **Poly()** 함수로 인해 다항식으로 변환된 식에서 계수만 골라내주는 것이라 생각하면 된다. <br><br>

위에서 든 예를 이어가보자면, 콩을 골라낸 상태에서 그 먹기 싫고 먹어봤자 맛이 없는게 자명해보이는 콩을 <br>
다시 하얀 밥에 두면 의미가 없으니 콩만 골라둘 그릇을 준비해둔다 생각해주면 된다. <br>
