# **다항식의 인수분해**

```python
import sympy as sp
sp.init_printing()


a = sp.symbols('a')
b = sp.symbols('b')

expr = a**3 + 3*a**2 *b + 3*a*b**2 + b**3

cal = sp.factor(expr)

display(cal)
```

```python
import sympy as sp

a, b = sp.symbols('a b')
expr = a**3 + 3*a**2*b + 3*a*b**2 + b**3

# 결과가 즉시 나오는지 확인
print(sp.factor(expr))
```
