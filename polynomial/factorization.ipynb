# **다항식의 인수분해**

```python
import sympy as sp
sp.init_printing()


a = sp.symbols('a')
b = sp.symbols('b')

expr = a**3 + 3*a**2 *b + 3*a*b**2 + b**3

cal_fac = sp.factor(expr)
cal_exp = sp.expand(expr)

print('인수분해 식')
display(cal_fac)

print('전개식')
display(cal_exp)
```

## 인수분해 공식

```python
import sympy as sp
sp.init_printing()

a = sp.symbols('a')
b = sp.symbols('b')
c = sp.symbols('c')


print('### 인수분해 공식 1 ###')

expr1 = a**2 + b**2 + c**2 + 2*a*b + 2*b*c + 2*c*a
cal_1_fac = sp.factor(expr1)
cal_1_exp = sp.expand(expr1)


print('인수분해 식')
display(cal_1_fac)

print('전개식')
display(cal_1_exp)

print('----------------------------------------------')

print('### 인수분해 공식 2(+) ###')

expr2_p = a**3 + 3*a**2*b + 3*a*b**2 + b**3
cal_2p_fac = sp.factor(expr2_p)
cal_2p_exp = sp.expand(expr2_p)

print('인수분해 식')
display(cal_2p_fac)

print('전개식')
display(cal_2p_exp)

print('### 인수분해 공식 2(-) ###')

expr2_m = a**3 - 3*a**2*b + 3*a*b**2 - b**3

cal_2m_fac = sp.factor(expr2_m)
cal_2m_exp = sp.expand(expr2_m)

print('인수분해 식')
display(cal_2m_fac)

print('전개식')
display(cal_2m_exp)

print('----------------------------------------------')

print('### 인수분해 공식 3(+) ###')

expr3_p = a**3 + b**3
cal_3p_fac = sp.factor(expr3_p)
cal_3p_exp = sp.expand(expr3_p)


print('인수분해 식')
display(cal_3p_fac)

print('전개식')
display(cal_3p_exp)

print('### 인수분해 공식 3(+) ###')

expr3_m = a**3 - b**3
cal_3m_fac = sp.factor(expr3_m)
cal_3m_exp = sp.expand(expr3_m)


print('인수분해 식')
display(cal_3m_fac)

print('전개식')
display(cal_3m_exp)
```
