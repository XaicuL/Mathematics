# **수열의 합**

```python
from sympy import symbols, summation, simplify

k, n = symbols('k n', positive=True, integer=True)

S = summation(k, (k, 1, n))
print(simplify(S))   # n*(n + 1)/2

# 등차수열: a1=3, d=5 → Σ(3 + 5*(k-1)), k=1~5
a1, d = 3, 5
S_arith = summation(a1 + d * (k - 1), (k, 1, 5))
print(S_arith)   # 65
```
