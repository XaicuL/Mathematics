# 수열의 합

기호 $\displaystyle\sum_{k=1}^{n} a_k = a_1 + a_2 + \cdots + a_n$.

자주 쓰는 공식:

$$
\sum_{k=1}^{n} 1 = n, \quad
\sum_{k=1}^{n} k = \dfrac{n(n+1)}{2}, \quad
\sum_{k=1}^{n} k^2 = \dfrac{n(n+1)(2n+1)}{6}
$$


## 1. Σ의 성질


상수 $c$에 대해:

1. $\displaystyle\sum_{k=1}^{n} (a_k + b_k) = \sum a_k + \sum b_k$
2. $\displaystyle\sum_{k=1}^{n} c\, a_k = c \sum a_k$
3. $\displaystyle\sum_{k=1}^{n} c = cn$


```python
import sympy as sp
sp.init_printing()

k, n = sp.symbols('k n', integer=True, positive=True)
c = sp.symbols('c', real=True)

# 성질 3
display(sp.summation(c, (k, 1, n)))

# 선형성 예: Σ(2k + 3)
expr = sp.summation(2*k + 3, (k, 1, n))
display(sp.Eq(sp.Sum(2*k + 3, (k, 1, n)), sp.simplify(expr)))
```

**sp.summation(expr, (k, 1, n))**: <br>
=> 닫힌 식으로 합산

**sp.Sum(...)**: <br>
=> 합 기호 표현 (계산은 `.doit()` 또는 `summation`)


## 2. 기본 공식: $\sum k$, $\sum k^2$


```python
S1 = sp.summation(k, (k, 1, n))
S2 = sp.summation(k**2, (k, 1, n))

display(sp.Eq(sp.Sum(k, (k, 1, n)), sp.factor(S1)))
display(sp.Eq(sp.Sum(k**2, (k, 1, n)), sp.factor(S2)))

# 수치 확인 n=10
print("Σk (n=10):", S1.subs(n, 10), "==", sum(range(1, 11)))
print("Σk² (n=10):", S2.subs(n, 10), "==", sum(i**2 for i in range(1, 11)))
```

## 3. 예제


```python
# Σ_{k=1}^{n} (3k - 1)
ex1 = sp.summation(3*k - 1, (k, 1, n))
display(sp.Eq(sp.Sum(3*k - 1, (k, 1, n)), sp.simplify(ex1)))

# 등차: a=3, d=5 → Σ (3 + 5(k-1)), k=1..5
a1, d = 3, 5
S_arith = sp.summation(a1 + d*(k - 1), (k, 1, 5))
print("등차 a=3,d=5, n=5 →", S_arith)

# Σ k(k+1) = Σ(k²+k)
ex2 = sp.summation(k*(k + 1), (k, 1, n))
display(sp.Eq(sp.Sum(k*(k + 1), (k, 1, n)), sp.factor(sp.simplify(ex2))))
```
