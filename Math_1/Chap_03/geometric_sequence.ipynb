# 등비수열

첫째항 $a$, 공비 $r$ 인 등비수열:

$$
a_n = a\, r^{n-1}
$$

합 $S_n$:

$$
S_n =
\begin{cases}
na & (r = 1) \\
a\dfrac{1-r^n}{1-r} = a\dfrac{r^n-1}{r-1} & (r \neq 1)
\end{cases}
$$


## 1. 일반항 $a_n = ar^{n-1}$


```python
import sympy as sp
sp.init_printing()

n = sp.symbols('n', integer=True, positive=True)
a, r = sp.symbols('a r', real=True)

a_n = a * r**(n - 1)
display(sp.Eq(sp.Function('a')(n), a_n))

def term(a1, r, k):
    return a1 * r**(k - 1)

a1, rr = 3, 2
seq = [term(a1, rr, k) for k in range(1, 6)]
print("수열 a1~a5:", seq)
print("공비 확인 a_{k+1}/a_k:", [seq[i+1]/seq[i] for i in range(len(seq)-1)])

# a5 / a3 = r^2
print(f"a5/a3 = {seq[4]/seq[2]} = r^2 = {rr**2}")
```

**성질:** $\dfrac{a_m}{a_n} = r^{m-n}$ $(a_n \neq 0)$


## 2. 합 $S_n$ ($r \neq 1$)


```python
def S(a1, r, n):
    if r == 1:
        return a1 * n
    return a1 * (1 - r**n) / (1 - r)

a1, rr, N = 3, 2, 5
Sn = S(a1, rr, N)
print(f"a={a1}, r={rr}, n={N} → S_n = {Sn}")
print("검산:", sum(a1 * rr**k for k in range(N)))

# 기호
S_sym = a * (1 - r**n) / (1 - r)
display(sp.Eq(sp.Function('S')(n), S_sym))
```

**주의:** $r=1$이면 분모 $1-r=0$ → **반드시 분기**하여 $S_n = na$.


## 3. 특수 경우 $r = 1$


```python
# r = 1 → 상수열 a, a, a, ... → S_n = n*a
a1, N = 7, 10
print("r=1:", S(a1, 1, N), "==", a1 * N)

# r = -1 도 공식 적용 가능 (분모 ≠ 0)
print("r=-1, n=5:", S(1, -1, 5), "검산:", sum((-1)**k for k in range(5)))
print("r=-1, n=6:", S(1, -1, 6), "검산:", sum((-1)**k for k in range(6)))
```

## Example 


```python
# 첫째항 1, 공비 1/2, 10항까지의 합
print("S_10 (a=1, r=1/2) =", S(1, 0.5, 10))
print("기호:", sp.simplify(1 * (1 - sp.Rational(1, 2)**10) / (1 - sp.Rational(1, 2))))
```
