# **등비수열**

```python
def a_n(a1, r, n):
    return a1 * r**(n-1)

a1 = 3
r = 2

a5 = a_n(a1, r, 5)
a3 = a_n(a1, r, 3)

print(f'a5 = {a5}')
print(f'a3 = {a3}')

sol = a5//a3

print(f'a5/a3 = {sol}')

if sol == 2**2:
    print('equal')
    
```

# **등비수열의 합**

```python
def S_n(a1, r, n):
    return a1 * (1 - r**n) / (1 - r)

S_5 = int(S_n(a1, r, 5))

print(f'S_5 = {S_5}')
```
