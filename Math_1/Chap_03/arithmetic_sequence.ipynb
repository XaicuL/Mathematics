# 등차수열

첫째항 $a$, 공차 $d$ 인 등차수열:

$$
a_n = a + (n-1)d
$$

합 $S_n = a_1 + a_2 + \cdots + a_n$:

$$
S_n = \dfrac{n}{2}\bigl(2a + (n-1)d\bigr) = \dfrac{n(a + \ell)}{2}
$$

여기서 $\ell = a_n$ (마지막 항).


## 1. 일반항 $a_n = a + (n-1)d$


```python
import sympy as sp
sp.init_printing()

n = sp.symbols('n', integer=True, positive=True)
a, d = sp.symbols('a d', real=True)

a_n = a + (n - 1) * d
display(sp.Eq(sp.Function('a')(n), a_n))

# 수치 예: a=3, d=5
def term(a1, d, k):
    return a1 + (k - 1) * d

a1, dd = 3, 5
seq = [term(a1, dd, k) for k in range(1, 6)]
print("수열 a1~a5:", seq)
print("a5 =", term(a1, dd, 5))

# 공차 확인: a_{k+1} - a_k = d
print("공차들:", [seq[i+1] - seq[i] for i in range(len(seq)-1)])
```

**성질:** $a_m - a_n = (m-n)d$  
예: $a_5 - a_3 = 2d$ (공차의 2배).


```python
a1, dd = 3, 5
a5 = a1 + 4*dd
a3 = a1 + 2*dd
print(f"a5 - a3 = {a5 - a3} = 2d = {2*dd}")  # 10 = 10
```

## 2. 합 $S_n = n(a+\ell)/2$


```python
# 기호 공식
ell = a + (n - 1) * d  # 마지막 항 a_n
S_n = n * (a + ell) / 2
S_n_alt = n * (2*a + (n - 1)*d) / 2

print("두 공식 일치?", sp.simplify(S_n - S_n_alt) == 0)
display(sp.Eq(sp.Function('S')(n), sp.simplify(S_n)))

def S(a1, d, n):
    # 등차수열 합: n(a + L)/2 = n(2a + (n-1)d)/2
    return n * (2*a1 + (n - 1)*d) / 2

a1, dd, N = 3, 5, 5
ell = a1 + (N - 1)*dd
Sn = S(a1, dd, N)
print(f"a={a1}, d={dd}, n={N} → L=a_n={ell}")
print(f"S_n = n(a+L)/2 = {N}*({a1}+{ell})/2 = {Sn}")
print(f"검산 (직접 합): {sum(a1 + k*dd for k in range(N))}")
```

**올바른 항등식:**
- $S_n = \dfrac{n}{2}(a + \ell)$  (첫항+끝항)
- $S_n = \dfrac{n}{2}\bigl(2a+(n-1)d\bigr)$

**잘못된 표기 주의:** `$a_1 + a_n/2$` 은 합이 아님.  
또한 $S_n = n \cdot a_{(n+1)/2}$ 는 **n이 홀수**일 때만 성립 (가운데 항 × n).


## 3. 가운데 항 (n 홀수)


```python
# n=5(홀수): S_5 = 5 * a_3
a1, dd, N = 3, 5, 5
a3 = a1 + 2*dd
Sn = N * (2*a1 + (N - 1)*dd) / 2
print(f"S_5 = {Sn}, 5*a_3 = {N * a3}")
print("같음?", Sn == N * a3)

# n=4(짝수): 가운데 항이 하나 아님
# S_n = n * (첫항+끝항)/2 는 항상 성립
N2 = 4
Sn2 = N2 * (2*a1 + (N2 - 1)*dd) / 2
a2 = a1 + dd
a3b = a1 + 2*dd
mid_avg = (a2 + a3b) / 2
print(f"S_4 = {Sn2}, 4*(a2+a3)/2 = {N2 * mid_avg}")
```

**API:** 순수 Python / sympy 기호식만으로 충분.  
일반항·합 검증 시 `sum(...)` 직접 합으로 검산하면 된다.
