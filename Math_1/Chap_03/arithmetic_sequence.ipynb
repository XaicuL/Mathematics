# **등차수열**

```python
def a_n(a1, d, n):
    return a1 + (n - 1) * d

a1, d = 3, 5

a5 = a_n(a1, d, 5)
#print(a5)
a3 = a_n(a1, d, 3)

sol = a5 - a3

print(f'a5 = {a5}')
print(f'a3 = {a3}')
print(f'a5 - a3 = {sol}')

if 2*d == sol:
    print(f'd = {sol/2}')
else:
    print('error')
```
