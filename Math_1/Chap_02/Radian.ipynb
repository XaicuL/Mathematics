```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
theta_circle = np.linspace(0, 2 * np.pi, 200)
x_circle = np.cos(theta_circle)
y_circle = np.sin(theta_circle)

target_radians = [1.0, np.pi/3, np.pi/2, np.pi]
colors = ['red', 'blue', 'green', 'purple']
```

```python
plt.figure(figsize=(8, 8))
plt.plot(x_circle, y_circle, color='gray', linestyle='--', label='Unit Circle (r=1)')
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)


for rad, color in zip(target_radians, colors):
    x = np.cos(rad)
    y = np.sin(rad)


    plt.plot([0, x], [0, y], color=color, linewidth=2,
             label=f'Angle: {rad:.2f} rad ({np.degrees(rad):.1f}°)')
    plt.scatter(x, y, color=color, s=100, zorder=5)


    plt.text(x * 1.1, y * 1.1, f'({x:.2f}, {y:.2f})', color=color, fontsize=10, ha='center')

plt.xlim(-1.5, 1.5)
plt.ylim(-1.5, 1.5)
plt.grid(True, linestyle=':', alpha=0.6)
plt.gca().set_aspect('equal', adjustable='box')

plt.legend(loc='upper right')
plt.show()
```
