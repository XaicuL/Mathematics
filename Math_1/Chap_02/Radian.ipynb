# 호도법 (Radian)

중심각이 호를 **반지름과 같은 길이**로 자를 때의 각을 $1$ 라디안이라 한다.

$$
\theta\text{(rad)} = \dfrac{\ell}{r}, \qquad 180^\circ = \pi\ \text{rad}
$$

| 도 | 라디안 |
|----|--------|
| $0^\circ$ | $0$ |
| $30^\circ$ | $\pi/6$ |
| $45^\circ$ | $\pi/4$ |
| $60^\circ$ | $\pi/3$ |
| $90^\circ$ | $\pi/2$ |
| $180^\circ$ | $\pi$ |
| $360^\circ$ | $2\pi$ |


## 1. 도 ↔ 라디안 변환


```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
sp.init_printing()

# 기호 변환
deg = sp.symbols('theta_deg', real=True)
rad = sp.symbols('theta_rad', real=True)

def deg_to_rad(d):
    return d * sp.pi / 180

def rad_to_deg(r):
    return r * 180 / sp.pi

angles_deg = [0, 30, 45, 60, 90, 180, 360]
print("도 → 라디안")
for d in angles_deg:
    r = deg_to_rad(d)
    print(f"  {d}° = {r} = {float(r):.4f} rad")

print("\n라디안 → 도")
for r in [sp.pi/6, sp.pi/4, sp.pi/3, sp.pi/2, sp.pi, 2*sp.pi]:
    print(f"  {r} = {rad_to_deg(r)}°")
```

**deg → rad**: $\theta_{\mathrm{rad}} = \theta_{\mathrm{deg}} \cdot \pi / 180$

**rad → deg**: $\theta_{\mathrm{deg}} = \theta_{\mathrm{rad}} \cdot 180 / \pi$

**np.deg2rad / np.rad2deg**: <br>
=> NumPy 수치 변환 함수


## 2. 호 길이 $\ell = r\theta$


라디안 $\theta$일 때 호 길이:

$$
\ell = r\,\theta
$$

(단위원 $r=1$이면 $\ell = \theta$)


```python
r, theta = 5, sp.pi / 3  # 반지름 5, 중심각 60°
ell = r * theta
print(f"r={r}, θ=π/3 → 호 길이 ℓ = {ell} = {float(ell):.4f}")

# 역으로: ℓ과 r로 θ
ell2, r2 = 4, 2
theta2 = ell2 / r2
print(f"ℓ={ell2}, r={r2} → θ = {theta2} rad = {rad_to_deg(theta2)}°")
```

## 3. 단위원과 호도


```python
theta_circle = np.linspace(0, 2 * np.pi, 400)
x_c, y_c = np.cos(theta_circle), np.sin(theta_circle)

# 표시할 각
marks = {
    r'$0$': 0,
    r'$\pi/6$': np.pi/6,
    r'$\pi/4$': np.pi/4,
    r'$\pi/3$': np.pi/3,
    r'$\pi/2$': np.pi/2,
    r'$\pi$': np.pi,
}

fig, ax = plt.subplots(figsize=(7, 7))
ax.plot(x_c, y_c, color='gray', ls='--', lw=1.5, label='unit circle $r=1$')
ax.axhline(0, color='lightgray', lw=0.8)
ax.axvline(0, color='lightgray', lw=0.8)

for label, th in marks.items():
    ax.plot([0, np.cos(th)], [0, np.sin(th)], color='steelblue', lw=1.2)
    ax.plot(np.cos(th), np.sin(th), 'o', color='crimson', ms=6)
    ax.annotate(label, (1.12*np.cos(th), 1.12*np.sin(th)), fontsize=11, ha='center')

# 호 강조: 0 → π/3
arc = np.linspace(0, np.pi/3, 50)
ax.plot(np.cos(arc), np.sin(arc), color='crimson', lw=3, label=r'Ark (length$=\theta$)')

ax.set_aspect('equal')
ax.set_xlim(-1.4, 1.4); ax.set_ylim(-1.4, 1.4)
ax.legend(loc='upper right')
ax.grid(True, alpha=0.3)
plt.show()
```

단위원에서 중심각 $\theta$(rad)에 대응하는 **호의 길이가 곧 $\theta$**.
