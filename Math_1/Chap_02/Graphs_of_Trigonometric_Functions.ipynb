# **삼각함수 그래프**

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2 * np.pi, 2 * np.pi, 1000)
y_sin = np.sin(x)
y_cos = np.cos(x)
y_tan = np.tan(x)
```

```python
plt.figure(figsize=(14, 7))

plt.plot(x, y_sin, label='y = sin(x)', color='red', linewidth=2.5)
plt.plot(x, y_cos, label='y = cos(x)', color='blue', linewidth=2.5, linestyle='--')
plt.plot(x, y_tan, label='y = tan(x)', color='green', linewidth=2)


plt.axvline(np.pi/2, color='purple', linestyle=':', alpha=0.5)
plt.axvline(0, color='purple', linestyle=':', alpha=0.5)

asymptotes = [-3*np.pi/2, -np.pi/2, np.pi/2, 3*np.pi/2]
for asympt in asymptotes:
    plt.axvline(asympt, color='gray', linestyle=':', linewidth=1.2, alpha=0.7)

plt.xlabel('x (radians)', fontsize=12)
plt.ylabel('y', fontsize=12)

plt.xticks([-2*np.pi, -3*np.pi/2, -np.pi, -np.pi/2, 0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi],
           ['-2π', '-3π/2', '-π', '-π/2', '0', 'π/2', 'π', '3π/2', '2π'], fontsize=11)
plt.yticks([-3, -2, -1, 0, 1, 2, 3])

plt.xlim(-2 * np.pi, 2 * np.pi)
plt.ylim(-3.5, 3.5)
plt.grid(True, linestyle=':', alpha=0.5)
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.legend(loc='upper right', fontsize=12)

plt.show()

```
