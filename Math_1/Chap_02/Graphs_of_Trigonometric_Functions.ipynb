# 삼각함수의 그래프

$$
y = \sin x,\quad y = \cos x,\quad y = \tan x
$$

| 함수 | 주기 | 진폭 | 정의역 | 점근선 |
|------|------|------|--------|--------|
| $\sin x$ | $2\pi$ | $1$ | $\mathbb{R}$ | 없음 |
| $\cos x$ | $2\pi$ | $1$ | $\mathbb{R}$ | 없음 |
| $\tan x$ | $\pi$ | — | $x \neq \tfrac{\pi}{2}+n\pi$ | $x = \tfrac{\pi}{2}+n\pi$ |


## 1. $\sin x$, $\cos x$ — 주기·진폭


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2 * np.pi, 2 * np.pi, 1000)
y_sin = np.sin(x)
y_cos = np.cos(x)

plt.figure(figsize=(12, 5))
plt.plot(x, y_sin, label=r'$y=\sin x$', color='crimson', lw=2)
plt.plot(x, y_cos, label=r'$y=\cos x$', color='steelblue', lw=2)
plt.axhline(0, color='gray', lw=0.8)
plt.axhline(1, color='lightgray', ls=':', lw=1)
plt.axhline(-1, color='lightgray', ls=':', lw=1)
plt.axvline(0, color='gray', lw=0.8)

# 주기 표시
plt.axvline(0, color='orange', ls='--', alpha=0.5)
plt.axvline(2*np.pi, color='orange', ls='--', alpha=0.5, label=r'period $2\pi$')

plt.xticks([-2*np.pi, -np.pi, 0, np.pi, 2*np.pi],
           [r'$-2\pi$', r'$-\pi$', '0', r'$\pi$', r'$2\pi$'])
plt.ylim(-1.5, 1.5)
plt.title(r'$\sin x$, $\cos x$ (amplitude 1, period $2\pi$)')
plt.legend(); plt.grid(True, alpha=0.3)
plt.show()
```

**진폭:** $|A|$ in $y = A\sin(Bx+C)+D$  
**주기:** $T = 2\pi / |B|$ (sin, cos)

$\cos x = \sin(x + \pi/2)$ — 위상 $\pi/2$ 이동.


## 2. $\tan x$ — 주기 $\pi$, 수직 점근선


```python
# tan은 불연속 → 점근선 근처에서 NaN 처리
x_tan = np.linspace(-2 * np.pi, 2 * np.pi, 2000)
y_tan = np.tan(x_tan)
# 점근선 근처 끊기
y_tan[np.abs(np.cos(x_tan)) < 0.02] = np.nan

plt.figure(figsize=(12, 5))
plt.plot(x_tan, y_tan, label=r'$y=\tan x$', color='seagreen', lw=2)

# 점근선 x = π/2 + nπ
for n in range(-2, 3):
    asym = np.pi/2 + n*np.pi
    plt.axvline(asym, color='gray', ls='--', lw=1, alpha=0.7)

plt.axhline(0, color='gray', lw=0.8)
plt.xticks([-2*np.pi, -np.pi, -np.pi/2, 0, np.pi/2, np.pi, 2*np.pi],
           [r'$-2\pi$', r'$-\pi$', r'$-\pi/2$', '0', r'$\pi/2$', r'$\pi$', r'$2\pi$'])
plt.ylim(-5, 5)
plt.title(r'$\tan x$ (period $\pi$, asymptote $x=\pi/2+n\pi$)')
plt.legend(); plt.grid(True, alpha=0.3)
plt.show()
```

**np.tan**: <br>
=> 라디안 입력. 점근선에서 발산하므로 그래프는 끊어서 그림.


## 3. 변형: $y = A\sin(Bx)$


```python
x = np.linspace(-2*np.pi, 2*np.pi, 1000)

plt.figure(figsize=(12, 5))
plt.plot(x, np.sin(x), label=r'$\sin x$ (A=1,B=1)', lw=1.5, alpha=0.7)
plt.plot(x, 2*np.sin(x), label=r'$2\sin x$ (amplitude 2)', color='crimson', lw=2)
plt.plot(x, np.sin(2*x), label=r'$\sin 2x$ (period $\pi$)', color='steelblue', lw=2)
plt.axhline(0, color='gray', lw=0.8)
plt.ylim(-2.5, 2.5)
plt.legend(); plt.grid(True, alpha=0.3)
plt.show()

print("2*sin(x): 진폭=2, 주기=2π")
print("sin(2x): 진폭=1, 주기=π")
```
