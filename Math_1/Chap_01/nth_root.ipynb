# n제곱근

실수 $a \ge 0$, 자연수 $n$ 에 대해 **n제곱근** $\sqrt[n]{a}$ 는

$$
(\sqrt[n]{a})^n = a, \quad \sqrt[n]{a} \ge 0
$$

을 만족하는 실수이다. (홀수 $n$이면 $a < 0$도 가능)

| 표현 | 의미 |
|------|------|
| $\sqrt[n]{a}$ | $a$의 n제곱근 |
| $a^{1/n}$ | 지수 표기 (같은 값) |
| $\sqrt{a}$ | 제곱근 ($n=2$) |


## 1. 정의와 지수 표기 $a^{1/n}$

```python
import sympy as sp
sp.init_printing()

a, n = sp.symbols('a n', positive=True, integer=True)
x = sp.symbols('x', real=True)

# √[n]{a} = a**(1/n)
expr_root = sp.root(16, 4)
expr_pow = sp.Pow(16, sp.Rational(1, 4))

print("16의 네제곱근:")
display(expr_root)
display(expr_pow)
print("수치:", float(expr_root))

# (√[n]{a})^n = a
print("검증: (16^(1/4))^4 =", expr_root**4)
```

**sp.root(a, n)**: <br>
=> $\sqrt[n]{a}$ 계산

**sp.Pow(a, sp.Rational(1, n))**: <br>
=> 지수 표기 $a^{1/n}$ (유리수 지수로 정확히 유지)

**sp.Rational(p, q)**: <br>
=> 유리수 $p/q$ (부동소수 오차 방지)


## 2. n제곱근의 성질

$a, b \ge 0$, 자연수 $n, m$ 에 대해:

1. $\sqrt[n]{a} \cdot \sqrt[n]{b} = \sqrt[n]{ab}$
2. $\dfrac{\sqrt[n]{a}}{\sqrt[n]{b}} = \sqrt[n]{\dfrac{a}{b}}$ $(b \neq 0)$
3. $\sqrt[n]{\sqrt[m]{a}} = \sqrt[nm]{a}$
4. $(\sqrt[n]{a})^m = \sqrt[n]{a^m} = a^{m/n}$


```python
# 성질 검증 (수치 예)
A, B = 16, 81

print("1. 곱:", sp.root(A, 4) * sp.root(B, 4), "==", sp.root(A * B, 4))
print("2. 몫:", sp.root(A, 4) / sp.root(B, 4), "==", sp.root(sp.Rational(A, B), 4))
print("3. 중첩:", sp.root(sp.root(A, 2), 2), "==", sp.root(A, 4))
print("4. 거듭제곱:", sp.root(A, 4)**2, "==", sp.root(A**2, 4), "==", sp.Pow(A, sp.Rational(2, 4)))
```

## 3. 지수법칙과의 관계

무리수 지수로 지수를 확장하면:

$$
a^{m/n} = \sqrt[n]{a^m} = (\sqrt[n]{a})^m \quad (a > 0)
$$

특히 $a^{1/n} = \sqrt[n]{a}$.


```python
# a^(m/n) 예제
a_val = 8
print("8^(2/3) =", sp.Pow(a_val, sp.Rational(2, 3)))
print("검증: (∛8)^2 =", sp.root(a_val, 3)**2)
print("검증: ∛(8^2) =", sp.root(a_val**2, 3))

# 기호: a^(m/n)
m = sp.symbols('m', integer=True, positive=True)
display(sp.Eq(sp.Pow(a, m / n), sp.root(a, n)**m))
display(sp.Eq(sp.root(a**m, n), sp.Pow(a, m / n)))
```

## 4. 예제: sympy로 간단히

```python
examples = [
    (27, 3),   # ∛27 = 3
    (32, 5),   # ⁵√32 = 2
    (81, 4),   # ⁴√81 = 3
    (1, 7),    # ⁷√1 = 1
]

for val, k in examples:
    r = sp.root(val, k)
    print(f"√[{k}]{{{val}}} = {r}  (수치: {float(r)})")
```
