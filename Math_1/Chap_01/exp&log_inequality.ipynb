# 지수·로그 부등식

밑의 크기에 따라 **부등호 방향**이 달라진다.

| 밑 | 지수부등식 $a^x > a^k$ | 로그부등식 $\log_a x > \log_a k$ |
|----|------------------------|-------------------------------------|
| $a > 1$ | $x > k$ (방향 유지) | $x > k$ (방향 유지, $x>0$) |
| $0 < a < 1$ | $x < k$ (방향 **반전**) | $x < k$ (방향 **반전**, $x>0$) |


## 1. 지수부등식 — 밑 $> 1$


```python
import sympy as sp
from sympy import solve_univariate_inequality, S
sp.init_printing()

x = sp.symbols('x', real=True)

# 2^x > 8 = 2^3  →  x > 3  (밑 2 > 1)
ineq1 = 2**x > 8
sol1 = solve_univariate_inequality(ineq1, x, relational=False)
print("2^x > 8")
display(sol1)

# 2^x ≤ 1/4 = 2^(-2)  →  x ≤ -2
ineq2 = 2**x <= sp.Rational(1, 4)
sol2 = solve_univariate_inequality(ineq2, x, relational=False)
print("2^x ≤ 1/4")
display(sol2)
```

**solve_univariate_inequality(ineq, x)**: <br>
=> 일변수 부등식의 해집합

**relational=False**: <br>
=> `Interval` / `Union` 형태로 반환


## 2. 지수부등식 — $0 <$ 밑 $< 1$ (부등호 반전)


```python
# (1/2)^x > 1/8 = (1/2)^3
# 밑 1/2 < 1 이므로 방향 반전: x < 3
ineq3 = (sp.Rational(1, 2))**x > sp.Rational(1, 8)
sol3 = solve_univariate_inequality(ineq3, x, relational=False)
print("(1/2)^x > 1/8  →  기대: x < 3")
display(sol3)

# 검산: x=0 → (1/2)^0=1 > 1/8 ✓ / x=4 → 1/16 < 1/8 ✗
```

## 3. 로그부등식 — 밑 $> 1$


```python
# log2(x) > 3  →  x > 8  (단 x > 0)
ineq4 = sp.log(x, 2) > 3
sol4 = solve_univariate_inequality(ineq4, x, relational=False)
print("log2(x) > 3")
display(sol4)

# log2(x) < 1  →  0 < x < 2
ineq5 = sp.log(x, 2) < 1
sol5 = solve_univariate_inequality(ineq5, x, relational=False)
print("log2(x) < 1")
display(sol5)
```

## 4. 로그부등식 — $0 <$ 밑 $< 1$ (부등호 반전)


```python
# log_{1/2}(x) > -2
# 밑 < 1 → 방향 반전: 0 < x < (1/2)^(-2) = 4
base = sp.Rational(1, 2)
ineq6 = sp.log(x, base) > -2
sol6 = solve_univariate_inequality(ineq6, x, relational=False)
print("log_{1/2}(x) > -2  →  기대: 0 < x < 4")
display(sol6)
```

## 5. 비교 요약 (같은 형태, 다른 밑)


```python
cases = [
    ("2^x > 4", 2**x > 4),
    ("(1/2)^x > 1/4", (sp.Rational(1, 2))**x > sp.Rational(1, 4)),
    ("log2(x) > 2", sp.log(x, 2) > 2),
    ("log_{1/2}(x) > -1", sp.log(x, sp.Rational(1, 2)) > -1),
]

for name, ineq in cases:
    sol = solve_univariate_inequality(ineq, x, relational=False)
    print(f"{name}  →  {sol}")
```

**핵심:** 밑 $>1$이면 함수 증가 → 부등호 유지.  
$0<$ 밑 $<1$이면 함수 감소 → 부등호 **반전**.  
로그는 항상 진수 $> 0$ (및 밑 조건)을 함께 적는다.
