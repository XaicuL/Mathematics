# 로그

$a > 0$, $a \neq 1$, $x > 0$ 일 때

$$
\log_a x = y \quad \Longleftrightarrow \quad a^y = x
$$

| 기호 | 의미 |
|------|------|
| $\log_a x$ | 밑 $a$, 진수 $x$ |
| $\ln x$ | 자연로그 ($\log_e x$) |
| $\log x$ | 상용로그 ($\log_{10} x$), 문맥에 따라 $\ln$ |


## 1. 정의: 지수의 역

```python
import sympy as sp
sp.init_printing()

a, x, y = sp.symbols('a x y', positive=True)

# 2^y = 9 → y = log2(9)
y_val = sp.log(9, 2)
print("2^y = 9 → y =")
display(y_val)
print("수치:", float(y_val))
print("검증 2^y =", float(2**y_val))

# 기본값
print("log2(8) =", sp.log(8, 2))
print("log3(81) =", sp.log(81, 3))
print("ln(e^5) =", sp.log(sp.E**5))
```

**sp.log(x, a)**: <br>
=> $\log_a x$ (두 번째 인자가 밑). 생략 시 자연로그 $\ln x$

**sp.E**: <br>
=> 자연상수 $e$


## 2. 로그의 성질

$M, N > 0$, 실수 $k$ 에 대해 ($a > 0,\, a \neq 1$):

1. $\log_a(MN) = \log_a M + \log_a N$
2. $\log_a\dfrac{M}{N} = \log_a M - \log_a N$
3. $\log_a(M^k) = k\,\log_a M$
4. $\log_a 1 = 0$, $\log_a a = 1$


```python
M, N = 4, 2
base = 2

lg = lambda t: sp.log(t, base)

print("1. 곱:", lg(M) + lg(N), "==", lg(M * N))
print("2. 몫:", lg(M) - lg(N), "==", lg(sp.Rational(M, N)))
print("3. 거듭제곱:", 3 * lg(M), "==", lg(M**3))
print("4. 기본:", lg(1), ",", lg(base))
```

## 3. 밑 변환 공식

$$
\log_a b = \dfrac{\log_c b}{\log_c a} = \dfrac{1}{\log_b a}
$$

특히 $\log_a b = \dfrac{\ln b}{\ln a}$.


```python
# 밑 변환: log2(9) = ln9 / ln2
lhs = sp.log(9, 2)
rhs = sp.log(9) / sp.log(2)
print("log2(9) =", lhs)
print("ln9/ln2 =", rhs)
print("같음?", sp.simplify(lhs - rhs) == 0)

# log_a b * log_b a = 1
print("log2(5) * log5(2) =", sp.simplify(sp.log(5, 2) * sp.log(2, 5)))
```

**sp.simplify(...)**: <br>
=> 식 단순화. 밑 변환 항등식 검증에 유용


## 4. 예제

```python
# log2(4) + log2(8) = log2(32) = 5
print(sp.log(4, 2) + sp.log(8, 2), "==", sp.log(32, 2))

# 3*log5(2) = log5(8)
print(3 * sp.log(2, 5), "==", sp.log(8, 5))

# 밑변환으로 log4(8) 계산
print("log4(8) =", sp.log(8, 4), "=", sp.simplify(sp.log(8) / sp.log(4)))
```
