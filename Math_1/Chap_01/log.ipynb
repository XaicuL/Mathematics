# **log**

```python
import math

#로그의 탄생 배경
re_1 = math.pow(2, 2)
re_2 = math.pow(2, 3)

#위와 같은 구조로,
# 2^x = 4 일 때 x 는?
# 2^x = 8 일 때 x 는?
# 이러한 질문들을 던지던 와중 문득 이러한 수가 나온다. 2^y = 9 일 때, y 는 ?
# 우리가 기존에 알던 지수로서의 표현으로는 나타내기가 어려운데, 이때 도입하게 된 것이 log이다.
```

```python
relg_1 = math.log2(9)
print(f'2^y = 9 일 때, y 는 ? {relg_1}')
```

```python
#log의 성질

relg_a = math.log2(4)
relg_b = math.log2(2)

#(1)

plus = relg_a + relg_b
print(f'log2(4) + log2(2) = {plus}')

#(2)
minus = relg_a - relg_b
print(f'log2(4) - log2(2) = {minus}')

#(3)
mul = relg_a * 3
print(f'log2(4) * 3 = {mul}')

#(4)
div = relg_a / 3
print(f'log2(4) / 3 = {div}')
```
