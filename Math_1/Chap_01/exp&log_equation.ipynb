# 지수·로그 방정식

밑이 같은 지수식·로그식을 정리하거나, 치환으로 다항방정식으로 바꾼다.

**정의역 주의:** 로그 진수 $> 0$, 밑 $a > 0,\, a \neq 1$.


## 1. 지수방정식 (밑 맞추기)


```python
import sympy as sp
sp.init_printing()

x = sp.symbols('x', real=True)

eq1 = sp.Eq(2**x, 16)
eq2 = sp.Eq(3**(x + 1), 27)
eq3 = sp.Eq(4**x, 8**(x - 1))  # (2^2)^x = (2^3)^(x-1)

for eq in [eq1, eq2, eq3]:
    sol = sp.solve(eq, x)
    print("=" * 40)
    display(eq)
    print("해:", sol)
```

**sp.Eq(lhs, rhs)**: <br>
=> 방정식 객체

**sp.solve(eq, x)**: <br>
=> 대수적 해 구하기


## 2. 치환: $4^x - 3\cdot 2^x + 2 = 0$


$4^x = (2^x)^2$ 이므로 $t = 2^x$ $(t > 0)$ 로 치환:

$$
t^2 - 3t + 2 = 0 \implies (t-1)(t-2)=0
$$


```python
t = sp.symbols('t', positive=True)

# 치환 방정식
quad = t**2 - 3*t + 2
t_sols = sp.solve(quad, t)
print("t 해:", t_sols)

# t = 2^x → x
for tv in t_sols:
    xs = sp.solve(sp.Eq(2**x, tv), x)
    print(f"2^x = {tv} → x = {xs}")

# 한 번에
eq = 4**x - 3*(2**x) + 2
print("직접 solve:", sp.solve(eq, x))
```

## 3. 로그방정식


```python
eq_log1 = sp.Eq(sp.log(x, 2), 5)       # → x = 32
eq_log2 = sp.Eq(sp.log(x - 1, 3), 2)  # → x = 10 (진수 x-1 > 0)
eq_log3 = sp.Eq(sp.log(x, 2) + sp.log(x - 2, 2), 3)  # log2(x(x-2))=3

domain = {
    id(eq_log1): lambda s: s > 0,
    id(eq_log2): lambda s: s > 1,
    id(eq_log3): lambda s: s > 2,
}

for eq in [eq_log1, eq_log2, eq_log3]:
    sol = sp.solve(eq, x)
    ok = [s for s in sol if s.is_real and domain[id(eq)](s)]
    print("=" * 40)
    display(eq)
    print("후보:", sol, "→ 정의역 통과:", ok)
```

로그 합: $\log_a M + \log_a N = \log_a(MN)$ 이용 후 $a^{\text{우변}}$으로 지수화.
진수 조건 $x > 0$, $x-2 > 0$ 등 **반드시 검산**.


## 4. 예제 정리


```python
problems = [
    ("2^x = 16", sp.Eq(2**x, 16)),
    ("3^(x+1) = 27", sp.Eq(3**(x + 1), 27)),
    ("4^x - 3*2^x + 2 = 0", 4**x - 3*(2**x) + 2),
    ("log2(x) = 5", sp.Eq(sp.log(x, 2), 5)),
    ("log3(x-1) = 2", sp.Eq(sp.log(x - 1, 3), 2)),
]

for name, eq in problems:
    print(f"{name} → x = {sp.solve(eq, x)}")
```
