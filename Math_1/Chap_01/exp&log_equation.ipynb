# **지수&로그 방정식**

```python
from sympy import symbols, Eq, solve

x = symbols('x')

eq1 = Eq(2**x, 16)
sol1 = solve(eq1, x)

print("1. Equation: 2^x = 16")
print(f"Solution: x = {sol1}")


eq2 = Eq(3**(x + 1), 27)
sol2 = solve(eq2, x)

print("2. Equation: 3^(x + 1) = 27")
print(f"Solution: x = {sol2}")


eq3 = 4**x - 3*(2**x) + 2
sol3 = solve(eq3, x)

print("3. Equation: 4^x - 3*(2^x) + 2 = 0")
print(f"Solution: x = {sol3}")
```

```python
from sympy import symbols, log, Eq, solve

x = symbols('x')

eq1 = Eq(2**x, 16)
sol1 = solve(eq1, x)

print("1. Equation: 2^x = 16")
print(f"Solution: x = {sol1}")


eq2 = 4**x - 3*(2**x) + 2
sol2 = solve(eq2, x)

print("2. Equation: 4^x - 3*(2^x) + 2 = 0")
print(f"Solution: x = {sol2}")


eq3 = Eq(log(x, 2), 5)
sol3 = solve(eq3, x)

print("3. Equation: log2(x) = 5")
print(f"Solution: x = {sol3}")


eq4 = Eq(log(x - 1, 3), 2)
sol4 = solve(eq4, x)

print("4. Equation: log3(x - 1) = 2")
print(f"Solution: x = {sol4}")
```
