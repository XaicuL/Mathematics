# 지수함수와 로그함수

$$
y = a^x \quad (a > 0,\, a \neq 1), \qquad y = \log_a x
$$

두 함수는 서로 **역함수**이며, 그래프는 직선 $y = x$ 에 대해 대칭이다.


## 1. 지수함수 $y = a^x$


| 밑 | 증감 | 점근선 | 치역 |
|----|------|--------|------|
| $a > 1$ | 증가 | $y = 0$ | $(0, \infty)$ |
| $0 < a < 1$ | 감소 | $y = 0$ | $(0, \infty)$ |

공통: 정의역 $\mathbb{R}$, $a^0 = 1$, 항상 $y > 0$.


```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

x = np.linspace(-3, 3, 400)

plt.figure(figsize=(8, 5))
plt.plot(x, 2**x, label=r'$y=2^x$ $(a>1)$', color='crimson', lw=2)
plt.plot(x, (0.5)**x, label=r'$y=(1/2)^x$ $(0<a<1)$', color='steelblue', lw=2)
plt.axhline(0, color='gray', ls='--', lw=1, label='asymptote $y=0$')
plt.axhline(1, color='lightgray', ls=':', lw=1)
plt.axvline(0, color='lightgray', ls=':', lw=1)
plt.ylim(-0.5, 8)
plt.xlabel('x'); plt.ylabel('y')
plt.title('exponential function')
plt.legend(); plt.grid(True, alpha=0.3)
plt.show()
```

## 2. 로그함수 $y = \log_a x$


| 밑 | 증감 | 점근선 | 정의역 |
|----|------|--------|--------|
| $a > 1$ | 증가 | $x = 0$ | $(0, \infty)$ |
| $0 < a < 1$ | 감소 | $x = 0$ | $(0, \infty)$ |

공통: 치역 $\mathbb{R}$, $\log_a 1 = 0$, $\log_a a = 1$.


```python
x_pos = np.linspace(0.05, 8, 400)

plt.figure(figsize=(8, 5))
plt.plot(x_pos, np.log2(x_pos), label=r'$y=\log_2 x$ $(a>1)$', color='crimson', lw=2)
plt.plot(x_pos, np.log(x_pos) / np.log(0.5),
         label=r'$y=\log_{1/2} x$ $(0<a<1)$', color='steelblue', lw=2)
plt.axvline(0, color='gray', ls='--', lw=1, label='asymptote $x=0$')
plt.axhline(0, color='lightgray', ls=':', lw=1)
plt.axvline(1, color='lightgray', ls=':', lw=1)
plt.ylim(-4, 4)
plt.xlabel('x'); plt.ylabel('y')
plt.title('logarithmic function')
plt.legend(); plt.grid(True, alpha=0.3)
plt.show()
```

**np.log2 / np.log**: <br>
=> 수치 로그. 임의 밑: $\log_a x = \ln x / \ln a$

**plt.axhline / axvline**: <br>
=> 수평·수직 점근선 표시


## 3. 역함수 관계 (대칭)


```python
x_exp = np.linspace(-3, 3, 300)
y_exp = 2**x_exp
x_log = np.linspace(0.05, 8, 300)
y_log = np.log2(x_log)
t = np.linspace(-1, 8, 100)

plt.figure(figsize=(7, 7))
plt.plot(x_exp, y_exp, label=r'$y=2^x$', color='crimson', lw=2)
plt.plot(x_log, y_log, label=r'$y=\log_2 x$', color='steelblue', lw=2)
plt.plot(t, t, 'k--', lw=1, alpha=0.6, label=r'$y=x$')
plt.axhline(0, color='gray', ls=':', lw=0.8)
plt.axvline(0, color='gray', ls=':', lw=0.8)
plt.xlim(-3, 8); plt.ylim(-3, 8)
plt.gca().set_aspect('equal', adjustable='box')
plt.title(r'inverse function: $y=2^x$ ↔ $y=\log_2 x$')
plt.legend(); plt.grid(True, alpha=0.3)
plt.show()

x_sym = sp.symbols('x', positive=True)
print("log2(2^x) =", sp.simplify(sp.log(2**x_sym, 2)))
print("2^(log2 x) =", sp.simplify(2**sp.log(x_sym, 2)))
```

합성: $\log_a(a^x) = x$, $a^{\log_a x} = x$ $(x > 0)$.
그래프는 $y=x$에 대해 대칭.
