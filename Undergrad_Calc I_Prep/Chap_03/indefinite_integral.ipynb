# 부정적분

$F'(x)=f(x)$ 일 때 $F$를 $f$의 **부정적분**이라 하고

$$
\int f(x)\,dx = F(x)+C
$$

($C$는 적분상수). 미분의 역연산이다.

## 1. 부정적분의 정의


```python
import sympy as sp
sp.init_printing()

x = sp.symbols('x')
C = sp.symbols('C')

f = x**2 + 3*x + 2
F = sp.integrate(f, x)

print("f(x) =")
display(f)
print("∫ f(x) dx =")
display(F + C)

print("검산 F'(x) = f(x)?")
display(sp.diff(F, x))
print(sp.diff(F, x) == f)
```

**sp.integrate(f, x)**: <br>
=> $x$에 대한 부정적분 (적분상수는 보통 생략)

**sp.diff(F, x)**: <br>
=> 미분해서 원래 함수로 돌아오는지 검산


## 시각화: 적분상수 $C$ — 곡선족

$F(x)+C$는 위아래로 평행이동한다. 기울기(도함수)는 모두 같다.

```python
import numpy as np
import matplotlib.pyplot as plt

f_num = sp.lambdify(x, f, "numpy")
F_num = sp.lambdify(x, F, "numpy")
xs = np.linspace(-3.5, 1.2, 400)

fig, axes = plt.subplots(1, 2, figsize=(10, 3.8))
axes[0].plot(xs, f_num(xs))
axes[0].axhline(0, color="gray", lw=0.8)
axes[0].axvline(0, color="gray", lw=0.8)
axes[0].set_title(r"$f(x)=x^2+3x+2$")
axes[0].grid(True, alpha=0.3)

for Cval in [-2, -1, 0, 1, 2]:
    axes[1].plot(xs, F_num(xs) + Cval, label=f"$C={Cval}$")
axes[1].axhline(0, color="gray", lw=0.8)
axes[1].axvline(0, color="gray", lw=0.8)
axes[1].set_title(r"$F(x)+C$")
axes[1].legend(fontsize=8)
axes[1].grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 2. 부정적분의 계산

선형성:

$$
\int \bigl(af(x)+bg(x)\bigr)\,dx = a\int f + b\int g
$$


```python
examples = [
    3*x**2 - 2*x + 1,
    sp.sin(x) + sp.cos(x),
    1/x,
    sp.exp(x),
]

for expr in examples:
    print("=" * 40)
    display(expr)
    display(sp.integrate(expr, x) + C)
```

## 3. 여러 가지 함수의 부정적분

| $f(x)$ | $\int f(x)\,dx$ |
|--------|------------------|
| $x^n\ (n\neq -1)$ | $\dfrac{x^{n+1}}{n+1}+C$ |
| $1/x$ | $\ln\|x\|+C$ |
| $e^x$ | $e^x+C$ |
| $\sin x$ | $-\cos x+C$ |
| $\cos x$ | $\sin x+C$ |
| $\sec^2 x$ | $\tan x+C$ |
| $1/\sqrt{1-x^2}$ | $\arcsin x+C$ |
| $1/(1+x^2)$ | $\arctan x+C$ |


```python
table = [
    x**4,
    1/x,
    sp.exp(x),
    sp.sin(x),
    sp.cos(x),
    sp.sec(x)**2,
    1/sp.sqrt(1 - x**2),
    1/(1 + x**2),
]

for expr in table:
    print("=" * 40)
    display(expr)
    display(sp.simplify(sp.integrate(expr, x)) + C)
```
