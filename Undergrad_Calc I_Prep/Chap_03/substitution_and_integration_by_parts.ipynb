# 치환적분 & 부분적분

합성함수의 미분 · 곱의 미분을 거꾸로 한 기법.

| 기법 | 대응 |
|------|------|
| 치환적분 | 합성함수의 미분 (연쇄법칙) |
| 부분적분 | 곱의 미분 |

## 1. 치환적분법

$u=g(x)$ 로 놓으면

$$
\int f\bigl(g(x)\bigr)g'(x)\,dx = \int f(u)\,du
$$

예: $\int 2x(x^2+1)^3\,dx$


```python
import sympy as sp
sp.init_printing()

x = sp.symbols('x')
u = sp.symbols('u')
C = sp.symbols('C')

expr = 2*x * (x**2 + 1)**3
print("원래 식:")
display(expr)

# 치환: u = x²+1, du = 2x dx
print("치환 u = x²+1 후:")
display(sp.integrate(u**3, u) + C)

print("다시 x로 (직접 적분과 비교):")
direct = sp.integrate(expr, x)
display(direct + C)
display(sp.expand(direct - (x**2 + 1)**4 / 4))
```

```python
more = [
    sp.cos(x) * sp.sin(x)**4,          # u = sin x
    sp.exp(x) / (1 + sp.exp(x)),       # u = 1+e^x
    x / sp.sqrt(x**2 + 1),             # u = x²+1
]

for expr in more:
    print("=" * 40)
    display(expr)
    display(sp.simplify(sp.integrate(expr, x)) + C)
```

## 2. 부분적분법

$$
\int u\,dv = uv - \int v\,du
$$

예: $\int x e^x\,dx$  ($u=x$, $dv=e^x dx$)


```python
expr = x * sp.exp(x)
print("∫ x e^x dx =")
display(sp.integrate(expr, x) + C)

# 수동 부분적분과 비교
u_fun, dv = x, sp.exp(x)
du, v = sp.diff(u_fun, x), sp.integrate(dv, x)
manual = u_fun*v - sp.integrate(v*du, x)
print("수동: uv − ∫ v du")
display(sp.simplify(manual) + C)
```

**LIATE 우선순위** (보통 $u$로 고르는 쪽): <br>
=> Log / Inverse trig / Algebraic / Trig / Exponential


```python
parts_ex = [
    x * sp.sin(x),
    x**2 * sp.exp(x),
    sp.log(x),
    sp.exp(x) * sp.sin(x),
]

for expr in parts_ex:
    print("=" * 40)
    display(expr)
    display(sp.simplify(sp.integrate(expr, x)) + C)
```
