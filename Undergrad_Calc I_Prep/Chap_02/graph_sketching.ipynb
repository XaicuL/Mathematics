# 3. 그래프의 개형

$f'$, $f''$ 부호로 **증감·볼록**을 파악해 개형을 그린다.

- $f'>0$: 증가 / $f'<0$: 감소
- $f''>0$: 아래로 볼록 / $f''<0$: 위로 볼록


## 1. Example


```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
sp.init_printing()
x = sp.symbols('x', real=True)

f = x**3 - 3*x**2 + 2
f1 = sp.diff(f, x)
f2 = sp.diff(f, x, 2)
crit = sp.solve(f1, x)
inflect = sp.solve(f2, x)

print("f'(x):"); display(f1)
print("f''(x):"); display(f2)
print("임계점:", crit, "변곡점:", inflect)

xs = np.linspace(-0.5, 3.5, 400)
y = [float(f.subs(x,t)) for t in xs]
plt.plot(xs, y, 'b', lw=2, label='f')
for c in crit:
    plt.axvline(float(c), color='r', ls=':', alpha=0.5)
for p in inflect:
    plt.plot(float(p), float(f.subs(x,p)), 'gs', ms=8)
plt.grid(True, alpha=0.3); plt.legend(); plt.title('Graph Sketching')
plt.show()
```

**개형 분석**: <br>
=> $f'$, $f''$ 부호표 + 임계점·변곡점 표시
