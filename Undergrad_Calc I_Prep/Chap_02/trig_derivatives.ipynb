# 1. 삼각함수의 도함수

$$
(\sin x)' = \cos x,\quad (\cos x)' = -\sin x,\quad (\tan x)' = \sec^2 x
$$


## 1. 기본 공식


```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
sp.init_printing()
x = sp.symbols('x', real=True)

trig = [sp.sin(x), sp.cos(x), sp.tan(x), sp.sin(2*x), sp.cos(x**2)]
for expr in trig:
    display(sp.Eq(sp.diff(expr, x), sp.diff(expr, x)))
```

**sp.diff(sp.sin(2*x), x)**: <br>
=> 연쇄법칙: $2\cos(2x)$


## 2. 시각화


```python
x_vals = np.linspace(-2*np.pi, 2*np.pi, 400)
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].plot(x_vals, np.sin(x_vals), label='sin x')
ax[0].plot(x_vals, np.cos(x_vals), label="sin' = cos")
ax[0].legend(); ax[0].grid(True, alpha=0.3); ax[0].set_title('sin & derivative')
ax[1].plot(x_vals, np.cos(x_vals), label='cos x')
ax[1].plot(x_vals, -np.sin(x_vals), label="cos' = -sin")
ax[1].legend(); ax[1].grid(True, alpha=0.3); ax[1].set_title('cos & derivative')
plt.tight_layout(); plt.show()
```
