# 1. 미분계수 & 도함수

$$
f'(x) = \lim_{h\to 0}\frac{f(x+h)-f(x)}{h}
$$

미분계수는 접선의 기울기, 순간 변화율이다.


## 1. 도함수 계산


```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

sp.init_printing()
x, h = sp.symbols('x h', real=True)

f = x**3 - 3*x**2 + 2

# 정의로 미분계수 (x=1)
f1_def = sp.limit((f.subs(x, x + h) - f) / h, h, 0)
f1 = sp.diff(f, x)
f2 = sp.diff(f, x, 2)

print("f(x):"); display(f)
print("정의 (극한):"); display(f1_def.subs(x, 1))
print("f'(x):"); display(f1)
print("f'(1) =", f1.subs(x, 1))
print("f''(x):"); display(f2)
```

**sp.limit(expr, h, 0)**: <br>
=> $h\to 0$ 극한

**sp.diff(f, x)**: <br>
=> $f$를 $x$로 미분

**sp.diff(f, x, 2)**: <br>
=> 2계 도함수
