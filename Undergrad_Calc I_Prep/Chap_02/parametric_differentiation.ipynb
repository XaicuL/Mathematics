# 5. 매개변수로 나타낸 함수의 미분법

$x=f(t)$, $y=g(t)$ 이면 $\dfrac{dy}{dx} = \dfrac{dy/dt}{dx/dt}$ ($dx/dt\neq 0$).


## 1. 매개변수 미분


```python
import sympy as sp
sp.init_printing()
t = sp.symbols('t', real=True)

x_t = sp.cos(t)
y_t = sp.sin(t)

dx_dt = sp.diff(x_t, t)
dy_dt = sp.diff(y_t, t)
dy_dx = sp.simplify(dy_dt / dx_dt)

print("x(t), y(t):")
display(x_t, y_t)
print("dy/dx = (dy/dt)/(dx/dt):")
display(dy_dx)

print("원 x²+y²=1 에서 dy/dx = -x/y:")
display(sp.simplify(-x_t / y_t))
print(f"일치?: {sp.simplify(dy_dx - (-x_t/y_t)) == 0}")
```

**매개변수 미분**: <br>
=> $\dfrac{dy}{dx}=\dfrac{y'(t)}{x'(t)}$. 2계는 $\dfrac{d}{dt}(dy/dx)\big/ x'(t)$.
