# 4. 음함수 & 역함수의 미분법

$F(x,y)=0$ 일 때 $\dfrac{dy}{dx} = -F_x/F_y$. 역함수 $g=f^{-1}$이면 $g'(y)=1/f'(g(y))$.


## 4. 음함수 미분법

$F(x,y)=0$ 일 때 $x$ 로 양변 미분:

$$
\frac{dy}{dx} = -\frac{F_x}{F_y}\quad (F_y \neq 0)
$$


```python
import sympy as sp
sp.init_printing()
x = sp.symbols('x', real=True)

y = sp.symbols('y', real=True)

F = x**2 + y**2 - 1

print("F(x,y) = 0:")
display(sp.Eq(F, 0))

dydx_idiff = sp.idiff(F, y, x)
print("sp.idiff:")
display(dydx_idiff)

dydx_formula = -sp.diff(F, x) / sp.diff(F, y)
print("-F_x/F_y:")
display(sp.simplify(dydx_formula))

print(f"일치?: {sp.simplify(dydx_idiff - dydx_formula) == 0}")
```

**sp.idiff(F, y, x)**: <br>
=> $F(x,y)=0$ 에서 $dy/dx$ 를 음함수 미분으로 계산


## 5. 역함수 미분법

$g = f^{-1}$ 이면

$$
g'(y) = \frac{1}{f'(g(y))}
$$


```python
import sympy as sp
sp.init_printing()
x = sp.symbols('x', real=True)

y = sp.symbols('y', real=True)

f = x**3
f_prime = sp.diff(f, x)
g = y**sp.Rational(1, 3)
g_prime_direct = sp.diff(g, y)
g_prime_formula = 1 / f_prime.subs(x, g)

print("f(x) = x³, 역함수 g(y) = y^(1/3)")
print("직접 미분 g':")
display(sp.simplify(g_prime_direct))
print("공식 1/f'(g(y)):")
display(sp.simplify(g_prime_formula))
print(f"일치?: {sp.simplify(g_prime_direct - g_prime_formula) == 0}")
```

**요지**: <br>
=> 역함수 미분은 원래 함수 미분의 역수. $f'(x_0)\neq 0$ 필요.
