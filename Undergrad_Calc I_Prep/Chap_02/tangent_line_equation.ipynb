# 2. 접선의 방정식

점 $(a,f(a))$에서 접선:

$$
y - f(a) = f'(a)(x-a)
$$


## 1. Example


```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
sp.init_printing()
x = sp.symbols('x', real=True)

f = x**3 - 3*x
a = 1
fa = f.subs(x, a)
m = sp.diff(f, x).subs(x, a)
tangent = m*(x-a) + fa

print(f"a={a}, f(a)={fa}, f'(a)={m}")
print("접선:"); display(sp.Eq(sp.Symbol('y'), sp.expand(tangent)))

xs = np.linspace(-2, 2.5, 300)
plt.plot(xs, [float(f.subs(x,t)) for t in xs], 'b', label='f(x)')
plt.plot(xs, [float(tangent.subs(x,t)) for t in xs], 'r--', label='tangent')
plt.plot(a, float(fa), 'ko')
plt.grid(True, alpha=0.3); plt.legend(); plt.title('tangent line equation')
plt.show()
```

**접선**: <br>
=> $y=f(a)+f'(a)(x-a)$
