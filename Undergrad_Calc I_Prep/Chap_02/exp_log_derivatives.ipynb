# 2. 지수함수 & 로그함수의 도함수

$$
(e^x)' = e^x,\quad (a^x)' = a^x \ln a,\quad (\ln x)' = \frac{1}{x}
$$


## 1. 기본 공식


```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
sp.init_printing()
x, a = sp.symbols('x a', positive=True, real=True)

for expr in [sp.exp(x), sp.log(x), a**x]:
    display(sp.Eq(sp.diff(expr, x), sp.diff(expr, x)))

# a=2 예시
print("2^x 의 도함수:"); display(sp.diff(2**x, x))
```

**sp.exp(x)**, **sp.log(x)**: <br>
=> 자연지수·자연로그


## 2. 시각화


```python
x_vals = np.linspace(0.1, 3, 300)
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].plot(x_vals, np.exp(x_vals), label='e^x')
ax[0].plot(x_vals, np.exp(x_vals), '--', alpha=0.6, label="(e^x)' = e^x")
ax[0].set_title('e^x'); ax[0].legend(); ax[0].grid(True, alpha=0.3)
ax[1].plot(x_vals, np.log(x_vals), label='ln x')
ax[1].plot(x_vals, 1/x_vals, '--', alpha=0.6, label="(ln x)' = 1/x")
ax[1].set_title('ln x'); ax[1].legend(); ax[1].grid(True, alpha=0.3)
plt.tight_layout(); plt.show()
```
