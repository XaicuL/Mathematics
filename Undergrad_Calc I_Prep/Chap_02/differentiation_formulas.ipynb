# 2. 미분법의 공식

| 함수 | 도함수 |
|------|--------|
| $x^n$ | $nx^{n-1}$ |
| $cf(x)$ | $cf'(x)$ |
| $f\pm g$ | $f'\pm g'$ |
| $f\cdot g$ | $f'g+fg'$ |


## 1. 기본 함수의 도함수


```python
import sympy as sp
sp.init_printing()
x = sp.symbols('x', real=True)

basics = [x**5, sp.sin(x), sp.cos(x), sp.exp(x), sp.log(x)]
for expr in basics:
    display(sp.Eq(sp.Derivative(expr, x), sp.diff(expr, x)))
```

**sp.Derivative(expr, x)**: <br>
=> 미분 기호만 표시 (계산은 `sp.diff`)


import sympy as sp
sp.init_printing()
x = sp.symbols('x', real=True)

f = x**2 + 2
g = x + 1
y = f / g

f_p, g_p = sp.diff(f, x), sp.diff(g, x)
quot_rule = (f_p*g - f*g_p) / g**2
direct = sp.diff(y, x)

print("y = f/g:"); display(y)
print("몫의 미분법:"); display(sp.simplify(quot_rule))
print("직접 미분:"); display(sp.simplify(direct))
print(f"일치?: {sp.simplify(quot_rule - direct) == 0}")


```python
f = x**2 + 2
g = x + 1
y = f / g

f_p, g_p = sp.diff(f, x), sp.diff(g, x)
quot_rule = (f_p*g - f*g_p) / g**2
direct = sp.diff(y, x)

print("y = f/g:"); display(y)
print("몫의 미분법:"); display(sp.simplify(quot_rule))
print("직접 미분:"); display(sp.simplify(direct))
print(f"일치?: {sp.simplify(quot_rule - direct) == 0}")
```

**몫의 미분**: <br>
=> $(f'g - fg')/g^2$
