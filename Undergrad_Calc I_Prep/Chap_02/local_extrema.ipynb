# 2. 함수의 극대와 극소

$f'(x)=0$인 **임계점**에서 부호 변화로 극대·극소 판정.

- $f'$: $+\to -$ → 극대
- $f'$: $-\to +$ → 극소


## 1. Example


```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
sp.init_printing()
x = sp.symbols('x', real=True)

f = x**3 - 3*x**2 + 2
f1 = sp.diff(f, x)
crit = sp.solve(f1, x)

print("f(x):"); display(f)
print("f'(x):"); display(f1)
print("임계점:", crit)

for c in crit:
    sign_left = f1.subs(x, c - sp.Rational(1,10))
    sign_right = f1.subs(x, c + sp.Rational(1,10))
    kind = '극대' if sign_left > 0 and sign_right < 0 else ('극소' if sign_left < 0 and sign_right > 0 else '판정불가')
    print(f"x={c}: f={f.subs(x,c)}, {kind}")

xs = np.linspace(-0.5, 3, 300)
plt.plot(xs, [float(f.subs(x,t)) for t in xs], 'b', lw=2)
for c in crit:
    plt.plot(float(c), float(f.subs(x,c)), 'ro')
plt.grid(True, alpha=0.3); plt.title('local maxima and minima')
plt.show()
```

**sp.solve(f1, x)**: <br>
=> $f'(x)=0$ 해 (임계점)
