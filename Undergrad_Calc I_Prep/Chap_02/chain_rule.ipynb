# 3. 합성함수의 미분법

$$
y = f(g(x)) \implies \frac{dy}{dx} = f'(g(x))\, g'(x)
$$


## 1. 연쇄법칙


```python
import sympy as sp
sp.init_printing()
x = sp.symbols('x', real=True)

u = sp.symbols('u', real=True)

g = 2*x + 1
f_u = u**3
y = f_u.subs(u, g)

dy_chain = sp.diff(f_u, u).subs(u, g) * sp.diff(g, x)
dy_direct = sp.diff(y, x)

print("y = f(g(x)):")
display(y)
print("연쇄법칙:")
display(sp.simplify(dy_chain))
print("직접 미분:")
display(sp.simplify(dy_direct))
print(f"일치?: {sp.simplify(dy_chain - dy_direct) == 0}")

print("Example) sin(2x):")
display(sp.Eq(sp.diff(sp.sin(2*x), x), sp.diff(sp.sin(2*x), x)))
```

**연쇄법칙 절차**: <br>
=> 바깥 미분 × 안쪽 미분. `subs` 로 $u=g(x)$ 대입 후 곱한다.
