# 1. 미분계수의 기하학적 의미

$f'(a)$는 곡선 $y=f(x)$ 위 점 $(a,f(a))$에서 **접선의 기울기**이다.


## 1. secant → tangent


```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
sp.init_printing()
x = sp.symbols('x', real=True)

f = x**2
a = 1
h_vals = [2, 1, 0.5, 0.1]

xs = np.linspace(-1, 3, 200)
plt.plot(xs, xs**2, 'b', lw=2, label='y=x²')
for h in h_vals:
    sec = (f.subs(x, a+h) - f.subs(x, a)) / h
    sec = float(sec)
    plt.plot(xs, float(f.subs(x,a)) + sec*(xs-a), '--', alpha=0.7, label=f'secant h={h}')
tan = float(sp.diff(f,x).subs(x,a))
plt.plot(xs, float(f.subs(x,a)) + tan*(xs-a), 'r', lw=2, label='tangent')
plt.plot(a, float(f.subs(x,a)), 'ko')
plt.xlim(-1,3); plt.ylim(-1,5); plt.grid(True, alpha=0.3); plt.legend(); plt.title("f'(1)=2: secant → tangent")
plt.show()
```

**해석**: <br>
=> $h\to 0$ 일 때 할선의 기울기가 접선의 기울기 $f'(a)$로 수렴
