# **미분계수**

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define the cubic function from your previous code
def f(x):
    return x**3 + x**2 + x + 1

# 2. Set the interval points [a, b]
a = 1
b = 3

# 3. Calculate the average rate of change (slope)
slope = (f(b) - f(a)) / (b - a)
print(f"Average Rate of Change between x={a} and x={b}: {slope}")

# 4. Generate X values for a smooth curve
x_vals = np.linspace(0.5, 3.5, 100)
y_vals = f(x_vals)

# 5. Create the plot
plt.figure(figsize=(8, 6))

# Plot the original function curve
plt.plot(x_vals, y_vals, label="f(x) = x³ + x² + x + 1", color="blue", linewidth=2)

# Plot the two distinct points
plt.scatter([a, b], [f(a), f(b)], color="black", zorder=5)

# Plot the red dashed secant line connecting the points
plt.plot([a, b], [f(a), f(b)], label=f"Secant Line (Slope = {slope:.1f})", color="red", linestyle="--", linewidth=2)

# Add text labels next to the points
plt.text(a, f(a) + 2, f"  ({a}, {f(a)})", fontsize=10, color="black")
plt.text(b, f(b) - 4, f"  ({b}, {f(b)})", fontsize=10, color="black")

# Graph styling
plt.title("Average Rate of Change Visualization", fontsize=14)
plt.xlabel("x", fontsize=12)
plt.ylabel("f(x)", fontsize=12)
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend(loc="upper left")

# Show the graph
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Convert SymPy function to a reusable numerical function
f_num = lambda val: val**3 + val**2 + val**1 + 1

# 2. Setup the tangent line data at x = 3
x1 = 3
y1 = f_num(x1)  # Evaluates to 40
m = 34  # The derivative result calculated by SymPy


# Tangent line equation: y = m*(x - x1) + y1
def tangent_eq(x_domain):
    return m * (x_domain - x1) + y1


# 3. Generate smooth continuous X intervals for plotting
x_scale = np.linspace(1.5, 4.0, 100)

# 4. Plot construction
plt.figure(figsize=(8, 6))

# Plot the target cubic function curve
plt.plot(x_scale, f_num(x_scale), label="f(x) Curve", color="blue", linewidth=2)

# Plot the red dashed tangent line
plt.plot(
    x_scale,
    tangent_eq(x_scale),
    label=f"Tangent Line (Slope/Derivative = {m})",
    color="red",
    linestyle="--",
    linewidth=1.5,
)

# Mark the exact point of tangency (3, 40)
plt.scatter([x1], [y1], color="black", zorder=5)
plt.text(x1, y1 + 5, f" Point (3, {y1})", color="black", fontsize=11)

# Formatting rules
plt.title("Derivative Visualization (Tangent Line)", fontsize=14)
plt.xlabel("x", fontsize=12)
plt.ylabel("f(x)", fontsize=12)
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend(loc="upper left")

# Display graphic window
plt.show()
```
