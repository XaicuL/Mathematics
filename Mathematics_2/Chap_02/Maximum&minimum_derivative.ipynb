# **함수의 최대 최소**

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return (x-1) * (x-2) * (x-3)

x = np.linspace(-2, 6, 400)   # 구간을 400등분
y = f(x)

max_idx = np.argmax(y)
min_idx = np.argmin(y)

max_x, max_y = x[max_idx], y[max_idx]
min_x, min_y = x[min_idx], y[min_idx]

plt.figure(figsize=(8, 5))
plt.plot(x, y, label='f(x)')
plt.scatter(max_x, max_y, color='red', label='max')
plt.scatter(min_x, min_y, color='blue', label='min')

plt.annotate(f'max=({max_x:.2f}, {max_y:.2f})', (max_x, max_y))
plt.annotate(f'min=({min_x:.2f}, {min_y:.2f})', (min_x, min_y))

plt.axhline(0, color='gray', linestyle='--', linewidth=0.8)
plt.axvline(0, color='gray', linestyle='--', linewidth=0.8)
plt.axvline(-2, color='gray', linestyle=':', linewidth=1)
plt.axvline(6, color='gray', linestyle=':', linewidth=1)
plt.legend()
plt.grid(True)
plt.show()
```
