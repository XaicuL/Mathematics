# **방정식과 부등식 With 미분**

```python
# 방정식 with 미분

import sympy as sp

x = sp.symbols('x', real=True)
f = x**3 - 3*x + 1
df = sp.diff(f, x)          # f'(x) = 3x^2 - 3
f_num = sp.lambdify(x, f)
df_num = sp.lambdify(x, df)

def newton(x0, tol=1e-10, max_iter=100):
    xn = x0
    for _ in range(max_iter):
        fx, dfx = f_num(xn), df_num(xn)
        if abs(dfx) < 1e-14:
            raise ValueError("도함수가 0에 가까워 발산 위험")
        x_next = xn - fx / dfx
        if abs(x_next - xn) < tol:
            return x_next
        xn = x_next
    return xn

root = newton(x0=2.0)
print(root)   # 약 1.532
```

```python
#부등식 with 미분
crit_points = sp.solve(sp.Eq(df, 0), x)   # [-1, 1]
crit_points = sorted([float(c) for c in crit_points])

# 임계점을 기준으로 구간을 나누고 각 구간의 대표점에서 f의 부호를 확인
boundaries = [-10] + crit_points + [10]
intervals = []
for i in range(len(boundaries) - 1):
    a, b = boundaries[i], boundaries[i+1]
    mid = (a + b) / 2
    sign = "양수 (f>0)" if f_num(mid) > 0 else "음수 (f<0)"
    intervals.append((a, b, sign))

for a, b, sign in intervals:
    print(f"({a}, {b}) 구간: {sign}")
```
