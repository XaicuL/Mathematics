```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-3, 3, 100)

y = 2**x

plt.figure(figsize=(8, 5))
plt.plot(x, y, label='$f(x) = 2^x$', color='red', linewidth=2)

plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)

plt.plot(0, 1, 'go', markersize=8, label='Vertex $(0,1)$')
plt.text(0.1, 1.2, '$(0, 1)$', fontsize=12, color='green', fontweight='bold')

plt.title('Exp Function Graph ($2^x$)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0.1, 8, 100)

y = np.log2(x)

plt.figure(figsize=(8, 6))
plt.plot(x, y, label=r'$f^{-1}(x) = \log_2(x)$', color='blue', linewidth=2)

plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)

plt.plot(1, 0, 'ro', markersize=8, label='Vertex $(1,0)$')
plt.text(1.2, 0.2, '$(1, 0)$', fontsize=12, color='red', fontweight='bold')

plt.title(r'Log Function Graph ($\log_2(x)$)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()

plt.xlim(-1, 8)
plt.ylim(-3, 4)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x_exp = np.linspace(-3, 3, 100)
y_exp = 2**x_exp

x_log = np.linspace(0.1, 8, 100)
y_log = np.log2(x_log)

x_line = np.linspace(-3, 8, 100)

plt.figure(figsize=(8, 8))

plt.plot(x_exp, y_exp, label=r'$f(x) = 2^x$', color='red', linewidth=2)
plt.plot(x_log, y_log, label=r'$f^{-1}(x) = \log_2(x)$', color='blue', linewidth=2)
plt.plot(x_line, x_line, label=r'$y = x$', color='gray', linestyle='--', linewidth=1.5)

plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)

plt.plot(0, 1, 'ro', markersize=8)
plt.text(-1.2, 1.2, '$(0, 1)$', fontsize=11, color='red', fontweight='bold')

plt.plot(1, 0, 'bo', markersize=8)
plt.text(1.2, -0.5, '$(1, 0)$', fontsize=11, color='blue', fontweight='bold')

plt.title(r'Exp & Log Functions', fontsize=14, pad=15)
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(fontsize=11, loc='upper left')

plt.gca().set_aspect('equal', adjustable='box')

plt.xlim(-3, 6)
plt.ylim(-3, 6)

plt.show()
```
