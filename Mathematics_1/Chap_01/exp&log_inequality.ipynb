# **지수&로그 부등식**

```python
from sympy import symbols, log, Eq, solve, solve_univariate_inequality

x = symbols('x')

eq1 = Eq(2**x, 16)
sol1 = solve(eq1, x)

print("1. Equation: 2^x = 16")
print(f"Solution: x = {sol1}")


eq2 = 4**x - 3*(2**x) + 2
sol2 = solve(eq2, x)

print("2. Equation: 4^x - 3*(2^x) + 2 = 0")
print(f"Solution: x = {sol2}")


eq3 = Eq(log(x, 2), 5)
sol3 = solve(eq3, x)

print("3. Equation: log2(x) = 5")
print(f"Solution: x = {sol3}")


eq4 = Eq(log(x - 1, 3), 2)
sol4 = solve(eq4, x)

print("4. Equation: log3(x - 1) = 2")
print(f"Solution: x = {sol4}")


ineq1 = 2**x > 16
sol_ineq1 = solve_univariate_inequality(ineq1, x)

print("5. Inequality: 2^x > 16")
print(f"Solution: {sol_ineq1}")


ineq2 = 3**(2*x - 1) <= 9
sol_ineq2 = solve_univariate_inequality(ineq2, x)

print("6. Inequality: 3^(2*x - 1) <= 9")
print(f"Solution: {sol_ineq2}")
```

```python
from sympy import symbols, log, Eq, solve, solve_univariate_inequality

x = symbols('x')

eq1 = Eq(2**x, 16)
sol1 = solve(eq1, x)

print("1. Equation: 2^x = 16")
print(f"Solution: x = {sol1}")


eq2 = 4**x - 3*(2**x) + 2
sol2 = solve(eq2, x)

print("2. Equation: 4^x - 3*(2^x) + 2 = 0")
print(f"Solution: x = {sol2}")


eq3 = Eq(log(x, 2), 5)
sol3 = solve(eq3, x)

print("3. Equation: log2(x) = 5")
print(f"Solution: x = {sol3}")


eq4 = Eq(log(x - 1, 3), 2)
sol4 = solve(eq4, x)

print("4. Equation: log3(x - 1) = 2")
print(f"Solution: x = {sol4}")


ineq1 = 2**x > 16
sol_ineq1 = solve_univariate_inequality(ineq1, x)

print("5. Inequality: 2^x > 16")
print(f"Solution: {sol_ineq1}")


ineq2 = 3**(2*x - 1) <= 9
sol_ineq2 = solve_univariate_inequality(ineq2, x)

print("6. Inequality: 3^(2*x - 1) <= 9")
print(f"Solution: {sol_ineq2}")


ineq3 = log(x, 2) < 4
sol_ineq3 = solve_univariate_inequality(ineq3, x, relational=False)

print("7. Inequality: log2(x) < 4")
print(f"Solution: {sol_ineq3}")


ineq4 = log(x - 2, 5) >= 1
sol_ineq4 = solve_univariate_inequality(ineq4, x, relational=False)

print("8. Inequality: log5(x - 2) >= 1")
print(f"Solution: {sol_ineq4}")
```
