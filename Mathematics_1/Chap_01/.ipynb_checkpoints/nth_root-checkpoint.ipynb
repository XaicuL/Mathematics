# **거듭제곱근**

```python
import math
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = 16

re1 = x ** (1/4) #math 모듈 사용X
print(f'16의 네제곱근 {re1}')

reM1 = math.sqrt(x)
print(f'16의 제곱근 {reM1}')

reM2 = math.cbrt(8)
print(f'8의 세제곱근 {reM2}')
```

```python
x = np.linspace(-2.5, 2.5, 400)
a_values = [2, 0, -2]
colors = ['green', 'orange', 'crimson']

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

def draw_math_graph(ax, title, y_func, label):
    ax.plot(x, y_func, color='black', linewidth=2, label=label)

    # X축과 Y축을 중앙(0)으로 이동
    ax.spines['left'].set_position('zero')   # Y축 이동
    ax.spines['bottom'].set_position('zero') # X축 이동

    # 위쪽과 오른쪽 테두리 제거
    ax.spines['right'].set_color('none')
    ax.spines['top'].set_color('none')

    # a 값 가로선 추가
    for a, c in zip(a_values, colors):
        if a != 0: # 0은 X축과 겹치므로 제외하거나 스타일 변경
            ax.axhline(y=a, color=c, linestyle='--', alpha=0.6, label=f'a={a}')

    ax.set_title(title, pad=20)
    ax.legend(loc='upper right')
    ax.grid(True, linestyle=':', alpha=0.5)

draw_math_graph(ax1, 'n is Even', x**2, '$y=x^2$')
ax1.set_ylim(-3, 5)

draw_math_graph(ax2, 'n is Odd', x**3, '$y=x^3$')
ax2.set_ylim(-5, 5)

plt.tight_layout()
plt.show()
```

# 성질

```python
#1 : 서로 같은 지수일 때 곱셈

x = 2
y = 3

expr1 = math.sqrt(x)
expr2 = math.sqrt(y)
law_sq = expr1 * expr2

print(f'x의 제곱근 * y의 제곱근 = {law_sq}')
```

```python
#2 : 서로 같은 지수일 때 나눗셈

x = 2
y = 3

expr1 = math.sqrt(x)
expr2 = math.sqrt(y)
law_dv = expr1 / expr2

print(f'x의 제곱근 / y의 제곱근 = {law_dv}')
```
