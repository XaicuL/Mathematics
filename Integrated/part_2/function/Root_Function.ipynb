# 무리함수 ($\sqrt{x}$)

제곱근함수의 기본형:

$$
y = \sqrt{x}
$$

- **정의역**: $x \ge 0$
- **치역**: $y \ge 0$
- 그래프는 원점에서 시작해 오른쪽으로 증가 (위로 볼록하지 않은 완만한 증가)


## 1. 기본형 $y=\sqrt{x}$


```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

sp.init_printing()
x = sp.symbols('x', real=True)

f = sp.sqrt(x)
print("식:")
display(f)
print("정의역(sympy):", sp.calculus.util.continuous_domain(f, x, sp.S.Reals))

xs = np.linspace(0, 9, 200)
ys = np.sqrt(xs)

plt.figure(figsize=(7, 4.5))
plt.axhline(0, color='gray', lw=0.8)
plt.axvline(0, color='gray', lw=0.8)
plt.plot(xs, ys, label=r'$y=\sqrt{x}$', color='C0', lw=2)
plt.xlabel('x'); plt.ylabel('y')
plt.title(r'original $y=\sqrt{x}$')
plt.legend(); plt.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()
```

**np.sqrt / sp.sqrt**: <br>
=> 제곱근 (numpy는 $x<0$에서 nan)

**sp.calculus.util.continuous_domain**: <br>
=> 실수에서 연속인 정의역


## 2. 평행이동 $y = \sqrt{x-a} + b$

| 변형 | 효과 |
|------|------|
| $\sqrt{x-a}$ | $x$축으로 $a$만큼 평행이동 ($a>0$이면 오른쪽) |
| $+\,b$ | $y$축으로 $b$만큼 평행이동 |
| 정의역 | $x-a \ge 0$ ⇒ $x \ge a$ |


```python
a, b = 2, 1  # y = sqrt(x-2) + 1
print(f"y = √(x - {a}) + {b}")
print(f"정의역: x ≥ {a},  시작점: ({a}, {b})")

xs2 = np.linspace(a, a + 9, 200)
ys2 = np.sqrt(xs2 - a) + b

plt.figure(figsize=(7, 4.5))
plt.axhline(0, color='gray', lw=0.8)
plt.axvline(0, color='gray', lw=0.8)
plt.plot(np.linspace(0, 9, 200), np.sqrt(np.linspace(0, 9, 200)),
         '--', color='gray', alpha=0.7, label=r'$y=\sqrt{x}$ (original)')
plt.plot(xs2, ys2, color='C1', lw=2, label=rf'$y=\sqrt{{x-{a}}}+{b}$')
plt.scatter([a], [b], color='C3', zorder=5)
plt.annotate(f'Starting point ({a},{b})', xy=(a, b), xytext=(a+1.2, b+0.8),
             arrowprops=dict(arrowstyle='->', color='gray'))
plt.xlabel('x'); plt.ylabel('y')
plt.legend(); plt.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()
```

## 3. 여러 평행이동 비교


```python
plt.figure(figsize=(8, 5))
plt.axhline(0, color='gray', lw=0.8)
plt.axvline(0, color='gray', lw=0.8)

configs = [(0, 0, r'$y=\sqrt{x}$'),
           (2, 0, r'$y=\sqrt{x-2}$'),
           (0, 1, r'$y=\sqrt{x}+1$'),
           (-1, -0.5, r'$y=\sqrt{x+1}-0.5$')]
for aa, bb, lab in configs:
    t = np.linspace(aa, aa + 8, 200)
    plt.plot(t, np.sqrt(t - aa) + bb, lw=2, label=lab)

plt.xlabel('x'); plt.ylabel('y')
plt.title(r'Comparison of $y=\sqrt{x-a}+b$')
plt.legend(); plt.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()
```
