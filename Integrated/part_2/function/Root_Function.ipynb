# **무리함수**

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 1. 무리함수 f(x) = sqrt(x)
# x는 0 이상이어야 하므로 np.linspace의 시작점을 0으로 설정
x_irrational = np.linspace(0, 4, 100)
y_irrational = np.sqrt(x_irrational)

# 2. 역함수인 이차함수 g(x) = x^2 (단, x >= 0)
x_quadratic = np.linspace(0, 2, 100)
y_quadratic = x_quadratic**2

# 3. 시각화
plt.figure(figsize=(8, 8))

# 무리함수 그래프
plt.plot(x_irrational, y_irrational, label=r'$f(x) = \sqrt{x}$', color='crimson', lw=2)

# 이차함수(역함수) 그래프
plt.plot(x_quadratic, y_quadratic, label=r'$g(x) = x^2 \ (x \geq 0)$', color='darkorange', lw=2)

# y = x 대칭선
plt.plot([0, 4], [0, 4], 'k--', alpha=0.5, label='$y = x$')

# 그래프 디테일
plt.title("Irrational Function & Its Inverse", fontsize=15)
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.axhline(0, color='black', lw=1)
plt.axvline(0, color='black', lw=1)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()
plt.gca().set_aspect('equal') # 대칭 확인을 위해 비율 1:1 유지
plt.show()
```
