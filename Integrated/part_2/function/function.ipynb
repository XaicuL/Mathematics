# 함수

함수 $f$는 정의역의 각 원소에 **함숫값**을 하나만 대응시키는 대응

$$
y = f(x)
$$

| 용어 | 의미 |
|------|------|
| 정의역 (domain) | $x$가 가질 수 있는 값의 집합 |
| 치역 (range) | 실제로 나오는 $f(x)$의 집합 |
| 공역 (codomain) | 함숫값이 속한다고 정한 집합 (치역 ⊇ 아님, 치역 ⊆ 공역) |


## 1. 함수 정의와 함숫값


```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

sp.init_printing()
x = sp.symbols('x')

# 기호: f(x) = 2x + 3
f_expr = 2*x + 3
f = sp.lambdify(x, f_expr, 'numpy')

print("함수:")
display(sp.Eq(sp.Function('f')(x), f_expr))
print(f"f(2) = {f_expr.subs(x, 2)}")
print(f"f(-1) = {f_expr.subs(x, -1)}")
```

**sp.lambdify(x, expr, 'numpy')**: <br>
=> 기호식을 numpy에서 쓸 수 있는 함수로 변환

**expr.subs(x, a)**: <br>
=> $x$에 $a$를 대입한 함숫값


## 2. 정의역 · 치역 (간단한 예)


```python
# 정의역을 [-5, 5]로 두고 치역(이미지)을 수치로 추정
domain = np.linspace(-5, 5, 201)
values = f(domain)

print(f"정의역 구간: [{domain.min()}, {domain.max()}]")
print(f"함숫값 최소·최대 (치역 추정): [{values.min():.3f}, {values.max():.3f}]")
# 일차함수 f(x)=2x+3 이면 치역도 구간 [-7, 13]
```

## 3. 일대일(단사) · 전사(onto) — 짧게

- **일대일(단사)**: $f(a)=f(b) \Rightarrow a=b$ (서로 다른 $x$가 같은 $y$를 갖지 않음)
- **전사(onto)**: 공역의 모든 원소가 어떤 $x$의 함숫값으로 나옴
- 실수 전체에서 $f(x)=2x+3$은 일대일이고 전사(역함수 존재)
- $g(x)=x^2$ ($x\in\mathbb{R}$, 공역 $\mathbb{R}$)은 일대일·전사 모두 아님


```python
g_expr = x**2
# 같은 함숫값 → 서로 다른 x 존재?
print("g(2) =", g_expr.subs(x, 2), ", g(-2) =", g_expr.subs(x, -2))
print("→ 일대일 아님 (정의역을 x≥0으로 제한하면 일대일)")
```

## 4. Example

```python
xs = np.linspace(-5, 5, 200)
ys = f(xs)

plt.figure(figsize=(7, 4.5))
plt.axhline(0, color='gray', lw=0.8)
plt.axvline(0, color='gray', lw=0.8)
plt.plot(xs, ys, label=r'$f(x)=2x+3$', color='C0', lw=2)
plt.scatter([2], [f(2)], color='C3', zorder=5)
plt.annotate(r'$(2, f(2))$', xy=(2, f(2)), xytext=(2.5, f(2)+2),
             arrowprops=dict(arrowstyle='->', color='gray'))
plt.xlabel('x'); plt.ylabel('y')
plt.legend(); plt.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()
```

**np.linspace(a, b, n)**: <br>
=> $[a,b]$를 $n$등분한 배열

**plt.plot / plt.scatter**: <br>
=> 곡선 · 점 표시
