# **함수**

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 1. 함수 정의 (수학의 f(x) 역할)
def f(x):
    return 2 * x + 3

# 2. 정의역(Domain) 설정: -5부터 5까지 100개의 숫자
x = np.linspace(-5, 5, 100)

# 3. 치역(Range) 계산: 함수에 x를 넣어서 나온 결과값들
y = f(x)
```

```python
plt.figure(figsize=(8, 5))
plt.plot(x, y, label='f(x) = 2x + 3', color='purple', lw=2)

# 특정 점 하나 찍어보기 (예: x=2일 때)
test_x = 2
test_y = f(test_x)
plt.scatter(test_x, test_y, color='red', zorder=5)
plt.text(test_x, test_y, f'  f({test_x}) = {test_y}', fontsize=12)

plt.axhline(0, color='black', lw=0.5)
plt.axvline(0, color='black', lw=0.5)
plt.title("Basic Function: Domain & Range")
plt.xlabel("Domain (x)")
plt.ylabel("Range (f(x))")
plt.grid(True, linestyle='--')
plt.legend()
plt.show()
```
