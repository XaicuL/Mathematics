# 행렬

행렬의 정의, 덧셈·뺄셈, 스칼라배, **행렬곱**, 단위행렬.

주의: `A * B` 는 **원소별(하마다) 곱**, 행렬곱은 `A @ B` 또는 `np.matmul(A, B)`.


## 1. 행렬 정의


```python
import numpy as np

A = np.array([
    [3, 18, 7, 55],
    [44, 7, 13, 32],
])

print(f"shape: {A.shape}  → {A.shape[0]}×{A.shape[1]} 행렬")
print(A)
```

**np.array([[...], [...]])**: <br>
=> 2차원 배열로 행렬 생성

**A.shape**: <br>
=> `(행 수, 열 수)`


## 2. 덧셈·뺄셈·스칼라배

같은 크기일 때만 덧셈·뺄셈 가능. 스칼라배는 모든 원소에 곱.


```python
k = 3
P = np.array([[15, 23],
              [19, 51]])
Q = np.array([[30, 47],
              [24, 35]])

P_plus_Q = P + Q
P_minus_Q = P - Q
kP = k * P

print("P + Q =")
print(P_plus_Q)
print("P - Q =")
print(P_minus_Q)
print(f"{k}P =")
print(kP)
```

## 3. 원소별 곱 vs 행렬곱

| 연산 | 의미 | numpy |
|------|------|-------|
| 원소별 곱 | $(AB)_{ij} = A_{ij} B_{ij}$ | `A * B` |
| 행렬곱 | 행·열 내적 | `A @ B` / `np.matmul` |


```python
A = np.array([[1, 2],
              [3, 4]])
B = np.array([[5, 6],
              [7, 8]])

hadamard = A * B
matmul1 = A @ B
matmul2 = np.matmul(A, B)

print("원소별 곱 A*B:")
print(hadamard)
print("행렬곱 A@B:")
print(matmul1)
print(f"matmul 일치?: {np.array_equal(matmul1, matmul2)}")

# (2×3)·(3×2) 예
C = np.array([[1, 2, 3],
              [4, 5, 6]])
D = np.array([[1, 0],
              [0, 1],
              [1, 1]])
print("C@D shape:", (C @ D).shape)
print(C @ D)
```

**A \* B**: <br>
=> 원소별 곱 (행렬곱 아님!)

**A @ B** / **np.matmul(A, B)**: <br>
=> 행렬곱. 열 수 = 행 수 조건 필요


## 4. 단위행렬

$I_n$ 은 대각이 1, 나머지가 0. $AI = IA = A$.


```python
I2 = np.eye(2)
A = np.array([[2, 1],
              [0, 3]])

print("I_2 =")
print(I2)
print("A @ I =")
print(A @ I2)
print("I @ A =")
print(I2 @ A)
print(f"A와 같음?: {np.array_equal(A @ I2, A) and np.array_equal(I2 @ A, A)}")
```

**np.eye(n)**: <br>
=> $n \times n$ 단위행렬


## 5. Example 


```python
M = np.array([[1, 2],
              [3, 4]])
N = np.array([[0, 1],
              [1, 0]])

print("M+N:\n", M + N)
print("2M:\n", 2 * M)
print("M@N:\n", M @ N)
print("N@M:\n", N @ M)  # 일반적으로 MN ≠ NM
```
