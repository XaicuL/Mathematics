# 삼차 & 사차 방정식

고차 방정식은 이차처럼 근의 공식이 간단하지 않다.  
통합수학에서는 주로 **인수분해 · 치환 · 조립제법**으로 다룬다.

관련: `factorization.ipynb`, `Remainder_Theorem.ipynb`, `Discriminant.ipynb`

```python
import sympy as sp
sp.init_printing()

x = sp.symbols('x')
```

## 1. 인수분해로 풀기

삼차·사차를 인수분해하면 $(x-a)(\cdots)=0$ 형태로 근을 읽는다.

```python
expr1 = x**3 + 1
expr2 = x**4 - 10*x**2 + 9

print("삼차: x³ + 1")
display(sp.factor(expr1))
print("근:")
display(sp.solve(expr1, x))

print("\n사차: x⁴ − 10x² + 9")
display(sp.factor(expr2))
print("근:")
display(sp.solve(expr2, x))
```

**sp.factor(expr)**: <br>
=> 다항식을 인수분해

**sp.solve(expr, x)**: <br>
=> `expr = 0` 의 근을 구함

## 2. 치환 (짝수 차수만 있는 사차)

$x^4 + px^2 + q = 0$ 꼴이면 $t = x^2$ 로 놓아 이차로 줄인다.

예: $x^4 - 5x^2 + 4 = 0$

```python
t = sp.symbols('t', nonnegative=True)

# t = x² 치환
quad_in_t = t**2 - 5*t + 4
print("t에 대한 이차식:")
display(quad_in_t)

t_roots = sp.solve(quad_in_t, t)
print("t의 근:")
display(t_roots)

# t = x² 이므로 x = ±√t
x_roots = []
for ti in t_roots:
    x_roots.extend(sp.solve(x**2 - ti, x))

print("원래 방정식의 근:")
display(x_roots)

print("검증 (직접 solve):")
display(sp.solve(x**4 - 5*x**2 + 4, x))
```

## 3. 인수정리 / 가능한 유리근

$a_n x^n + \cdots + a_0 = 0$ 에서 유리근 후보는  
$\pm$ (상수항의 약수) / (최고차 계수의 약수).

예: $x^3 - 6x^2 + 11x - 6 = 0$

```python
expr = x**3 - 6*x**2 + 11*x - 6
poly = sp.Poly(expr, x)

# 유리근 정리: ± (상수항 약수) / (최고차 계수 약수)
const = int(poly.TC())
lead = int(poly.LC())
cands = {
    sp.Rational(p, q)
    for p in sp.divisors(abs(const))
    for q in sp.divisors(abs(lead))
}
cands |= {-c for c in cands}

print("가능한 유리근 후보:")
display(sorted(cands, key=lambda r: (float(r),)))

roots_found = sorted((r for r in cands if expr.subs(x, r) == 0), key=lambda r: float(r))
print("실제로 근인 후보:")
display(roots_found)

print("인수분해:")
display(sp.factor(expr))
print("모든 근:")
display(sp.solve(expr, x))
```

**유리근 정리**: <br>
=> 유리근 후보는 $\pm$(상수항의 약수)/(최고차 계수의 약수)

**sp.divisors(n)**: <br>
=> $n$의 양의 약수 목록

(조립제법은 `Remainder_Theorem.ipynb` 참고)


## 4. 삼차 = 이차 × 일차 (한 근을 알 때)

한 근 $x = r$ 을 알면 $(x-r)$ 로 나눈 나머지가 0이고, 몫은 이차.
그 이차는 판별식/`Discriminant.ipynb`로 이어서 푼다.

```python
expr = x**3 - 3*x + 2   # (x-1)²(x+2)
r = 1

quot, rem = sp.div(expr, x - r)
print(f"(x − {r})로 나눈 몫 / 나머지:")
display(quot, rem)

print("몫(이차)의 판별식:")
qpoly = sp.Poly(quot, x)
display(sp.discriminant(qpoly))

print("이차의 근:")
display(sp.solve(quot, x))

print("전체 근:")
display(sp.solve(expr, x))
```

**sp.div(A, B)**: <br>
=> 다항식 나눗셈 → (몫, 나머지)

**sp.discriminant(poly)**: <br>
=> 몫이 이차일 때 근의 개수/성질 판별

## 5. 중근이 있는 경우

고차에서도 **중근**이 있으면 인수에 제곱이 나타난다.  
(판별식 $D=0$ 과 같은 역할의 “겹근” 직관)

```python
expr = (x - 2)**2 * (x + 1)
expanded = sp.expand(expr)

print("전개:")
display(expanded)
print("인수분해:")
display(sp.factor(expanded))
print("근 (중근 포함):")
display(sp.roots(expanded, x))  # {근: 중복도}
```

**sp.roots(expr, x)**: <br>
=> `{근: 중복도}` 딕셔너리

## 6. 복소수 근이 나오는 삼차

$x^3 - 1 = 0$ → 실근 1개 + 허근 2개  
(허근 연산은 `Complex_Number.ipynb`)

```python
expr = x**3 - 1
print("인수분해:")
display(sp.factor(expr))
print("근:")
display(sp.solve(expr, x))

# 이차 인수 x² + x + 1 의 판별식
quad = x**2 + x + 1
print("이차 인수의 판별식:")
display(sp.discriminant(sp.Poly(quad, x)))
```

## 정리

| 상황 | 방법 |
|------|------|
| 눈에 보이는 인수 | `factor` |
| $x^4 + px^2 + q$ | $t=x^2$ 치환 |
| 유리근이 있을 때 | 유리근 후보 → 나눗셈 → 이차 |
| 몫이 이차 | 판별식 / 근의 공식 |
| 허근 | 복소수 노트와 연결 |
