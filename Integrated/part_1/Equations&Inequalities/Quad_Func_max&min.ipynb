# 이차함수의 최대·최소

$y = ax^2 + bx + c$ $(a \neq 0)$

- 꼭짓점: $\displaystyle\left(-\frac{b}{2a},\; f\!\left(-\frac{b}{2a}\right)\right)$
- $a > 0$: 아래로 볼록 → 꼭짓점에서 **최솟값**
- $a < 0$: 위로 볼록 → 꼭짓점에서 **최댓값**

닫힌 구간 $[p,q]$ 에서는 **끝점과 구간 안 꼭짓점**만 비교하면 된다.

관련: `Discriminant.ipynb`, `Quadratic_Inequality.ipynb`.


## 1. 꼭짓점 공식

$$
x = -\frac{b}{2a}
$$


```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
sp.init_printing()

x = sp.symbols('x')
a, b, c = 1, -4, 3  # y = x^2 - 4x + 3
f = a*x**2 + b*x + c

vx = -b / (2*a)
vy = f.subs(x, vx)

print("함수:")
display(f)
print(f"꼭짓점: ({vx}, {vy})")
print(f"a={a} > 0 → 최솟값 {vy}")
```

**-b/(2*a)**: <br>
=> 꼭짓점의 $x$좌표

**f.subs(x, vx)**: <br>
=> 그 점에서의 함숫값


## 2. 완전제곱으로 변형

$$
ax^2 + bx + c = a\left(x + \frac{b}{2a}\right)^2 + \left(c - \frac{b^2}{4a}\right)
$$


```python
completed_expr = sp.simplify(a*(x + b/(2*a))**2 + (c - b**2/(4*a)))
print("완전제곱 형태:")
display(completed_expr)
print(f"전개 일치?: {sp.expand(completed_expr) == f}")
```

## 3. 닫힌 구간에서의 최대·최소

고정 예: $f(x) = x^2 - 4x + 3$, 구간 $[-1, 5]$


```python
a, b, c = 1, -4, 3
x_start, x_end = -1, 5

def f_num(t):
    return a*t**2 + b*t + c

vertex_x = -b / (2*a)
candidates = [x_start, x_end]
if x_start <= vertex_x <= x_end:
    candidates.append(vertex_x)

pairs = [(t, f_num(t)) for t in candidates]
max_x, max_y = max(pairs, key=lambda p: p[1])
min_x, min_y = min(pairs, key=lambda p: p[1])

print("후보점:", pairs)
print(f"최댓값: f({max_x}) = {max_y}")
print(f"최솟값: f({min_x}) = {min_y}")
```

## 4. 그래프 (고정 계수)

`input()` 없이 위에서 정한 $a,b,c$ 와 구간으로 시각화.


```python
margin = (x_end - x_start) * 0.2
x_plot = np.linspace(x_start - margin, x_end + margin, 400)
y_plot = f_num(x_plot)

x_range = np.linspace(x_start, x_end, 200)
y_range = f_num(x_range)

plt.figure(figsize=(8, 5))
plt.plot(x_plot, y_plot, label=r"$f(x)=x^2-4x+3$", color="C0", alpha=0.5)
plt.plot(x_range, y_range, color="C3", linewidth=3, label="구간 [-1, 5]")

xs_c = [p[0] for p in pairs]
ys_c = [p[1] for p in pairs]
plt.scatter(xs_c, ys_c, color="black", zorder=5)
for tx, ty in pairs:
    plt.annotate(f"({tx:g}, {ty:g})", (tx, ty), textcoords="offset points", xytext=(5, 8))

plt.axhline(0, color="gray", lw=0.8)
plt.axvline(0, color="gray", lw=0.8)
plt.grid(True, alpha=0.3)
plt.legend()
plt.title("이차함수 구간 최대·최소")
plt.tight_layout()
plt.show()
```

## 5. $a < 0$ 인 경우

$f(x) = -x^2 + 2x + 3$, 구간 $[0, 3]$ → 꼭짓점에서 최댓값.


```python
a2, b2, c2 = -1, 2, 3
lo, hi = 0, 3
vx2 = -b2 / (2*a2)

def g(t):
    return a2*t**2 + b2*t + c2

cands = [lo, hi] + ([vx2] if lo <= vx2 <= hi else [])
vals = [(t, g(t)) for t in cands]
print("후보:", vals)
print(f"최댓값: {max(vals, key=lambda p: p[1])}")
print(f"최솟값: {min(vals, key=lambda p: p[1])}")
```
