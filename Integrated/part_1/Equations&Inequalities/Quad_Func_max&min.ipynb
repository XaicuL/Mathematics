# 이차함수의 최대 최소

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a = float(input("이차항의 계수(a): "))
b = float(input("일차항의 계수(b): "))
c = float(input("상수항(c): "))
x_start = float(input("범위 시작(x_min): "))
x_end = float(input("범위 끝(x_max): "))
```

```python
p = -b / (2 * a)
x_candidates = [x_start, x_end]
if x_start <= p <= x_end:
    x_candidates.append(p)

y_values = [a*x**2 + b*x + c for x in x_candidates]
max_y, min_y = max(y_values), min(y_values)

margin = (x_end - x_start) * 0.2
x_plot = np.linspace(x_start - margin, x_end + margin, 400)
y_plot = a * x_plot**2 + b * x_plot + c
```

```python
plt.figure(figsize=(8, 6))
plt.plot(x_plot, y_plot, label=f'f(x) = {a}x² + {b}x + {c}', color='blue')

x_range = np.linspace(x_start, x_end, 200)
y_range = a * x_range**2 + b * x_range + c
plt.plot(x_range, y_range, color='red', linewidth=3, label='Defined Range')

plt.scatter(x_candidates, y_values, color='black', zorder=5)
for x, y in zip(x_candidates, y_values):
    plt.text(x, y, f'({x:.1f}, {y:.1f})', verticalalignment='bottom')

plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.grid(True, linestyle='--')
plt.legend()
plt.title("Quadratic Function Max/Min")
plt.xlabel("x")
plt.ylabel("y")

print(f"\n[결과] 최댓값: {max_y}, 최솟값: {min_y}")
plt.show()
```
