# 근과 계수의 관계

이차방정식 $ax^2 + bx + c = 0$ $(a \neq 0)$ 의 두 근을 $\alpha, \beta$ 라 하면:

$$
\alpha + \beta = -\frac{b}{a}, \qquad \alpha\beta = \frac{c}{a}
$$

판별식·실근과의 연결은 `Discriminant.ipynb`, 허근은 `Complex_Number.ipynb`.


## 1. 기본: 계수 → 합·곱


```python
import sympy as sp
sp.init_printing()

x = sp.symbols('x')
expr = x**2 - 5*x + 6
a, b, c = sp.Poly(expr, x).all_coeffs()

sum_roots = -b / a
prod_roots = c / a
roots = sp.solve(expr, x)

print("식:")
display(sp.Eq(expr, 0))
print(f"계수: a={a}, b={b}, c={c}")
print(f"합 α+β = -b/a = {sum_roots}")
print(f"곱 αβ = c/a = {prod_roots}")
print("실제 근:")
display(roots)
print(f"검증: 합={sum(roots)}, 곱={sp.prod(roots)}")
```

**sp.Poly(expr, x).all_coeffs()**: <br>
=> 최고차항부터 계수 추출

**sp.solve(expr, x)**: <br>
=> 방정식의 근


## 2. 근의 공식과의 연결

$$
\alpha,\beta = \frac{-b \pm \sqrt{D}}{2a}, \quad D = b^2 - 4ac
$$

- 합·곱은 $D$ 부호와 무관하게 성립 (허근이어도 성립)
- $D \ge 0$ 이면 실근의 합·곱으로 바로 해석


```python
def viete_quad(a, b, c):
    return -b/a, c/a

examples = [
    (1, -5, 6),   # D > 0
    (1, -4, 4),   # D = 0
    (1,  2, 5),   # D < 0
]

for a, b, c in examples:
    D = b**2 - 4*a*c
    s, p = viete_quad(a, b, c)
    roots = sp.solve(a*x**2 + b*x + c, x)
    print("=" * 40)
    display(sp.Eq(a*x**2 + b*x + c, 0))
    print(f"D = {D},  합={s},  곱={p}")
    display(roots)
```

## 3. 응용 예제

Ex1) 두 근의 합이 3, 곱이 2 → 이차방정식 만들기  
Ex2) $x^2 - 7x + k = 0$ 의 두 근의 차가 3일 때 $k$


```python
# Ex1: 합=3, 곱=2 → x^2 - (합)x + (곱) = 0
eq1 = x**2 - 3*x + 2
print("합=3, 곱=2 인 방정식:")
display(sp.Eq(eq1, 0))
display(sp.solve(eq1, x))

# Ex2: 근 α, β → (α-β)^2 = (α+β)^2 - 4αβ = 9
k = sp.symbols('k')
# 합=7, 곱=k
diff_sq = 7**2 - 4*k
k_sol = sp.solve(sp.Eq(diff_sq, 9), k)[0]
print(f"두 근의 차가 3일 때 k = {k_sol}")
print("그때의 근:")
display(sp.solve(x**2 - 7*x + k_sol, x))
```

## 4. 삼차방정식 (짧게)

$ax^3 + bx^2 + cx + d = 0$ 의 근 $\alpha,\beta,\gamma$:

$$
\alpha+\beta+\gamma = -\frac{b}{a}, \quad
\alpha\beta+\beta\gamma+\gamma\alpha = \frac{c}{a}, \quad
\alpha\beta\gamma = -\frac{d}{a}
$$

고차 실습은 `higher-order_equations.ipynb`.


```python
expr3 = x**3 - 6*x**2 + 11*x - 6
poly3 = sp.Poly(expr3, x)
a3, b3, c3, d3 = poly3.all_coeffs()
roots3 = sp.solve(expr3, x)

print("삼차식:")
display(sp.Eq(expr3, 0))
print(f"합 = -b/a = {-b3/a3}")
print(f"둘씩 곱의 합 = c/a = {c3/a3}")
print(f"곱 = -d/a = {-d3/a3}")
print("근:")
display(roots3)
print(f"검증: 합={sum(roots3)}, 곱={sp.prod(roots3)}")
```

**sp.prod(roots)**: <br>
=> 근들의 곱
