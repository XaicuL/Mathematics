# 복소수

```python
import sympy as sp
sp.init_printing()

z1 = 3 + 4*sp.I
z2 = 1 - 2*sp.I

print('(3 + 4i) + (1 - 2i) = ')
display(z1 + z2)

print('(3 + 4i) - (1 - 2i) = ')
display(z1 - z2)
```

**n * sp.I** (단, n = 1,2,3...etc): <br>
=>복소수를 나타내는 방법

# **복소수 동치 조건**

```python
import sympy as sp
sp.init_printing()

a = sp.symbols('a' ,real = True)
b = sp.symbols('b' ,real = True)

z1 = a + 4*sp.I
z2 = -1 + b*sp.I

eq = sp.Eq(z1, z2)
sol_eq = sp.solve(eq)

print(f'{z1} = {z2}')
display(sol_eq)
```

**sp.symbol('name', real = True)** 에서 **real = True**의 의미 <br>
=> 해당 **name** 이라는 변수는 **실수** 라는 의미

**sp.Eq()**: <br>
방정식에서의 **=(등호)** 역할이라고 보면 된다.
<br> 그러나 **Eq** 함수까지만 쓰면 단지 식을 세팅해둔 것이니, 방정식을 계산해내기 위해 **sp.solve()** 함수를 사용해야한다.

```python
#복소수의 덧셈과 뺄셈

import sympy as sp

sp.init_printing()

a = sp.symbols('a' ,real = True)
b = sp.symbols('b' ,real = True)
c = sp.symbols('c' ,real = True)
d = sp.symbols('d', real = True)

z1 = (a + b*sp.I) + (c + d*sp.I)
z2 = (a + b*sp.I) - (c + d*sp.I)

z1_a = (a+c) + (b+d)*sp.I
z2_a = (a-c) + (b-d)*sp.I

display(z1)
print('=')
display(z1_a)

print('################################################')

display(z2)
print('=')
display(z2_a)
```

```python
# 복소수의 곱셈

import sympy as sp

sp.init_printing()

a = sp.symbols('a' ,real = True)
b = sp.symbols('b' ,real = True)
c = sp.symbols('c' ,real = True)
d = sp.symbols('d', real = True)

z = (a + b*sp.I) * (c + d*sp.I)
zCal = (a*c - b*d) + (a*d + b*c)*sp.I

display(z)
print('=')
display(zCal)
```

```python
# 복소수의 나눗셈

import sympy as sp

sp.init_printing()

a = sp.symbols('a' ,real = True)
b = sp.symbols('b' ,real = True)
c = sp.symbols('c' ,real = True)
d = sp.symbols('d', real = True)

exP = c + d*sp.I
exC = a + b*sp.I

ex_pa = c**2 + d**2
ex_ca = a*c + b*d

ex_pb = c**2 + d**2
ex_cb = b*c - a*d

ex1 = exC/exP
ex2 = ex_ca/ex_pa
ex3 = ex_cb/ex_pb

display(ex1)
print('=')

display(ex2 + ex3)

```

# **켤레복소수와 절댓값**

$z = a + bi$ 일 때

- 켤레: $\bar{z} = a - bi$
- 절댓값: $|z| = \sqrt{a^2 + b^2}$
- 성질: $z\bar{z} = |z|^2$, $\dfrac{1}{z} = \dfrac{\bar{z}}{|z|^2}$ $(z \neq 0)$


```python
import sympy as sp
sp.init_printing()

a, b = sp.symbols('a b', real=True)
z = a + b*sp.I

print("z =")
display(z)
print("켤레 z̄ =")
display(sp.conjugate(z))
print("|z| =")
display(sp.Abs(z))
print("z · z̄ = |z|² ?")
display(sp.simplify(z * sp.conjugate(z)))
display(sp.simplify(sp.Abs(z)**2))
```

**sp.conjugate(z)**: <br>
=> 켤레복소수

**sp.Abs(z)**: <br>
=> 절댓값 (복소 모듈)

# **이차방정식의 허근**

판별식 $D < 0$ 이면 서로 다른 두 허근.  
(판별식 자체는 `Discriminant.ipynb`)

예: $x^2 + 2x + 5 = 0$ → $D = 4 - 20 = -16$


```python
import sympy as sp
sp.init_printing()

x = sp.symbols('x')
expr = x**2 + 2*x + 5
poly = sp.Poly(expr, x)
a_, b_, c_ = poly.all_coeffs()
D = b_**2 - 4*a_*c_

print(f"D = {D}")
print("근:")
roots = sp.solve(expr, x)
display(roots)

print("두 근은 서로 켤레인가?")
display(sp.simplify(sp.conjugate(roots[0]) - roots[1]))
```

실계수 이차방정식에서 $D<0$ 이면 **두 허근은 서로 켤레**이다.
