# 이차부등식

```python
from sympy import Symbol, solve_poly_inequality, solve_univariate_inequality, Poly, Interval
```

```python
x = Symbol('x')

ineq1 = x**2 - 4
ineq2 = x**2 - 9
```

```python
sol1 = solve_poly_inequality(Poly(ineq1, x), '>')
sol2 = solve_poly_inequality(Poly(ineq2, x), '<')
```

```python
from sympy import Union, Intersection

set1 = Union(*sol1)
set2 = Union(*sol2)
```

```python
final_solution = Intersection(set1, set2)

print(f"연립부등식의 해: {final_solution}")
```
