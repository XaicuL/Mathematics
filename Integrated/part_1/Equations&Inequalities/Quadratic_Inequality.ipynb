# 이차부등식

$ax^2 + bx + c \gtrless 0$ 의 해는 **부호표**와 판별식 $D = b^2 - 4ac$ 로 정리한다.

관련: `Discriminant.ipynb` (실근·중근), `Viète's_formulas.ipynb` (근의 합·곱).


## 1. 부호표의 아이디어

$a > 0$ 이고 서로 다른 두 실근 $\alpha < \beta$ 이면:

| 구간 | $x < \alpha$ | $\alpha < x < \beta$ | $x > \beta$ |
|------|--------------|----------------------|-------------|
| $ax^2+bx+c$ | $+$ | $-$ | $+$ |

- $a < 0$ 이면 부호가 반대
- $D = 0$ (중근 $\alpha$): $a(x-\alpha)^2 \ge 0$ 또는 $\le 0$ 형태로 단순화
- $D < 0$: 근이 없으므로 $a$ 의 부호만으로 전 구간 결정


## 2. $D$ 와 해의 관계 ($a > 0$ 기준)


```python
import sympy as sp
sp.init_printing()

x = sp.symbols('x')

cases = [
    ("D>0", x**2 - 5*x + 6, ">"),   # (x-2)(x-3)>0
    ("D=0", x**2 - 4*x + 4, ">="),  # (x-2)^2 >= 0
    ("D<0", x**2 + 2*x + 5, ">"),   # 항상 양
]

for name, expr, rel in cases:
    a, b, c = sp.Poly(expr, x).all_coeffs()
    D = b**2 - 4*a*c
    print("=" * 40)
    print(f"{name}: D={D}")
    display(sp.Rel(expr, 0, rel))
    sol = sp.solve_poly_inequality(sp.Poly(expr, x), rel)
    display(sol)
```

**sp.solve_poly_inequality(Poly, rel)**: <br>
=> 다항부등식의 해를 구간(Interval) 리스트로 반환. `rel`은 `'>'`, `'>='`, `'<'`, `'<='`

**sp.Rel(expr, 0, rel)**: <br>
=> 부등식 객체를 보기 좋게 표시


## 3. 부호표로 풀기 (수동)

$(x-2)(x-3) > 0$ → 근 $2, 3$ 을 경계로 구간을 나눈다.


```python
expr = (x - 2)*(x - 3)
expanded = sp.expand(expr)
print("전개:")
display(expanded)

# 경계점
roots = sorted(sp.solve(expr, x))
print(f"경계(근): {roots}")

# 각 구간의 대표점으로 부호 확인
tests = [0, sp.Rational(5, 2), 4]
for t in tests:
    val = expr.subs(x, t)
    print(f"x={t} → f(x)={val} ({'양' if val > 0 else '음'})")

print("해 (sympy):")
display(sp.solve_poly_inequality(sp.Poly(expanded, x), '>'))
```

## 4. 연립이차부등식

예: $x^2 - 4 > 0$ 이고 $x^2 - 9 < 0$


```python
from sympy import Union, Intersection

ineq1 = x**2 - 4
ineq2 = x**2 - 9

sol1 = sp.solve_poly_inequality(sp.Poly(ineq1, x), '>')
sol2 = sp.solve_poly_inequality(sp.Poly(ineq2, x), '<')

set1 = Union(*sol1)
set2 = Union(*sol2)
final = Intersection(set1, set2)

print("x^2 - 4 > 0:")
display(set1)
print("x^2 - 9 < 0:")
display(set2)
print("교집합(연립 해):")
display(final)
```

**Union(\*intervals)**: <br>
=> 구간들의 합집합

**Intersection(A, B)**: <br>
=> 교집합 — 연립부등식의 공통 해


## 5. 등호 포함 / 한 점으로 퇴화

$(x-1)^2 \ge 0$ 은 모든 실수, $(x-1)^2 < 0$ 은 공집합.


```python
sq = (x - 1)**2
for rel in [">=", ">", "<=", "<"]:
    sol = sp.solve_poly_inequality(sp.Poly(sp.expand(sq), x), rel)
    print(f"(x-1)^2 {rel} 0 → {sol}")
```

## 6. 그래프 관점

$y = ax^2+bx+c$ 가 $x$축 위에 있는 $x$ 구간이 곧 부등식의 해.

관련: `Quad_Func_max&min.ipynb`, `Discriminant.ipynb`.


```python
import numpy as np
import matplotlib.pyplot as plt

xs = np.linspace(-5, 5, 400)
ys = xs**2 - 4  # (x-2)(x+2)

plt.figure(figsize=(7, 4))
plt.plot(xs, ys, label=r"$y=x^2-4$")
plt.axhline(0, color="gray", lw=0.8)
plt.axvline(0, color="gray", lw=0.8)
plt.fill_between(xs, ys, 0, where=(ys > 0), alpha=0.25, label=r"$y>0$")
plt.scatter([-2, 2], [0, 0], zorder=5)
plt.legend()
plt.grid(True, alpha=0.3)
plt.title(r"$x^2-4>0$ 해: $(−\infty,-2)\cup(2,\infty)$")
plt.tight_layout()
plt.show()
```
