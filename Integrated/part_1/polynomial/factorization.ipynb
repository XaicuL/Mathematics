# 다항식의 인수분해

공통인수 묶기 → 공식 → (필요 시) 중항 추가·조립제법.

## 1. 공통인수


```python
import sympy as sp
sp.init_printing()

x, a, b, c = sp.symbols('x a b c')

expr = 6*x**3 - 9*x**2 + 3*x
print("원식 / 인수분해:")
display(expr)
display(sp.factor(expr))
```

**sp.factor(expr)**: <br>
=> 다항식 인수분해

**sp.expand(expr)**: <br>
=> 전개 (검산용)


## 2. 기본 공식

$$
(a \pm b)^2 = a^2 \pm 2ab + b^2
$$

$$
a^2 - b^2 = (a-b)(a+b)
$$

$$
a^3 \pm b^3 = (a \pm b)(a^2 \mp ab + b^2)
$$


```python
formulas = [
    ("(a+b)^2", (a + b)**2),
    ("(a-b)^2", (a - b)**2),
    ("a^2-b^2", a**2 - b**2),
    ("a^3+b^3", a**3 + b**3),
    ("a^3-b^3", a**3 - b**3),
]

for name, e in formulas:
    print("=" * 40)
    print(name)
    print("전개:")
    display(sp.expand(e))
    print("인수분해(원식으로 되돌리기):")
    display(sp.factor(sp.expand(e)))
```

## 3. $(a+b+c)^2$ / $(a \pm b)^3$


```python
print("(a+b+c)^2 전개:")
display(sp.expand((a + b + c)**2))

print("(a+b)^3:")
display(sp.expand((a + b)**3))
display(sp.factor(a**3 + 3*a**2*b + 3*a*b**2 + b**3))

print("(a-b)^3:")
display(sp.factor(a**3 - 3*a**2*b + 3*a*b**2 - b**3))
```

## 4. 중항 만들기

$x^2 + 6x + 5 = (x+3)^2 - 4 = (x+3-2)(x+3+2)$


```python
expr = x**2 + 6*x + 5
print("원식:")
display(expr)
print("완전제곱 + 차:")
mid = (x + 3)**2 - 2**2
display(mid)
print("인수분해:")
display(sp.factor(expr))
print(f"일치?: {sp.expand(mid) == expr}")
```

## 5. 조셈(치환) 인수분해

$x^4 + x^2 + 1$ → $t = x^2$ 로 두면 $t^2 + t + 1$ 은 더 이상 안 되지만,  
$x^4 + 4$ 등은 중항을 넣어 인수분해.


```python
# x^4 + 4 = (x^2+2)^2 - (2x)^2
expr = x**4 + 4
print("원식:")
display(expr)
print("중항 트릭:")
trick = (x**2 + 2)**2 - (2*x)**2
display(trick)
print("인수분해:")
display(sp.factor(expr))

# 이차식 조셈: (x^2+3x)^2 형태 예
u = x**2 + 3*x
expr2 = u**2 + 2*u - 8
print("치환 u=x^2+3x 후 인수분해:")
display(sp.factor(sp.expand(expr2)))
```

```python
practice = [
    2*x**2 + 4*x,
    x**2 - 9,
    x**3 - 8,
    x**2 + 5*x + 6,
    a**2 + 2*a*b + b**2 - c**2,
]

for e in practice:
    print("-" * 30)
    display(e)
    display(sp.factor(e))
```
