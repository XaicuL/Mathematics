2. 항등식의 핵심 성질
    - 좌변과 우변의 일치: 식을 정리하면 \(0=0\) 꼴이 되거나, 양변의 동류항 계수가 모두 같다. <br><br>

**미정계수법: 항등식 성질을 이용해 계수를 구함** <br>

        - 계수비교법: 양변의 동류항 계수를 비교.
        - 수치대입법: 문자에 편리한 숫자를 대입하여 계산

## 계수비교법

- 양변의 동류항 계수를 비교. <br>

계수비교법은 두 가지의 전제를 필요로 한다. <br>

1. 항등식이여야 한다.
2. 두 개의 식이 있어야 한다.

```python
import sympy as sp
sp.init_printing()

x = sp.symbols('x')
a = sp.symbols('a')
b = sp.symbols('b')

lhs = a*(x - 1)**2 + b * (x - 1) - 2
rhs = x**2 -2*x - 1

display(lhs)
print('위 식을 좌변이라 가정하고')
display(rhs)
print('위 식을 우변이라 가정하자')
print('우선 좌변을 정리하자.')

lhs_exp = sp.expand(lhs)
display(lhs_exp)


expr = lhs - rhs

ex = sp.Poly(expr, x)
coeffs = ex.coeffs()

ans = sp.solve(coeffs, (a, b))


display(ans)
```

sp.Poly(cal_variableName, x): <br>
=> 수식을 다항식으로 변환한다고 생각하면 된다. <br>
**,x** 는 인터프리터에게 **x에 대한 식** 이라는 점을 시사한다고 생각하면 된다. <br><br>

coeffs(): <br>
=> **Poly()** 함수로 인해 다항식으로 변환된 식에서 계수만 골라내주는 것이라 생각하면 된다. <br><br>

sp.solve(coeffsVarName, (cfs1, cfs2)): <br>
=> 방정식의 해를 찾는 기능을 해준다고 생각하면 된다. <br>
여기서 **coeffsVarName** 은 뒤에 있는 **cfs1** **cfs2** 를 모두 **= 0** 을 취한다고 생각하면 된다.

## 수치대입법
- 문자에 편리한 숫자를 대입하여 계산

```python
import sympy as sp
sp.init_printing()

x, a, b = sp.symbols('x a b')


lhs = a*(x - 1)**2 + b*(x - 1) - 2
rhs = x**2 - 2*x - 1


eq = lhs - rhs

print("--- 수치대입법 시작 ---")

eq1 = eq.subs(x, 1)
print(f"x=1 대입 시 식: {eq1} = 0")
display(eq1)


eq2 = eq.subs(x, 0)
print(f"x=0 대입 시 식: {eq2} = 0")
display(eq2)

eq3 = eq.subs(x,2)
print(f"x = 2 대입 시 식: {eq3} = 0")
display(eq3)

ans_sub = sp.solve([eq1, eq2], (a, b))

print("\n--- 최종 결과 ---")
print(f"미정계수 해: {ans_sub}")
display(ans_sub)
```

subs(cal_variableName, n): (n = 1,2,3,4,5...) <br>
=> **subs()** 함수는 Substitution(대입/교체)의 약자인데, **n** 은 대입을 시킬 수를 지칭한다 <br>

solve([eqN , eqN], (sb1,sb2))<br>
=> **solve()** 함수는 해당 식이 **= 0** 이 되게 하는 값은 무엇인지를 구하게 하는 기능을 한다.

# 나머지 정리
나머지 정리란 **다항식의 나눗셈**을 진행하지 않고, 쉽게 **몫**과 **나머지** 를 구하기 위해 사용한다고 이해하면 된다.

```python
import sympy as sp
sp.init_printing()

x = sp.symbols('x')

px = 2*x**3 + 1 * x ** 2 + 1 * x + 1
display(px)

R = px.subs(x,-1)
display(R)

quotient, remainder = sp.div(px, x + 1)

print("몫:", quotient)       # 2*x**2 - x + 2
print("나머지:", remainder) # -1
```

# 인수 정리
    - 나머지 정리의 나머지는 상수 OR 0
    - 인수 정리의 나머지는 Must 0

다항식 P(x)에 대하여
1. P(α) = 0 이면 일차식 x - α 로 나누어 떨어진다.
2. P(x)가 일차식 x - α 로 나누어떨어지면 P(α) = 0이다.
