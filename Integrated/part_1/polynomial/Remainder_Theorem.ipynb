# 나머지정리 · 인수정리

항등식 → 미정계수(계수비교 / 수치대입) → 나머지정리 $f(a)=r$ → 인수정리 → 조립제법.

## 1. 항등식

모든 $x$ 에 대해 좌변 = 우변이면 **항등식**.

미정계수법:
- **계수비교법**: 동류항 계수를 같다고 놓음
- **수치대입법**: 편리한 수를 대입해 연립


## 2. 계수비교법


```python
import sympy as sp
sp.init_printing()

x, a, b = sp.symbols('x a b')

lhs = a*(x - 1)**2 + b*(x - 1) - 2
rhs = x**2 - 2*x - 1

print("좌변 / 우변:")
display(lhs)
display(rhs)

lhs_exp = sp.expand(lhs)
print("좌변 전개:")
display(lhs_exp)

diff = sp.Poly(lhs - rhs, x)
coeffs = diff.coeffs()  # 각 차수 계수 = 0
ans = sp.solve(coeffs, (a, b))
print("계수비교 해:")
display(ans)
```

**sp.Poly(expr, x).coeffs()**: <br>
=> 최고차부터 계수 리스트 (항등식이면 모두 0)

**sp.solve(coeffs, (a, b))**: <br>
=> 계수 = 0 연립을 풀어 미정계수 결정

**sp.expand**: <br>
=> 전개 후 동류항 정리


## 3. 수치대입법


```python
lhs = a*(x - 1)**2 + b*(x - 1) - 2
rhs = x**2 - 2*x - 1
eq = sp.expand(lhs - rhs)

# x=1 이면 (x-1) 항이 사라져 단순해짐 — 다만 항등식이므로 0=0이 될 수 있음
eq0 = eq.subs(x, 0)
eq2 = eq.subs(x, 2)
eq3 = eq.subs(x, 3)

print("대입 식 = 0:")
display(eq0, eq2, eq3)

ans_sub = sp.solve([eq0, eq2], (a, b))
print("수치대입 해:")
display(ans_sub)

# 검산
print("다른 점으로 검산 eq3 with 해:")
display(eq3.subs(ans_sub))
```

**expr.subs(x, n)**: <br>
=> $x$ 에 수 $n$ 대입

## 4. 나머지정리

$f(x)$ 를 $x - \alpha$ 로 나눈 나머지가 $r$ 이면:

$$
f(x) = (x - \alpha)\,q(x) + r \quad\Rightarrow\quad f(\alpha) = r
$$

나눗셈을 하지 않고 $f(\alpha)$ 만으로 나머지를 알 수 있다.


```python
f = 2*x**3 + x**2 + x + 1
alpha = -1  # 제수 x - (-1) = x+1

r = f.subs(x, alpha)
q, rem = sp.div(f, x - alpha)

print("f(x):")
display(f)
print(f"f({alpha}) = {r}  ← 나머지정리")
print(f"sp.div 몫: {q}, 나머지: {rem}")
print(f"일치?: {r == rem}")
```

**f.subs(x, alpha)**: <br>
=> 나머지정리로 나머지 계산

**sp.div(f, x - alpha)**: <br>
=> 다항식 나눗셈 (몫, 나머지) — 검산용


## 5. 인수정리

$$
f(\alpha) = 0 \;\Longleftrightarrow\; (x - \alpha) \text{ 가 } f(x) \text{ 의 인수}
$$


```python
f = x**3 - 6*x**2 + 11*x - 6
candidates = [1, 2, 3, -1]

print("인수정리 후보 검사:")
for t in candidates:
    val = f.subs(x, t)
    print(f"f({t}) = {val}", "→ 인수" if val == 0 else "")

print("인수분해:")
display(sp.factor(f))
```

## 6. 조립제법 예

$f(x) = 2x^3 + x^2 + x + 1$ 을 $x+1$ 로 나눔 (합성나눗셈).


```python
# 계수: 2, 1, 1, 1  /  뿌리 -1
coeffs = [2, 1, 1, 1]
root = -1

synth = [coeffs[0]]
for c in coeffs[1:]:
    synth.append(root * synth[-1] + c)

print("조립제법 결과 계수(몫) + 나머지:")
print(f"몫 계수: {synth[:-1]}, 나머지: {synth[-1]}")

q, rem = sp.div(2*x**3 + x**2 + x + 1, x + 1)
print("검산:")
display(q, rem)
```

조립제법: 맨 앞 계수를 내려쓰고, 「×뿌리 + 다음계수」를 반복. 마지막 수가 나머지.
