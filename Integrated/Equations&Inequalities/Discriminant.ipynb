# 판별식

```python
import sympy as sp
sp.init_printing()

x = sp.symbols('x')
expr = x**2 + 3*x - 6

poly = sp.Poly(expr, x)
coeffs = poly.all_coeffs()

if len(coeffs) == 3:
    a, b, c = coeffs
elif len(coeffs) == 2:
    a, b, c = 0, coeffs[0], coeffs[1]
else:
    a, b, c = 0, 0, coeffs[0]

def dcm(a, b, c):
    return b**2 - 4*a*c

re = dcm(a, b, c)

print("식:")
display(expr)
print(f"계수: a={a}, b={b}, c={c}")
print(f"판별식: {re}")

if re > 0:
    print("서로 다른 두 실근")
elif re == 0:
    print("중근")
else:
    print("서로 다른 두 허근")
```
