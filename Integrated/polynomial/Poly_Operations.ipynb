# **다항식의 계산**

- 다항식의 덧셈: 동류항끼리 모아서 정리한다.
- 다항식의 뺄셈: 각 항의 부호를 바꾸어 더한다
- 다항식의 곱셈: 분배법칙을 이용하여 식을 전개한 다음 동류항끼리 모아서 정리한다.
- 다항식의 나눗셈: 다항식을 내림차순으로 정리한 다음 계산한다.


## **다항식의 덧셈**

- 동류항끼리 모아서 정리한다.

### 성질
    - 교환법칙: A+B = B+Α
    - 결합법칙: (A+B)+C = A+(B+C)

```python
import sympy as sp
sp.init_printing()

x = sp.symbols('x')

ex1 = x**3 + 2*x + 3
ex2 = 2*x**3 -3*x**2 + 1

cal = ex1 + ex2

display(cal)

######################################################
print('-----------------------성질----------------------------')

print('###교환법칙###')

cal_a = ex1 + ex2
cal_b = ex2 + ex1

display(cal_a)
display(cal_b)

if cal_a == cal_b:
    print('교환법칙 성립')
else:
    print('교환법칙 성립X')

print('###결합법칙###')

ex3 = x**3 + 4*x + 2

cal_c = (ex1 + ex2) + ex3
cal_d = ex1 + (ex2 + ex3)

display(cal_c)
display(cal_d)

if cal_c == cal_d:
    print('결합법칙 성립')
else:
    print('결합법칙 성립X')
```

## 다항식의 뺄셈

```python
import sympy as sp
sp.init_printing()

x = sp.symbols('x')

ex1 = x**2 - 1
ex2 = x**2 + x
ex3 = x+3

cal_expanded = sp.expand(ex1 * ex2 * ex3)
display(cal_expanded)
######################################################
print('-----------------------성질----------------------------')

print('###교환법칙###')

cal_A = ex1 * ex2
cal_A_ep = sp.expand(cal_A)

display(cal_A)
display(cal_A_ep)

cal_B = ex2 * ex1
cal_B_ep = sp.expand(cal_B)

display(cal_B)
display(cal_B_ep)

print('-----------------검증-----------------')

if cal_A_ep == cal_B_ep:
    print('교환법칙 성립')
else:
    print('교환법칙 성립X')

######################################################

print('###결합법칙###')

cal_C = (ex1 * ex2) * ex3
cal_C_ep = sp.expand(cal_C)

display(cal_C)
display(cal_C_ep)

cal_D = ex1* (ex2 * ex3)
cal_D_ep = sp.expand(cal_D)

display(cal_D)
display(cal_D_ep)


print('-----------------검증2-----------------')

if cal_C_ep == cal_D_ep:
    print('결합법칙 성립')
else:
    print('결합법칙 성립X')

######################################################

print('###분배법칙(1)###')

cal_E = ex1 * (ex2 + ex3)
cal_E_ep = sp.expand(cal_E)

display(cal_E)
display(cal_E_ep)

cal_F = (ex1 * ex2) + (ex1 * ex3)
cal_F_ep = sp.expand(cal_F)

display(cal_F)
display(cal_F_ep)

print('-----------------검증3-----------------')

if cal_E_ep == cal_F_ep:
    print('분배법칙 성립(1)')
else:
    print('분배법칙 성립X (1)')

######################################################

print('###분배법칙(2)###')

cal_G = (ex1 + ex2) * ex3
cal_G_ep = sp.expand(cal_G)

display(cal_G)
display(cal_G_ep)


cal_H = (ex1 * ex3) + (ex2 * ex3)
cal_H_ep = sp.expand(cal_H)

display(cal_H)
display(cal_H_ep)

print('-----------------검증3-----------------')

if cal_G_ep == cal_H_ep:
    print('분배법칙 성립(2)')
else:
    print('분배법칙 성립X (2)')


# cf
# 하나하나 다 치다가 진짜 미칠 뻔 했다. 수학자 분들에게 존경을 표한다
#그런데, 여기서 하나 보여주고픈게 있다. 우리는 def와 친해져야 할 이유가 있다. 다음의 코드를 보자

def verify_all_polynomial_laws(A, B, C):

    sp.init_printing()

    # 1. 교환법칙
    is_commutable = sp.expand(A * B) == sp.expand(B * A)

    # 2. 결합법칙
    is_associative = sp.expand((A * B) * C) == sp.expand(A * (B * C))

    # 3. 분배법칙
    dist_left = sp.expand(A * (B + C)) == sp.expand(A * B + A * C)
    dist_right = sp.expand((A + B) * C) == sp.expand(A * C + B * C)

    # --- 결과 출력 섹션 ---
    print("\n" + "="*30)
    print('-----------------검증-----------------')
    print(f"1. 교환법칙: {'✅ 통과' if is_commutable else '❌ 실패'}")
    print(f"2. 결합법칙: {'✅ 통과' if is_associative else '❌ 실패'}")
    print(f"3. 분배법칙(좌): {'✅ 통과' if dist_left else '❌ 실패'}")
    print(f"4. 분배법칙(우): {'✅ 통과' if dist_right else '❌ 실패'}")
    print("="*30)


    if all([is_commutable, is_associative, dist_left, dist_right]):
        print('최종 결과 A*B*C')
        display(sp.expand(A * B * C))
    else:
        print("\n오류 발생")


verify_all_polynomial_laws(ex1, ex2, ex3)
#hahahahahahahahahahahahahahahahahahaaaaaaaaaaaaa 내 10분은 무너졌어.
```

```python
#중학Ver

import sympy as sp
sp.init_printing()

a = sp.symbols('a')
b = sp.symbols('b')
c = sp.symbols('c')
d = sp.symbols('d')
x = sp.symbols('x')


ex1_a = (a + b)**2
ex1_b = (a - b)**2
ex2 = (a + b) * (a - b)
ex3 = (x + a) * (x + b)
ex4 = (a*x + b) * (c*x + d)

ex1_a_ep = sp.expand(ex1_a)
ex1_b_ep = sp.expand(ex1_b)
ex2_ep = sp.expand(ex2)
ex3_ep = sp.expand(ex3)
ex4_ep = sp.expand(ex4)

display(ex1_a)
print('해당 식을 계산하면 다음과 같다.')
display(ex1_a_ep)

print('-----------------------------------------------------')

display(ex1_b)
print('해당 식을 계산하면 다음과 같다.')
display(ex1_b_ep)

print('-----------------------------------------------------')

display(ex2)
print('해당 식을 계산하면 다음과 같다.')
display(ex2_ep)

print('-----------------------------------------------------')

display(ex3)
print('해당 식을 계산하면 다음과 같다.')
display(ex3_ep)

print('-----------------------------------------------------')

display(ex4)
print('해당 식을 계산하면 다음과 같다.')
display(ex4_ep)

print('-----------------검증-----------------')


def verify_polynomial_identity(name, expr):
    # 1. 전개 (Forward: Expansion)
    expanded = sp.expand(expr)

    # 2. 인수분해 (Backward: Factorization)
    factored = sp.factor(expanded)

    # 3. 검증 (Verification)
    # 원래 식과 인수분해된 식이 수학적으로 같은지 확인 (A - B = 0)
    is_valid = sp.simplify(expr - factored) == 0

    print(f"--- [{name}] 검증 리포트 ---")
    print(f"1. 원본 수식 (Origin):")
    display(expr)
    print(f"2. 전개 결과 (Expanded):")
    display(expanded)
    print(f"3. 복원 결과 (Factored):")
    display(factored)

    if is_valid:
        print(f"✅\n")
    else:
        print(f"❌\n")

# --- 실행부 ---
sp.init_printing()
a, b, c, d, x = sp.symbols('a b c d x')

# 검증할 식들의 리스트 (Dictionary 활용)
identities = {
    "완전제곱식(합)": (a + b)**2,
    "완전제곱식(차)": (a - b)**2,
    "합차공식": (a + b) * (a - b),
    "일차식의 곱(1)": (x + a) * (x + b),
    "일차식의 곱(2)": (a*x + b) * (c*x + d)
}

for name, expression in identities.items():
    verify_polynomial_identity(name, expression)
```

## 다항식의 나눗셈
- 각 다항식을 내림차순으로 정리한 다음 자연수의 나눗셈과 같은 방식으로 계산한다

```python
import sympy as sp
sp.init_printing()

ex1 = x**2 + 2*x + 3
ex2 = x + 1
q1, r1 = sp.div(ex1, ex2)

display(q1)
display(r1)

print('---------------------------------여기부턴 divide function 출력--------------------------------------')

#이걸로 함수를 정의한다면 다음과 같이 제작할 수 있다.

def divide(A,B):
    q, r = sp.div(A, B)
    display(q)
    display(r)

#cf, 여기서 q는 몫, r은 나머지라는 뜻

divide(x**2 + 2*x + 3, x + 1)
```
